```python
# ============================================================
# 강판(Steel Plates) 표면 결함 유형 분류
# ============================================================
# 데이터: steel_plates_faults.csv (1,941행 × 34열)
#   - 입력 변수 27개: 결함 위치·크기·둘레·밝기·강종·두께·형상 지수
#   - 타깃 7개(원-핫): Pastry, Z_Scratch, K_Scatch, Stains,
#                     Dirtiness, Bumps, Other_Faults
#
# 예시(SNP) 코드와 달라진 점
#   1. 범주형 SNP → 연속형 수치 변수이므로 원-핫 인코딩 대신
#      로그 변환 + 표준화(선형 모델용) / 원본 그대로(트리 모델용)
#   2. 카이제곱 검정 → Kruskal-Wallis 검정 + 효과크기 + 상호정보량
#   3. 모델: LightGBM(주력 추천) / Random Forest / 로지스틱 회귀(기준선)
#   4. GridSearchCV → RandomizedSearchCV (탐색 비용 절감)
#   5. 변수 중요도: 모델 내장 중요도 + Permutation Importance
#   6. Hamming 거리 유사도 분석은 연속형 데이터에 맞지 않아 제외
#   7. 외부 test.csv가 없으므로 최종 모델을 저장하고
#      새 데이터 예측 함수를 제공
# ============================================================


# ============================================================
# 0. 라이브러리
# ============================================================

import time
import warnings

import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

try:
    from IPython.display import display
except ImportError:          # 주피터가 아닌 환경에서도 실행되도록
    display = print

from scipy.stats import kruskal, loguniform, randint, uniform

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_validate,
    RandomizedSearchCV,
)
from sklearn.preprocessing import StandardScaler, FunctionTransformer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.decomposition import PCA
from sklearn.feature_selection import mutual_info_classif
from sklearn.inspection import permutation_importance
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
)

from lightgbm import LGBMClassifier

warnings.filterwarnings("ignore")


# ============================================================
# 1. 기본 설정
# ============================================================

plt.rcParams["font.family"] = "Malgun Gothic"   # macOS: "AppleGothic"
plt.rcParams["axes.unicode_minus"] = False

RANDOM_STATE = 42
DATA_PATH = "steel_plates_faults.csv"

FAULT_COLS = [
    "Pastry",
    "Z_Scratch",
    "K_Scatch",
    "Stains",
    "Dirtiness",
    "Bumps",
    "Other_Faults",
]

class_mapping = {name: i for i, name in enumerate(FAULT_COLS)}
reverse_class_mapping = {i: name for name, i in class_mapping.items()}
LABELS = list(range(len(FAULT_COLS)))


# ============================================================
# 2. 데이터 불러오기
# ============================================================

raw_df = pd.read_csv(DATA_PATH)

print("데이터 크기:", raw_df.shape)
display(raw_df.head())


# ============================================================
# 3. 데이터 기본 확인
# ============================================================

print("\n[데이터 정보]")
raw_df.info()

print("\n[결측치 합계]:", raw_df.isnull().sum().sum())
print("[중복 행 수]:", raw_df.duplicated().sum())

# 타깃이 행마다 정확히 1개만 1인지 확인 (원-핫 타깃 검증)
fault_sum = raw_df[FAULT_COLS].sum(axis=1)
print("\n[행별 결함 표시 개수]")
display(fault_sum.value_counts())
assert (fault_sum == 1).all(), "결함 라벨이 1개가 아닌 행이 있습니다."

# 강종 변수 확인: A300 + A400 = 1 이면 둘 중 하나는 중복 정보
steel_sum = raw_df["TypeOfSteel_A300"] + raw_df["TypeOfSteel_A400"]
print("\n[A300 + A400 값 분포]")
display(steel_sum.value_counts())


# ============================================================
# 4. 타깃(Class) 생성 및 분포
# ============================================================
# 7개의 원-핫 결함 열을 하나의 범주형 Class로 합친다.
# ============================================================

df = raw_df.copy()
df["class"] = df[FAULT_COLS].idxmax(axis=1)

class_counts = df["class"].value_counts().reindex(FAULT_COLS)
class_distribution = pd.DataFrame({
    "count": class_counts,
    "ratio(%)": (class_counts / len(df) * 100).round(2),
})

print("\n[Class 분포]")
display(class_distribution)

print(
    "\n최대/최소 Class 비율:",
    round(class_counts.max() / class_counts.min(), 1), "배",
    "→ 불균형 데이터이므로 Macro F1을 주 평가지표로 사용",
)

plt.figure(figsize=(8, 4))
plt.bar(class_distribution.index, class_distribution["count"])
plt.xlabel("결함 유형")
plt.ylabel("개수")
plt.title("결함 유형별 개수")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()


# ============================================================
# 5. 파생 변수 생성
# ============================================================
# - TypeOfSteel_A400: A300과 완전히 반대(합=1)이므로 제거
# - 결함 영역의 가로/세로 길이, 가로세로비
# - 밝기 범위, 평균 밝기
# - 면적 대비 둘레(형상이 얼마나 가늘고 복잡한지)
# ============================================================

def add_features(data: pd.DataFrame) -> pd.DataFrame:

    out = data.copy()

    out["X_Range"] = out["X_Maximum"] - out["X_Minimum"]
    out["Y_Range"] = out["Y_Maximum"] - out["Y_Minimum"]
    out["Aspect_Ratio"] = (out["X_Range"] + 1) / (out["Y_Range"] + 1)

    out["Luminosity_Range"] = (
        out["Maximum_of_Luminosity"] - out["Minimum_of_Luminosity"]
    )
    out["Mean_Luminosity"] = (
        out["Sum_of_Luminosity"] / out["Pixels_Areas"]
    )

    out["Perimeter_Area_Ratio"] = (
        (out["X_Perimeter"] + out["Y_Perimeter"]) / out["Pixels_Areas"]
    )

    out = out.drop(columns=["TypeOfSteel_A400"], errors="ignore")

    return out


feature_df = add_features(df.drop(columns=FAULT_COLS + ["class"]))
feature_cols = feature_df.columns.tolist()

print("\n입력 변수 수:", len(feature_cols))
print(feature_cols)

# 로그 변환할 치우친(양수) 변수 → 선형 모델에만 적용
SKEWED_COLS = [
    "Pixels_Areas",
    "X_Perimeter",
    "Y_Perimeter",
    "Sum_of_Luminosity",
    "X_Range",
    "Y_Range",
    "Aspect_Ratio",
    "Perimeter_Area_Ratio",
]

print("\n[치우친 변수의 왜도]")
display(feature_df[SKEWED_COLS].skew().round(2).to_frame("skewness"))


# ============================================================
# 6. 변수 간 상관관계
# ============================================================
# 크기 관련 변수(면적, 둘레, 밝기 합 등)는 서로 강하게 상관된다.
# 트리 모델은 영향이 적지만, 중요도 해석 시 주의해야 한다.
# ============================================================

corr = feature_df.corr(method="spearman")

plt.figure(figsize=(12, 10))
plt.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar(label="Spearman 상관계수")
plt.xticks(range(len(feature_cols)), feature_cols, rotation=90, fontsize=8)
plt.yticks(range(len(feature_cols)), feature_cols, fontsize=8)
plt.title("변수 간 Spearman 상관관계")
plt.tight_layout()
plt.show()

high_corr_pairs = (
    corr.where(np.triu(np.ones(corr.shape, dtype=bool), k=1))
    .stack()
    .rename("rho")
    .reset_index()
    .query("abs(rho) >= 0.9")
    .sort_values("rho", key=abs, ascending=False)
)

print("\n[|상관계수| ≥ 0.9 변수 쌍]")
display(high_corr_pairs.round(3))


# ============================================================
# 7. 변수-Class 연관성 검정 (Kruskal-Wallis)
# ============================================================
#
# H0: 모든 결함 유형에서 해당 변수의 분포가 같다.
# H1: 적어도 한 결함 유형의 분포가 다르다.
#
# 변수 대부분이 정규분포가 아니므로 ANOVA 대신 비모수 검정 사용.
# 표본이 커서 p-value는 대부분 매우 작으므로,
# 효과크기 ε²(epsilon squared)로 실제 차이의 크기를 비교한다.
# 상호정보량(Mutual Information)도 함께 계산한다.
# ============================================================

y_all = df["class"].map(class_mapping)
n_obs, n_class = len(df), len(FAULT_COLS)

stat_results = []

for col in feature_cols:

    groups = [
        feature_df.loc[df["class"] == c, col] for c in FAULT_COLS
    ]
    h_stat, p_value = kruskal(*groups)

    epsilon_sq = max(0.0, (h_stat - n_class + 1) / (n_obs - n_class))

    stat_results.append({
        "Feature": col,
        "H": h_stat,
        "p-value": p_value,
        "Epsilon²": epsilon_sq,
    })

stat_df = pd.DataFrame(stat_results)

stat_df["Mutual Info"] = mutual_info_classif(
    feature_df, y_all, random_state=RANDOM_STATE
)

stat_df = stat_df.sort_values("Epsilon²", ascending=False).reset_index(drop=True)

stat_df_display = stat_df.copy()
stat_df_display["p-value"] = stat_df_display["p-value"].apply(
    lambda x: "<0.001" if x < 0.001 else f"{x:.4f}"
)

print("\n[변수-Class Kruskal-Wallis 검정 결과]")
display(stat_df_display.round(4))

fig, axes = plt.subplots(1, 2, figsize=(15, 6))

axes[0].barh(stat_df["Feature"][::-1], stat_df["Epsilon²"][::-1])
axes[0].set_title("Kruskal-Wallis 효과크기 (ε²)")
axes[0].set_xlabel("ε²")

mi_sorted = stat_df.sort_values("Mutual Info")
axes[1].barh(mi_sorted["Feature"], mi_sorted["Mutual Info"])
axes[1].set_title("상호정보량 (Mutual Information)")
axes[1].set_xlabel("MI")

plt.tight_layout()
plt.show()


# ============================================================
# 8. 상위 변수의 Class별 분포
# ============================================================

top_features = stat_df["Feature"].head(6).tolist()

fig, axes = plt.subplots(2, 3, figsize=(16, 9))

for ax, col in zip(axes.flat, top_features):

    data_by_class = [
        feature_df.loc[df["class"] == c, col] for c in FAULT_COLS
    ]

    ax.boxplot(data_by_class, showfliers=False)
    ax.set_xticks(range(1, len(FAULT_COLS) + 1))
    ax.set_xticklabels(FAULT_COLS, rotation=40)
    ax.set_title(col)

plt.suptitle("효과크기 상위 6개 변수의 결함 유형별 분포 (이상치 숨김)")
plt.tight_layout()
plt.show()


# ============================================================
# 9. 분석 변수 설정 및 데이터 분할
# ============================================================

X = feature_df.copy()
y = y_all.copy()

X_train, X_eval, y_train, y_eval = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=RANDOM_STATE,
)

print("\n[데이터 분할]")
print("학습 데이터:", X_train.shape)
print("평가 데이터:", X_eval.shape)

split_ratio = pd.DataFrame({
    "Train(%)": y_train.value_counts(normalize=True).sort_index() * 100,
    "Eval(%)": y_eval.value_counts(normalize=True).sort_index() * 100,
}).rename(index=reverse_class_mapping)

print("\n[Class 비율 비교]")
display(split_ratio.round(2))


# ============================================================
# 10. 교차검증 및 평가 지표 설정
# ============================================================

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

scoring = {
    "accuracy": "accuracy",
    "balanced_accuracy": "balanced_accuracy",
    "precision_macro": "precision_macro",
    "recall_macro": "recall_macro",
    "f1_macro": "f1_macro",
}


# ============================================================
# 11. 기본 모델 정의 (개선 전)
# ============================================================
#
# ▶ LightGBM (주력 추천 모델)
#   - 연속형 수치 변수가 많은 표 형식 데이터에서 성능이 좋음
#   - 스케일링/로그 변환 불필요, 학습 속도가 빠름
#   - class_weight="balanced"로 소수 Class(Dirtiness, Stains) 보정
#
# ▶ Random Forest (비교 모델)
#   - 튜닝에 덜 민감한 안정적인 비교 대상
#
# ▶ 로지스틱 회귀 (기준선)
#   - 로그 변환 + 표준화 후 학습
#   - 단순한 선형 모델 대비 트리 모델이 얼마나 나은지 확인
# ============================================================

def make_logistic_pipeline():

    other_cols = [c for c in feature_cols if c not in SKEWED_COLS]

    preprocessor = ColumnTransformer([
        ("log", FunctionTransformer(np.log1p, feature_names_out="one-to-one"),
         SKEWED_COLS),
        ("pass", "passthrough", other_cols),
    ])

    return Pipeline([
        ("preprocess", preprocessor),
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(
            C=1.0,
            class_weight="balanced",
            max_iter=5000,
            random_state=RANDOM_STATE,
        )),
    ])


lgbm_model = LGBMClassifier(
    objective="multiclass",
    n_estimators=300,
    learning_rate=0.05,
    num_leaves=31,
    class_weight="balanced",
    random_state=RANDOM_STATE,
    n_jobs=1,          # 교차검증/탐색 단계에서 병렬 처리하므로 1로 고정
    verbose=-1,
)

rf_model = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced_subsample",
    random_state=RANDOM_STATE,
    n_jobs=1,
)

models = {
    "LightGBM": lgbm_model,
    "Random Forest": rf_model,
    "Logistic Regression": make_logistic_pipeline(),
}


# ============================================================
# 12. 교차검증 결과 정리 함수
# ============================================================

def run_cv(model_dict, title):

    rows = []

    print("\n" + "=" * 60)
    print(title)
    print("=" * 60)

    for name, model in model_dict.items():

        print(f"{name} CV 진행 중...")
        start = time.time()

        result = cross_validate(
            model, X_train, y_train,
            cv=cv, scoring=scoring, n_jobs=-1,
        )

        rows.append({
            "Model": name,
            "CV Accuracy": result["test_accuracy"].mean(),
            "CV Balanced Acc": result["test_balanced_accuracy"].mean(),
            "CV Precision": result["test_precision_macro"].mean(),
            "CV Recall": result["test_recall_macro"].mean(),
            "CV Macro F1": result["test_f1_macro"].mean(),
            "CV Macro F1 SD": result["test_f1_macro"].std(),
            "CV Time(s)": time.time() - start,
        })

    result_df = (
        pd.DataFrame(rows)
        .sort_values("CV Macro F1", ascending=False)
        .reset_index(drop=True)
    )

    display(result_df.round(4))

    return result_df


# ============================================================
# 13. 개선 전 5-Fold 교차검증
# ============================================================

before_cv_results_df = run_cv(models, "개선 전 5-Fold 교차검증")


# ============================================================
# 14. 하이퍼파라미터 탐색 (RandomizedSearchCV)
# ============================================================
#
# 평가 데이터 X_eval / y_eval은 사용하지 않는다.
# 모든 탐색은 X_train / y_train과 5-Fold CV로 수행한다.
# 연속 범위에서 무작위 표본을 뽑으므로
# 같은 시간에 GridSearch보다 넓은 공간을 탐색할 수 있다.
# ============================================================

search_spaces = {

    "LightGBM": {
        "n_estimators": randint(200, 1000),
        "learning_rate": loguniform(0.01, 0.2),
        "num_leaves": randint(8, 64),
        "min_child_samples": randint(5, 40),
        "subsample": uniform(0.6, 0.4),
        "subsample_freq": [1],
        "colsample_bytree": uniform(0.5, 0.5),
        "reg_lambda": loguniform(1e-3, 10),
    },

    "Random Forest": {
        "n_estimators": randint(200, 800),
        "max_depth": [None, 10, 15, 20, 30],
        "min_samples_split": randint(2, 11),
        "min_samples_leaf": randint(1, 5),
        "max_features": ["sqrt", "log2", 0.3, 0.5],
    },

    "Logistic Regression": {
        "model__C": loguniform(1e-2, 1e2),
    },
}

n_iter_map = {
    "LightGBM": 40,
    "Random Forest": 30,
    "Logistic Regression": 20,
}

searches = {}

for name, model in models.items():

    print(f"\n{name} 하이퍼파라미터 탐색 중...")
    start = time.time()

    search = RandomizedSearchCV(
        estimator=model,
        param_distributions=search_spaces[name],
        n_iter=n_iter_map[name],
        scoring="f1_macro",
        cv=cv,
        n_jobs=-1,
        random_state=RANDOM_STATE,
        refit=True,           # 최적 조합으로 X_train 전체에 재학습
    )
    search.fit(X_train, y_train)
    searches[name] = search

    print(f"  소요 시간: {time.time() - start:.1f}초")
    print(f"  최적 Macro F1: {search.best_score_:.4f}")
    print(f"  최적 파라미터: {search.best_params_}")


# ============================================================
# 15. 개선 후 5-Fold 교차검증
# ============================================================
# best_score_는 탐색 과정의 최댓값이라 약간 낙관적이므로,
# 개선 전과 동일한 방식(cross_validate)으로 다시 측정한다.
# ============================================================

improved_models = {
    name: search.best_estimator_ for name, search in searches.items()
}

after_cv_results_df = run_cv(improved_models, "개선 후 5-Fold 교차검증")


# ============================================================
# 16. 개선 전후 비교
# ============================================================

compare_cols = ["Model", "CV Accuracy", "CV Balanced Acc", "CV Macro F1"]

before_after_cv = pd.merge(
    before_cv_results_df[compare_cols],
    after_cv_results_df[compare_cols],
    on="Model",
    suffixes=("_Before", "_After"),
)
before_after_cv["Macro F1 변화"] = (
    before_after_cv["CV Macro F1_After"]
    - before_after_cv["CV Macro F1_Before"]
)

print("\n[개선 전후 5-Fold CV 비교]")
display(before_after_cv.round(4))

x_pos = np.arange(len(before_after_cv))
width = 0.35

plt.figure(figsize=(9, 5))
plt.bar(x_pos - width / 2, before_after_cv["CV Macro F1_Before"], width, label="개선 전")
plt.bar(x_pos + width / 2, before_after_cv["CV Macro F1_After"], width, label="개선 후")
plt.xticks(x_pos, before_after_cv["Model"])
plt.ylabel("교차검증 Macro F1")
plt.title("모델별 하이퍼파라미터 개선 전후 Macro F1")
plt.ylim(
    max(0, before_after_cv[["CV Macro F1_Before", "CV Macro F1_After"]].min().min() - 0.1),
    1.0,
)
plt.legend()
plt.tight_layout()
plt.show()


# ============================================================
# 17. 최종 모델 선택 (개선 후 CV Macro F1 기준)
# ============================================================

best_model_name = after_cv_results_df.iloc[0]["Model"]
best_model = improved_models[best_model_name]

print("\n" + "=" * 60)
print("최종 모델 선정")
print("=" * 60)
print("선택 모델:", best_model_name)
print("CV Macro F1:", round(after_cv_results_df.iloc[0]["CV Macro F1"], 4))


# ============================================================
# 18. 학습 / 평가 데이터 성능 (모든 모델)
# ============================================================
# RandomizedSearchCV(refit=True)가 이미 X_train으로 학습을 마쳤으므로
# 다시 fit하지 않고 바로 예측한다.
#
# 주의: 모델 선택은 17단계에서 CV로 끝났다.
#       평가 데이터 결과는 일반화 성능 확인 및 참고용이다.
# ============================================================

def score_row(y_true, y_hat):
    return {
        "Accuracy": accuracy_score(y_true, y_hat),
        "Balanced Acc": balanced_accuracy_score(y_true, y_hat),
        "Precision": precision_score(y_true, y_hat, average="macro", zero_division=0),
        "Recall": recall_score(y_true, y_hat, average="macro", zero_division=0),
        "Macro F1": f1_score(y_true, y_hat, average="macro", zero_division=0),
    }


train_rows, holdout_rows = [], []

for name, model in improved_models.items():

    train_rows.append({"Model": name, **score_row(y_train, model.predict(X_train))})
    holdout_rows.append({"Model": name, **score_row(y_eval, model.predict(X_eval))})

train_results_df = pd.DataFrame(train_rows)
holdout_results_df = pd.DataFrame(holdout_rows)

overfit_check = pd.DataFrame({
    "Model": train_results_df["Model"],
    "Train Macro F1": train_results_df["Macro F1"],
    "Eval Macro F1": holdout_results_df["Macro F1"],
})
overfit_check["Gap"] = overfit_check["Train Macro F1"] - overfit_check["Eval Macro F1"]

print("\n[평가 데이터 모델 비교]")
display(holdout_results_df.round(4))

print("\n[과적합 점검: 학습 vs 평가 Macro F1]")
display(overfit_check.round(4))


# ============================================================
# 19. 최종 모델 평가 지표
# ============================================================

y_pred = best_model.predict(X_eval)
final_scores = score_row(y_eval, y_pred)

print("\n" + "=" * 60)
print(f"최종 평가 결과 - {best_model_name}")
print("=" * 60)
for metric, value in final_scores.items():
    print(f"{metric}: {value:.4f}")


# ============================================================
# 20. Classification Report / Class별 성능
# ============================================================

print("\n[Classification Report]")
print(classification_report(
    y_eval, y_pred,
    labels=LABELS, target_names=FAULT_COLS,
    digits=4, zero_division=0,
))

report_dict = classification_report(
    y_eval, y_pred,
    labels=LABELS, target_names=FAULT_COLS,
    output_dict=True, zero_division=0,
)

class_report_df = (
    pd.DataFrame(report_dict).T
    .loc[FAULT_COLS, ["precision", "recall", "f1-score", "support"]]
    .sort_values("f1-score")
)

print("\n[Class별 성능 (F1 낮은 순)]")
display(class_report_df.round(4))


# ============================================================
# 21. 혼동행렬 (개수 / 행 기준 비율)
# ============================================================

cm = confusion_matrix(y_eval, y_pred, labels=LABELS)
cm_norm = confusion_matrix(y_eval, y_pred, labels=LABELS, normalize="true")

fig, axes = plt.subplots(1, 2, figsize=(18, 7))

ConfusionMatrixDisplay(cm, display_labels=FAULT_COLS).plot(
    ax=axes[0], xticks_rotation=45, colorbar=False
)
axes[0].set_title(f"혼동행렬 (개수) - {best_model_name}")

ConfusionMatrixDisplay(cm_norm, display_labels=FAULT_COLS).plot(
    ax=axes[1], xticks_rotation=45, colorbar=False, values_format=".2f"
)
axes[1].set_title("혼동행렬 (실제 Class 기준 재현율)")

for ax in axes:
    ax.set_xlabel("예측 결함")
    ax.set_ylabel("실제 결함")

plt.tight_layout()
plt.show()


# ============================================================
# 22. 오분류 분석
# ============================================================

result_df = pd.DataFrame({
    "Actual": y_eval.map(reverse_class_mapping),
    "Predicted": pd.Series(y_pred, index=y_eval.index).map(reverse_class_mapping),
})

misclassified = result_df[result_df["Actual"] != result_df["Predicted"]]

print("\n" + "=" * 60)
print("최종 모델 오분류 분석")
print("=" * 60)
print("평가 데이터:", len(y_eval), "개")
print("오분류:", len(misclassified), "개")
print("오분류 비율:", round(len(misclassified) / len(y_eval) * 100, 2), "%")

misclassification_summary = (
    misclassified
    .groupby(["Actual", "Predicted"])
    .size()
    .reset_index(name="Count")
    .sort_values("Count", ascending=False)
    .reset_index(drop=True)
)

print("\n[주요 오분류 조합]")
display(misclassification_summary.head(10))

# Other_Faults가 얽힌 오분류 비중
# (Other_Faults는 '기타' 범주라 다른 결함과 경계가 모호할 수 있음)
other_related = misclassified.query(
    "Actual == 'Other_Faults' or Predicted == 'Other_Faults'"
)
print(
    "\nOther_Faults가 관련된 오분류 비중:",
    round(len(other_related) / max(len(misclassified), 1) * 100, 1), "%",
)

misclassified_rows = raw_df.loc[misclassified.index].copy()
misclassified_rows["Actual_Class"] = misclassified["Actual"]
misclassified_rows["Predicted_Class"] = misclassified["Predicted"]

if hasattr(best_model, "predict_proba"):
    proba = best_model.predict_proba(X_eval.loc[misclassified.index])
    misclassified_rows["Pred_Confidence"] = proba.max(axis=1)

print("\n[오분류 데이터 행]")
display(misclassified_rows.head(20))


# ============================================================
# 23. 변수 중요도
# ============================================================
# (1) 모델 내장 중요도: 트리 모델만 제공, 학습 데이터 기준
# (2) Permutation Importance: 모든 모델 가능, 평가 데이터 기준
#     → 변수 값을 섞었을 때 Macro F1이 얼마나 떨어지는지 측정
#
# 상관이 높은 변수끼리는 중요도가 나눠지거나 과소평가될 수 있다.
# (6단계 상관관계 표 참고)
# ============================================================

final_estimator = (
    best_model.named_steps["model"] if isinstance(best_model, Pipeline) else best_model
)

importance_tables = {}

if hasattr(final_estimator, "feature_importances_"):

    builtin_imp = pd.Series(
        final_estimator.feature_importances_, index=feature_cols
    )
    builtin_imp = builtin_imp / builtin_imp.sum()
    importance_tables["Built-in (정규화)"] = builtin_imp

print("\nPermutation Importance 계산 중...")

perm = permutation_importance(
    best_model, X_eval, y_eval,
    scoring="f1_macro",
    n_repeats=20,
    random_state=RANDOM_STATE,
    n_jobs=-1,
)

importance_tables["Permutation 평균"] = pd.Series(perm.importances_mean, index=feature_cols)
importance_tables["Permutation SD"] = pd.Series(perm.importances_std, index=feature_cols)

feature_importance_df = (
    pd.DataFrame(importance_tables)
    .sort_values("Permutation 평균", ascending=False)
    .rename_axis("Feature")
    .reset_index()
)
feature_importance_df.insert(0, "Rank", range(1, len(feature_importance_df) + 1))

print("\n[변수 중요도]")
display(feature_importance_df.round(4))

top_n = 15
plot_df = feature_importance_df.head(top_n).iloc[::-1]

plt.figure(figsize=(9, 7))
plt.barh(
    plot_df["Feature"],
    plot_df["Permutation 평균"],
    xerr=plot_df["Permutation SD"],
)
plt.xlabel("섞었을 때 Macro F1 감소량")
plt.title(f"Permutation Importance 상위 {top_n}개 - {best_model_name}")
plt.tight_layout()
plt.show()


# ============================================================
# 24. PCA (EDA용, 모델 학습에는 사용하지 않음)
# ============================================================

pca_input = X.copy()
pca_input[SKEWED_COLS] = np.log1p(pca_input[SKEWED_COLS])
pca_input = StandardScaler().fit_transform(pca_input)

pca = PCA(n_components=2, random_state=RANDOM_STATE)
X_pca = pca.fit_transform(pca_input)

print("\n[PCA 설명 분산 비율]")
print("PC1:", round(pca.explained_variance_ratio_[0], 4))
print("PC2:", round(pca.explained_variance_ratio_[1], 4))
print("누적:", round(pca.explained_variance_ratio_.sum(), 4))

plt.figure(figsize=(9, 7))
for class_name in FAULT_COLS:
    mask = (df["class"] == class_name).values
    plt.scatter(X_pca[mask, 0], X_pca[mask, 1], s=12, alpha=0.6, label=class_name)
plt.xlabel("주성분 1 (PC1)")
plt.ylabel("주성분 2 (PC2)")
plt.title("강판 결함 데이터 PCA (로그 변환 + 표준화)")
plt.legend(markerscale=2)
plt.tight_layout()
plt.show()


# ============================================================
# 25. 최종 분석 요약
# ============================================================

performance_comparison = pd.merge(
    after_cv_results_df[["Model", "CV Macro F1", "CV Macro F1 SD", "CV Balanced Acc"]],
    holdout_results_df.rename(columns=lambda c: c if c == "Model" else f"Eval {c}"),
    on="Model",
)

print("\n" + "=" * 60)
print("최종 분석 요약")
print("=" * 60)
print("전체 데이터:", len(df), "개")
print("학습 데이터:", len(X_train), "개 / 평가 데이터:", len(X_eval), "개")
print("입력 변수:", len(feature_cols), "개 (원본 26 + 파생 6)")
print("\n최종 모델:", best_model_name)
print("최적 파라미터:", searches[best_model_name].best_params_)

print("\n[CV 및 평가 데이터 통합 결과]")
display(performance_comparison.round(4))

print("\n가장 어려운 결함 유형:", class_report_df.index[0],
      f"(F1 {class_report_df['f1-score'].iloc[0]:.3f})")
print("가장 쉬운 결함 유형:", class_report_df.index[-1],
      f"(F1 {class_report_df['f1-score'].iloc[-1]:.3f})")
print("가장 중요한 변수 Top 5:",
      feature_importance_df["Feature"].head(5).tolist())


# ============================================================
# 26. 최종 모델을 전체 데이터로 재학습 및 저장
# ============================================================
# 평가가 끝났으므로 전체 1,941개로 다시 학습하여
# 새로운 강판 데이터 예측에 사용한다.
# ============================================================

final_model = improved_models[best_model_name]
final_model.fit(X, y)

joblib.dump(
    {
        "model": final_model,
        "feature_cols": feature_cols,
        "class_names": FAULT_COLS,
    },
    "steel_fault_model.joblib",
)

print("\n최종 모델 저장 완료: steel_fault_model.joblib")


# ============================================================
# 27. 새로운 데이터 예측 함수
# ============================================================
# 원본과 같은 27개 입력 열을 가진 DataFrame을 넣으면
# 예측 결함 유형과 Class별 확률을 반환한다.
#
# 사용 예:
#   new_df = pd.read_csv("new_plates.csv")
#   pred = predict_faults(new_df)
# ============================================================

def predict_faults(new_data: pd.DataFrame,
                   model_path: str = "steel_fault_model.joblib") -> pd.DataFrame:

    bundle = joblib.load(model_path)
    model = bundle["model"]
    names = bundle["class_names"]

    X_new = add_features(new_data.drop(columns=names, errors="ignore"))
    X_new = X_new[bundle["feature_cols"]]

    proba = model.predict_proba(X_new)

    output = pd.DataFrame(proba, columns=[f"prob_{n}" for n in names],
                          index=new_data.index)
    output.insert(0, "predicted_fault", [names[i] for i in proba.argmax(axis=1)])
    output.insert(1, "confidence", proba.max(axis=1))

    return output


# 동작 확인: 평가 데이터 앞 5개로 예측해 보기
# (전체 데이터로 재학습했으므로 성능 확인용이 아닌 동작 확인용)
demo = predict_faults(raw_df.loc[X_eval.index[:5]])
demo.insert(0, "actual_fault", df.loc[X_eval.index[:5], "class"].values)
print("\n[예측 함수 동작 확인]")
display(demo.round(3))


# ============================================================
# 28. 결과 파일 저장
# ============================================================

stat_df.to_csv("feature_kruskal_results.csv", index=False)
before_after_cv.to_csv("before_after_cv_comparison.csv", index=False)
after_cv_results_df.to_csv("cv_model_comparison.csv", index=False)
holdout_results_df.to_csv("holdout_model_comparison.csv", index=False)
class_report_df.to_csv("class_performance.csv")
feature_importance_df.to_csv("feature_importance.csv", index=False)
misclassified_rows.to_csv("misclassified_rows.csv", index=False)

result_df.assign(
    Correct=result_df["Actual"] == result_df["Predicted"]
).to_csv("eval_predictions.csv")

print("\n결과 파일 저장 완료")
for f in [
    "steel_fault_model.joblib",
    "feature_kruskal_results.csv",
    "before_after_cv_comparison.csv",
    "cv_model_comparison.csv",
    "holdout_model_comparison.csv",
    "class_performance.csv",
    "feature_importance.csv",
    "misclassified_rows.csv",
    "eval_predictions.csv",
]:
    print("-", f)
```
