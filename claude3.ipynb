```python
# ============================================================
# 강판(Steel Plates) 표면 결함 유형 분류
# 학습 / 검증 / 테스트 3분할 통합 코드
# ============================================================
# 데이터: steel_plates_faults.csv (1,941행 × 34열)
#
# 전처리
#   - 강종 2열(TypeOfSteel_A300 / A400) → Steel_Type 1열 (A300 / A400)
#   - 결함 7열(Pastry ~ Other_Faults)   → Fault_Type 1열 (타깃)
#
# 분할 (층화 추출, 60 / 20 / 20)
#   - 학습(Train)     : 모델 학습
#   - 검증(Validation): 하이퍼파라미터 선택, 모델 선택,
#                       LightGBM 조기 종료(early stopping)
#   - 테스트(Test)    : 모든 결정이 끝난 뒤 마지막에 한 번만 평가
#
# 모델
#   - LightGBM (주력 추천)
#   - Random Forest (비교)
#   - 로지스틱 회귀 (기준선)
# ============================================================


# ============================================================
# 0. 라이브러리
# ============================================================

import time
import warnings

import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

try:
    from IPython.display import display
except ImportError:
    display = print

from scipy.stats import chi2_contingency, kruskal, loguniform, randint, uniform

from sklearn.model_selection import train_test_split, ParameterSampler
from sklearn.preprocessing import StandardScaler, FunctionTransformer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.decomposition import PCA
from sklearn.inspection import permutation_importance
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
)

from lightgbm import LGBMClassifier, early_stopping

warnings.filterwarnings("ignore")
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)


# ============================================================
# 1. 기본 설정
# ============================================================

plt.rcParams["font.family"] = "Malgun Gothic"   # macOS: "AppleGothic"
plt.rcParams["axes.unicode_minus"] = False

RANDOM_STATE = 42
DATA_PATH = "steel_plates_faults.csv"

STEEL_COLS = ["TypeOfSteel_A300", "TypeOfSteel_A400"]

FAULT_COLS = [
    "Pastry",
    "Z_Scratch",
    "K_Scatch",
    "Stains",
    "Dirtiness",
    "Bumps",
    "Other_Faults",
]

TARGET = "Fault_Type"

class_mapping = {name: i for i, name in enumerate(FAULT_COLS)}
reverse_class_mapping = {i: name for name, i in class_mapping.items()}
LABELS = list(range(len(FAULT_COLS)))

# 모델 입력용 강종 코드 (고정 규칙이므로 데이터 누수 없음)
STEEL_CODE = {"A300": 0, "A400": 1}


# ============================================================
# 2. 데이터 불러오기
# ============================================================

raw_df = pd.read_csv(DATA_PATH)

print("원본 데이터 크기:", raw_df.shape)
display(raw_df.head())


# ============================================================
# 3. 데이터 기본 확인
# ============================================================

print("\n[결측치 합계]:", raw_df.isnull().sum().sum())
print("[중복 행 수]:", raw_df.duplicated().sum())

# 합치기 전에 원-핫 구조가 맞는지 검증
steel_sum = raw_df[STEEL_COLS].sum(axis=1)
fault_sum = raw_df[FAULT_COLS].sum(axis=1)

print("\n[강종 열 합계 분포] (모두 1이어야 함)")
display(steel_sum.value_counts())

print("\n[결함 열 합계 분포] (모두 1이어야 함)")
display(fault_sum.value_counts())

assert (steel_sum == 1).all(), "강종이 하나로 정해지지 않는 행이 있습니다."
assert (fault_sum == 1).all(), "결함 유형이 하나로 정해지지 않는 행이 있습니다."


# ============================================================
# 4. 강종 / 결함 유형을 각각 한 열로 합치기
# ============================================================

def merge_columns(data: pd.DataFrame) -> pd.DataFrame:
    """
    강종 2열 → Steel_Type, 결함 7열 → Fault_Type
    (새 데이터에 결함 열이 없으면 Steel_Type만 합친다)
    """

    out = data.copy()

    if set(STEEL_COLS).issubset(out.columns):

        position = out.columns.get_loc("TypeOfSteel_A300")
        steel_type = np.where(out["TypeOfSteel_A300"] == 1, "A300", "A400")

        out = out.drop(columns=STEEL_COLS)
        out.insert(position, "Steel_Type", steel_type)

    if set(FAULT_COLS).issubset(out.columns):

        out[TARGET] = out[FAULT_COLS].idxmax(axis=1)
        out = out.drop(columns=FAULT_COLS)

    return out


df = merge_columns(raw_df)

print("\n합친 후 데이터 크기:", df.shape)
print("(34열 → 강종 2열이 1열로, 결함 7열이 1열로 → 27열)")
display(df.head())

# 합친 결과 확인용 저장
df.to_csv("steel_plates_merged.csv", index=False)


# ============================================================
# 5. 결함 유형(타깃) 분포
# ============================================================

class_counts = df[TARGET].value_counts().reindex(FAULT_COLS)

class_distribution = pd.DataFrame({
    "count": class_counts,
    "ratio(%)": (class_counts / len(df) * 100).round(2),
})

print("\n[결함 유형 분포]")
display(class_distribution)
print(
    "최대/최소 비율:",
    round(class_counts.max() / class_counts.min(), 1), "배",
    "→ 불균형 데이터이므로 Macro F1을 주 지표로 사용",
)

plt.figure(figsize=(8, 4))
plt.bar(class_distribution.index, class_distribution["count"])
plt.xlabel("결함 유형")
plt.ylabel("개수")
plt.title("결함 유형별 개수")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()


# ============================================================
# 6. 강종 - 결함 유형 관계 (카이제곱 검정)
# ============================================================
#
# H0: 강종과 결함 유형은 독립이다.
# H1: 강종과 결함 유형은 독립이 아니다.
#
# Cramér's V로 연관성의 크기도 함께 본다.
# (0.1 약함 / 0.3 중간 / 0.5 이상 강함)
# ============================================================

steel_fault_table = pd.crosstab(df[TARGET], df["Steel_Type"]).reindex(FAULT_COLS)

print("\n[결함 유형 × 강종 교차표]")
display(steel_fault_table)

print("\n[결함 유형별 강종 비율]")
display(
    pd.crosstab(df[TARGET], df["Steel_Type"], normalize="index")
    .reindex(FAULT_COLS)
    .round(3)
)

chi2, p_value, dof, expected = chi2_contingency(steel_fault_table)
cramers_v = np.sqrt(chi2 / (len(df) * (min(steel_fault_table.shape) - 1)))

print("\n[카이제곱 검정]")
print(f"Chi2 = {chi2:.2f}, 자유도 = {dof}, p-value = "
      f"{'<0.001' if p_value < 0.001 else round(p_value, 4)}")
print(f"Cramér's V = {cramers_v:.3f}")

steel_fault_table.div(steel_fault_table.sum(axis=1), axis=0).plot(
    kind="bar", stacked=True, figsize=(9, 4)
)
plt.title("결함 유형별 강종 구성")
plt.xlabel("결함 유형")
plt.ylabel("비율")
plt.xticks(rotation=30)
plt.legend(title="강종")
plt.tight_layout()
plt.show()


# ============================================================
# 7. 모델 입력 변수 생성
# ============================================================
# - Steel_Type: A300 → 0, A400 → 1 (한 열 유지)
# - 파생 변수: 가로/세로 길이, 가로세로비, 밝기 범위,
#             평균 밝기, 면적 대비 둘레
# ============================================================

def build_features(data: pd.DataFrame) -> pd.DataFrame:

    out = data.drop(columns=[TARGET], errors="ignore").copy()

    out["Steel_Type"] = out["Steel_Type"].map(STEEL_CODE)

    if out["Steel_Type"].isnull().any():
        raise ValueError("Steel_Type에 A300 / A400 이외의 값이 있습니다.")

    out["X_Range"] = out["X_Maximum"] - out["X_Minimum"]
    out["Y_Range"] = out["Y_Maximum"] - out["Y_Minimum"]
    out["Aspect_Ratio"] = (out["X_Range"] + 1) / (out["Y_Range"] + 1)

    out["Luminosity_Range"] = (
        out["Maximum_of_Luminosity"] - out["Minimum_of_Luminosity"]
    )
    out["Mean_Luminosity"] = out["Sum_of_Luminosity"] / out["Pixels_Areas"]

    out["Perimeter_Area_Ratio"] = (
        (out["X_Perimeter"] + out["Y_Perimeter"]) / out["Pixels_Areas"]
    )

    return out


X_all = build_features(df)
y_all = df[TARGET].map(class_mapping)

feature_cols = X_all.columns.tolist()

print("\n입력 변수 수:", len(feature_cols))
print(feature_cols)

# 로지스틱 회귀에서만 로그 변환할 치우친 변수
SKEWED_COLS = [
    "Pixels_Areas",
    "X_Perimeter",
    "Y_Perimeter",
    "Sum_of_Luminosity",
    "X_Range",
    "Y_Range",
    "Aspect_Ratio",
    "Perimeter_Area_Ratio",
]


# ============================================================
# 8. 수치 변수 - 결함 유형 연관성 (Kruskal-Wallis)
# ============================================================
# 수치 변수는 정규분포가 아니므로 비모수 검정을 사용한다.
# 표본이 커서 p-value는 대부분 매우 작으므로
# 효과크기 ε²로 크기를 비교한다.
# ============================================================

n_obs, n_class = len(df), len(FAULT_COLS)
kw_results = []

for col in feature_cols:

    if col == "Steel_Type":
        continue

    groups = [X_all.loc[y_all == k, col] for k in LABELS]
    h_stat, p_val = kruskal(*groups)

    kw_results.append({
        "Feature": col,
        "H": h_stat,
        "p-value": "<0.001" if p_val < 0.001 else round(p_val, 4),
        "Epsilon²": max(0.0, (h_stat - n_class + 1) / (n_obs - n_class)),
    })

kw_df = (
    pd.DataFrame(kw_results)
    .sort_values("Epsilon²", ascending=False)
    .reset_index(drop=True)
)

print("\n[수치 변수 Kruskal-Wallis 검정 결과]")
display(kw_df.round(4))


# ============================================================
# 9. 학습 / 검증 / 테스트 분할 (60 / 20 / 20, 층화)
# ============================================================
# 1차: 전체에서 테스트 20% 분리
# 2차: 남은 80%에서 25%를 검증으로 분리 (전체 기준 20%)
# 두 단계 모두 stratify를 걸어 결함 유형 비율을 유지한다.
# ============================================================

X_temp, X_test, y_temp, y_test = train_test_split(
    X_all, y_all,
    test_size=0.20,
    stratify=y_all,
    random_state=RANDOM_STATE,
)

X_train, X_val, y_train, y_val = train_test_split(
    X_temp, y_temp,
    test_size=0.25,
    stratify=y_temp,
    random_state=RANDOM_STATE,
)

print("\n[데이터 분할]")
print(f"학습  : {len(X_train):>5}개 ({len(X_train) / len(X_all):.0%})")
print(f"검증  : {len(X_val):>5}개 ({len(X_val) / len(X_all):.0%})")
print(f"테스트: {len(X_test):>5}개 ({len(X_test) / len(X_all):.0%})")

split_table = pd.DataFrame({
    "Train": y_train.value_counts().sort_index(),
    "Val": y_val.value_counts().sort_index(),
    "Test": y_test.value_counts().sort_index(),
}).rename(index=reverse_class_mapping)

split_ratio = split_table.div(split_table.sum(axis=0), axis=1).mul(100).round(2)
split_ratio.columns = [f"{c}(%)" for c in split_ratio.columns]

print("\n[분할별 결함 유형 개수 / 비율]")
display(pd.concat([split_table, split_ratio], axis=1))

# 분할 결과를 원본 행 번호와 함께 저장 (재현성 확인용)
split_info = pd.Series("train", index=X_all.index)
split_info[X_val.index] = "val"
split_info[X_test.index] = "test"
df.assign(Split=split_info).to_csv("data_split_info.csv", index_label="row_id")


# ============================================================
# 10. 모델 정의
# ============================================================

def build_model(name, params=None):

    params = params or {}

    if name == "LightGBM":
        base = dict(
            objective="multiclass",
            n_estimators=2000,        # 상한값. 실제 개수는 조기 종료로 결정
            learning_rate=0.05,
            num_leaves=31,
            class_weight="balanced",
            random_state=RANDOM_STATE,
            n_jobs=-1,
            verbose=-1,
        )
        base.update(params)
        return LGBMClassifier(**base)

    if name == "Random Forest":
        base = dict(
            n_estimators=300,
            class_weight="balanced_subsample",
            random_state=RANDOM_STATE,
            n_jobs=-1,
        )
        base.update(params)
        return RandomForestClassifier(**base)

    if name == "Logistic Regression":
        other_cols = [c for c in feature_cols if c not in SKEWED_COLS]
        preprocessor = ColumnTransformer([
            ("log", FunctionTransformer(np.log1p, feature_names_out="one-to-one"),
             SKEWED_COLS),
            ("pass", "passthrough", other_cols),
        ])
        base = dict(
            C=1.0,
            class_weight="balanced",
            max_iter=5000,
            random_state=RANDOM_STATE,
        )
        base.update(params)
        return Pipeline([
            ("preprocess", preprocessor),
            ("scaler", StandardScaler()),   # 학습 데이터로만 fit됨
            ("model", LogisticRegression(**base)),
        ])

    raise ValueError(name)


def fit_model(name, params=None):
    """학습 데이터로 학습. LightGBM은 검증 데이터로 조기 종료."""

    model = build_model(name, params)

    if name == "LightGBM":
        model.fit(
            X_train, y_train,
            eval_set=[(X_val, y_val)],
            eval_metric="multi_logloss",
            callbacks=[early_stopping(stopping_rounds=50, verbose=False)],
        )
    else:
        model.fit(X_train, y_train)

    return model


def score_row(y_true, y_hat):
    return {
        "Accuracy": accuracy_score(y_true, y_hat),
        "Balanced Acc": balanced_accuracy_score(y_true, y_hat),
        "Precision": precision_score(y_true, y_hat, average="macro", zero_division=0),
        "Recall": recall_score(y_true, y_hat, average="macro", zero_division=0),
        "Macro F1": f1_score(y_true, y_hat, average="macro", zero_division=0),
    }


MODEL_NAMES = ["LightGBM", "Random Forest", "Logistic Regression"]


# ============================================================
# 11. 개선 전: 기본 하이퍼파라미터 → 검증 데이터 평가
# ============================================================

before_rows = []
base_models = {}

print("\n" + "=" * 60)
print("개선 전 (기본 하이퍼파라미터) 검증 성능")
print("=" * 60)

for name in MODEL_NAMES:

    start = time.time()
    model = fit_model(name)
    base_models[name] = model

    before_rows.append({
        "Model": name,
        **score_row(y_val, model.predict(X_val)),
        "Time(s)": time.time() - start,
    })

before_val_df = pd.DataFrame(before_rows)
display(before_val_df.round(4))


# ============================================================
# 12. 하이퍼파라미터 탐색 (검증 데이터 기준)
# ============================================================
#
# 후보 조합을 무작위로 뽑아
#   학습 데이터로 학습 → 검증 데이터 Macro F1 계산
# 을 반복하고, 검증 Macro F1이 가장 높은 조합을 고른다.
#
# 테스트 데이터는 이 단계에서 전혀 사용하지 않는다.
# ============================================================

search_spaces = {

    "LightGBM": {
        "learning_rate": loguniform(0.01, 0.2),
        "num_leaves": randint(8, 64),
        "min_child_samples": randint(5, 40),
        "subsample": uniform(0.6, 0.4),
        "subsample_freq": [1],
        "colsample_bytree": uniform(0.5, 0.5),
        "reg_lambda": loguniform(1e-3, 10),
    },

    "Random Forest": {
        "n_estimators": randint(200, 800),
        "max_depth": [None, 10, 15, 20, 30],
        "min_samples_split": randint(2, 11),
        "min_samples_leaf": randint(1, 5),
        "max_features": ["sqrt", "log2", 0.3, 0.5],
    },

    "Logistic Regression": {
        "C": loguniform(1e-2, 1e2),
    },
}

n_iter_map = {
    "LightGBM": 40,
    "Random Forest": 30,
    "Logistic Regression": 20,
}

search_logs = {}
best_params = {}
tuned_models = {}

for name in MODEL_NAMES:

    print(f"\n{name} 하이퍼파라미터 탐색 중... ({n_iter_map[name]}개 조합)")
    start = time.time()

    sampler = ParameterSampler(
        search_spaces[name],
        n_iter=n_iter_map[name],
        random_state=RANDOM_STATE,
    )

    log_rows = []
    best_score, best_model = -np.inf, None

    for params in sampler:

        params = {
            k: (v.item() if hasattr(v, "item") else v) for k, v in params.items()
        }

        model = fit_model(name, params)
        val_f1 = f1_score(y_val, model.predict(X_val), average="macro")

        row = {**params, "Val Macro F1": val_f1}
        if name == "LightGBM":
            row["best_iteration"] = model.best_iteration_
        log_rows.append(row)

        if val_f1 > best_score:
            best_score, best_model, best_param = val_f1, model, params

    search_logs[name] = (
        pd.DataFrame(log_rows)
        .sort_values("Val Macro F1", ascending=False)
        .reset_index(drop=True)
    )
    best_params[name] = best_param
    tuned_models[name] = best_model

    print(f"  소요 시간: {time.time() - start:.1f}초")
    print(f"  최고 검증 Macro F1: {best_score:.4f}")
    print(f"  최적 파라미터: {best_param}")
    if name == "LightGBM":
        print(f"  조기 종료 트리 수: {best_model.best_iteration_}")

print("\n[LightGBM 탐색 상위 5개 조합]")
display(search_logs["LightGBM"].head().round(4))


# ============================================================
# 13. 개선 전후 검증 성능 비교
# ============================================================

after_rows = [
    {"Model": name, **score_row(y_val, tuned_models[name].predict(X_val))}
    for name in MODEL_NAMES
]
after_val_df = pd.DataFrame(after_rows)

before_after_val = pd.merge(
    before_val_df[["Model", "Accuracy", "Macro F1"]],
    after_val_df[["Model", "Accuracy", "Macro F1"]],
    on="Model",
    suffixes=("_Before", "_After"),
)
before_after_val["Macro F1 변화"] = (
    before_after_val["Macro F1_After"] - before_after_val["Macro F1_Before"]
)

print("\n[개선 전후 검증 성능 비교]")
display(before_after_val.round(4))

x_pos = np.arange(len(MODEL_NAMES))
width = 0.35

plt.figure(figsize=(9, 5))
plt.bar(x_pos - width / 2, before_after_val["Macro F1_Before"], width, label="개선 전")
plt.bar(x_pos + width / 2, before_after_val["Macro F1_After"], width, label="개선 후")
plt.xticks(x_pos, before_after_val["Model"])
plt.ylabel("검증 Macro F1")
plt.title("하이퍼파라미터 개선 전후 검증 Macro F1")
plt.ylim(max(0, before_after_val[["Macro F1_Before", "Macro F1_After"]].min().min() - 0.1), 1.0)
plt.legend()
plt.tight_layout()
plt.show()


# ============================================================
# 14. LightGBM 학습 곡선 (검증 데이터의 조기 종료 역할 확인)
# ============================================================

lgbm_best = tuned_models["LightGBM"]
val_curve = lgbm_best.evals_result_["valid_0"]["multi_logloss"]

plt.figure(figsize=(8, 4))
plt.plot(range(1, len(val_curve) + 1), val_curve)
plt.axvline(lgbm_best.best_iteration_, color="red", linestyle="--",
            label=f"조기 종료 지점 ({lgbm_best.best_iteration_})")
plt.xlabel("트리 수")
plt.ylabel("검증 Log Loss")
plt.title("LightGBM 검증 손실 곡선")
plt.legend()
plt.tight_layout()
plt.show()


# ============================================================
# 15. 최종 모델 선택 (검증 Macro F1 기준)
# ============================================================

best_model_name = after_val_df.sort_values("Macro F1", ascending=False).iloc[0]["Model"]
best_model = tuned_models[best_model_name]

print("\n" + "=" * 60)
print("최종 모델 선정 (검증 데이터 기준)")
print("=" * 60)
print("선택 모델:", best_model_name)
print("검증 Macro F1:", round(after_val_df.set_index("Model").loc[best_model_name, "Macro F1"], 4))
print("최적 파라미터:", best_params[best_model_name])


# ============================================================
# 16. 테스트 데이터 평가 (마지막 1회)
# ============================================================
# 모든 결정(파라미터, 모델 선택)이 끝났으므로 여기서 처음 테스트를 사용한다.
# 다른 모델의 테스트 성능은 참고용이며, 이를 보고 선택을 바꾸지 않는다.
# ============================================================

summary_rows = []

for name in MODEL_NAMES:

    model = tuned_models[name]

    summary_rows.append({
        "Model": name,
        "Train Macro F1": f1_score(y_train, model.predict(X_train), average="macro"),
        "Val Macro F1": f1_score(y_val, model.predict(X_val), average="macro"),
        **{f"Test {k}": v for k, v in score_row(y_test, model.predict(X_test)).items()},
    })

summary_df = pd.DataFrame(summary_rows)
summary_df["Val-Test 차이"] = summary_df["Val Macro F1"] - summary_df["Test Macro F1"]

print("\n[학습 / 검증 / 테스트 성능 비교]")
display(summary_df.round(4))

y_pred = best_model.predict(X_test)
test_scores = score_row(y_test, y_pred)

print("\n" + "=" * 60)
print(f"최종 테스트 결과 - {best_model_name}")
print("=" * 60)
for metric, value in test_scores.items():
    print(f"{metric}: {value:.4f}")


# ============================================================
# 17. Classification Report / 결함 유형별 성능
# ============================================================

print("\n[Classification Report - Test]")
print(classification_report(
    y_test, y_pred,
    labels=LABELS, target_names=FAULT_COLS,
    digits=4, zero_division=0,
))

class_report_df = (
    pd.DataFrame(classification_report(
        y_test, y_pred,
        labels=LABELS, target_names=FAULT_COLS,
        output_dict=True, zero_division=0,
    )).T
    .loc[FAULT_COLS, ["precision", "recall", "f1-score", "support"]]
    .sort_values("f1-score")
)

print("\n[결함 유형별 성능 (F1 낮은 순)]")
display(class_report_df.round(4))


# ============================================================
# 18. 혼동행렬
# ============================================================

cm = confusion_matrix(y_test, y_pred, labels=LABELS)
cm_norm = confusion_matrix(y_test, y_pred, labels=LABELS, normalize="true")

fig, axes = plt.subplots(1, 2, figsize=(18, 7))

ConfusionMatrixDisplay(cm, display_labels=FAULT_COLS).plot(
    ax=axes[0], xticks_rotation=45, colorbar=False
)
axes[0].set_title(f"혼동행렬 (개수) - {best_model_name}")

ConfusionMatrixDisplay(cm_norm, display_labels=FAULT_COLS).plot(
    ax=axes[1], xticks_rotation=45, colorbar=False, values_format=".2f"
)
axes[1].set_title("혼동행렬 (실제 유형 기준 비율)")

for ax in axes:
    ax.set_xlabel("예측 결함 유형")
    ax.set_ylabel("실제 결함 유형")

plt.tight_layout()
plt.show()


# ============================================================
# 19. 오분류 분석
# ============================================================

test_result_df = pd.DataFrame({
    "Actual": y_test.map(reverse_class_mapping),
    "Predicted": pd.Series(y_pred, index=y_test.index).map(reverse_class_mapping),
    "Confidence": best_model.predict_proba(X_test).max(axis=1),
})

misclassified = test_result_df[test_result_df["Actual"] != test_result_df["Predicted"]]

print("\n" + "=" * 60)
print("오분류 분석 (테스트 데이터)")
print("=" * 60)
print("테스트 데이터:", len(y_test), "개")
print("오분류:", len(misclassified), "개",
      f"({len(misclassified) / len(y_test):.2%})")

misclassification_summary = (
    misclassified
    .groupby(["Actual", "Predicted"])
    .size()
    .reset_index(name="Count")
    .sort_values("Count", ascending=False)
    .reset_index(drop=True)
)

print("\n[주요 오분류 조합]")
display(misclassification_summary.head(10))

other_related = misclassified.query(
    "Actual == 'Other_Faults' or Predicted == 'Other_Faults'"
)
print("Other_Faults가 관련된 오분류 비중:",
      f"{len(other_related) / max(len(misclassified), 1):.1%}")

misclassified_rows = df.loc[misclassified.index].join(
    misclassified[["Predicted", "Confidence"]]
)

print("\n[오분류 데이터 행]")
display(misclassified_rows.head(20))


# ============================================================
# 20. 변수 중요도 (Permutation Importance, 테스트 데이터)
# ============================================================
# 변수 값을 무작위로 섞었을 때 Macro F1이 얼마나 떨어지는지 측정.
# 상관이 높은 변수끼리는 중요도가 나뉘어 낮게 나올 수 있다.
# ============================================================

print("\nPermutation Importance 계산 중...")

perm = permutation_importance(
    best_model, X_test, y_test,
    scoring="f1_macro",
    n_repeats=20,
    random_state=RANDOM_STATE,
    n_jobs=-1,
)

importance_df = pd.DataFrame({
    "Feature": feature_cols,
    "Importance Mean": perm.importances_mean,
    "Importance SD": perm.importances_std,
})

final_estimator = (
    best_model.named_steps["model"] if isinstance(best_model, Pipeline) else best_model
)
if hasattr(final_estimator, "feature_importances_"):
    builtin = final_estimator.feature_importances_
    importance_df["Built-in (정규화)"] = builtin / builtin.sum()

importance_df = (
    importance_df
    .sort_values("Importance Mean", ascending=False)
    .reset_index(drop=True)
)
importance_df.insert(0, "Rank", range(1, len(importance_df) + 1))

print("\n[변수 중요도]")
display(importance_df.round(4))

plot_df = importance_df.head(15).iloc[::-1]

plt.figure(figsize=(9, 7))
plt.barh(plot_df["Feature"], plot_df["Importance Mean"], xerr=plot_df["Importance SD"])
plt.xlabel("섞었을 때 Macro F1 감소량")
plt.title(f"Permutation Importance 상위 15개 - {best_model_name}")
plt.tight_layout()
plt.show()


# ============================================================
# 21. PCA (EDA용, 모델 학습에는 사용하지 않음)
# ============================================================

pca_input = X_all.copy()
pca_input[SKEWED_COLS] = np.log1p(pca_input[SKEWED_COLS])
pca_input = StandardScaler().fit_transform(pca_input)

pca = PCA(n_components=2, random_state=RANDOM_STATE)
X_pca = pca.fit_transform(pca_input)

print("\n[PCA 설명 분산 비율]")
print("PC1:", round(pca.explained_variance_ratio_[0], 4),
      "/ PC2:", round(pca.explained_variance_ratio_[1], 4),
      "/ 누적:", round(pca.explained_variance_ratio_.sum(), 4))

plt.figure(figsize=(9, 7))
for k, name in reverse_class_mapping.items():
    mask = (y_all == k).values
    plt.scatter(X_pca[mask, 0], X_pca[mask, 1], s=12, alpha=0.6, label=name)
plt.xlabel("주성분 1 (PC1)")
plt.ylabel("주성분 2 (PC2)")
plt.title("강판 결함 데이터 PCA")
plt.legend(markerscale=2)
plt.tight_layout()
plt.show()


# ============================================================
# 22. 최종 분석 요약
# ============================================================

print("\n" + "=" * 60)
print("최종 분석 요약")
print("=" * 60)
print("전체 데이터:", len(df), "개 (27열: 입력 26 + 타깃 1)")
print(f"분할: 학습 {len(X_train)} / 검증 {len(X_val)} / 테스트 {len(X_test)}")
print("모델 입력 변수:", len(feature_cols), "개 (원본 25 + Steel_Type 1 + 파생 6)")
print(f"강종-결함 유형 연관성: Cramér's V = {cramers_v:.3f}")
print("\n최종 모델:", best_model_name)
print("최적 파라미터:", best_params[best_model_name])
print("검증 Macro F1:", round(summary_df.set_index("Model").loc[best_model_name, "Val Macro F1"], 4))
print("테스트 Macro F1:", round(test_scores["Macro F1"], 4))
print("테스트 Accuracy:", round(test_scores["Accuracy"], 4))
print("\n가장 어려운 결함 유형:", class_report_df.index[0],
      f"(F1 {class_report_df['f1-score'].iloc[0]:.3f})")
print("가장 쉬운 결함 유형:", class_report_df.index[-1],
      f"(F1 {class_report_df['f1-score'].iloc[-1]:.3f})")
print("중요 변수 Top 5:", importance_df["Feature"].head(5).tolist())


# ============================================================
# 23. 최종 모델을 전체 데이터로 재학습 및 저장
# ============================================================
# 평가가 끝났으므로 학습+검증+테스트 전체로 다시 학습한다.
# 검증 데이터가 없으므로 LightGBM은 조기 종료로 찾은
# 트리 수를 그대로 고정해서 사용한다.
# ============================================================

final_params = dict(best_params[best_model_name])

if best_model_name == "LightGBM":
    final_params["n_estimators"] = int(lgbm_best.best_iteration_)

final_model = build_model(best_model_name, final_params)
final_model.fit(X_all, y_all)

joblib.dump(
    {
        "model": final_model,
        "model_name": best_model_name,
        "params": final_params,
        "feature_cols": feature_cols,
        "class_names": FAULT_COLS,
    },
    "steel_fault_model.joblib",
)

print("\n최종 모델 저장 완료: steel_fault_model.joblib")
print("재학습 파라미터:", final_params)


# ============================================================
# 24. 새로운 데이터 예측 함수
# ============================================================
# 원본 형식(TypeOfSteel_A300/A400 두 열) 또는
# 합친 형식(Steel_Type 한 열) 둘 다 입력 가능.
#
# 사용 예:
#   new_df = pd.read_csv("new_plates.csv")
#   pred = predict_faults(new_df)
# ============================================================

def predict_faults(new_data: pd.DataFrame,
                   model_path: str = "steel_fault_model.joblib") -> pd.DataFrame:

    bundle = joblib.load(model_path)
    model, names = bundle["model"], bundle["class_names"]

    merged = merge_columns(new_data.drop(columns=names, errors="ignore"))
    X_new = build_features(merged)[bundle["feature_cols"]]

    proba = model.predict_proba(X_new)

    output = pd.DataFrame(
        proba, columns=[f"prob_{n}" for n in names], index=new_data.index
    )
    output.insert(0, "predicted_fault", [names[i] for i in proba.argmax(axis=1)])
    output.insert(1, "confidence", proba.max(axis=1))

    return output


# 동작 확인 (전체 데이터로 재학습한 모델이므로 성능 확인용이 아님)
demo = predict_faults(raw_df.loc[X_test.index[:5]])
demo.insert(0, "actual_fault", df.loc[X_test.index[:5], TARGET].values)

print("\n[예측 함수 동작 확인]")
display(demo.round(3))


# ============================================================
# 25. 결과 파일 저장
# ============================================================

output_files = {
    "steel_plates_merged.csv": None,     # 4단계에서 저장
    "data_split_info.csv": None,         # 9단계에서 저장
    "numeric_kruskal_results.csv": kw_df,
    "before_after_val_comparison.csv": before_after_val,
    "train_val_test_comparison.csv": summary_df,
    "lgbm_search_log.csv": search_logs["LightGBM"],
    "class_performance_test.csv": class_report_df,
    "feature_importance.csv": importance_df,
    "misclassified_rows.csv": misclassified_rows,
    "test_predictions.csv": test_result_df,
}

for filename, table in output_files.items():
    if table is not None:
        table.to_csv(filename, index=filename in {
            "class_performance_test.csv", "test_predictions.csv",
            "misclassified_rows.csv",
        })

print("\n결과 파일 저장 완료")
for filename in ["steel_fault_model.joblib", *output_files]:
    print("-", filename)
```
