```python
# ============================================================
# 강판(Steel Plates) 표면 결함 유형 분류 - 성능 개선판 (v2)
# ============================================================
# 데이터: steel_plates_faults.csv (1,941행 × 34열)
#
# 전처리
#   - 강종 2열(TypeOfSteel_A300 / A400) → Steel_Type 1열
#   - 결함 7열(Pastry ~ Other_Faults)   → Fault_Type 1열 (타깃)
#
# 분할 (층화 추출, 60 / 20 / 20)
#   - 학습(Train)     : 모델 학습
#   - 검증(Validation): ① 부스팅 모델 트리 수 결정(조기 종료)
#                       ② 학습→검증 홀드아웃 점수로 CV 결과 2차 확인
#   - 학습+검증(Dev)  : 5-Fold CV로 하이퍼파라미터 탐색·모델 선택
#   - 테스트(Test)    : 모든 결정이 끝난 뒤 마지막에 한 번만 평가
#
# 이전 버전 대비 개선 사항
#   ① 테스트 전에 학습+검증 데이터로 재학습 (학습 데이터 33% 증가)
#   ② ExtraTrees 추가 (이 데이터에서 단일 모델 최고 성능)
#   ③ 컨베이어 기준 상대 위치 등 파생 변수 6개 추가
#   ④ 소프트 보팅 앙상블 (OOF 확률로 최적 조합 선택)
#   ⑤ 모델 선택 기준: 검증 1회 점수 → 5-Fold CV 점수
# ============================================================


# ============================================================
# 0. 라이브러리
# ============================================================

import itertools
import time
import warnings

import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

try:
    from IPython.display import display
except ImportError:
    display = print

from scipy.stats import chi2_contingency, kruskal, randint, uniform

from sklearn.base import clone
from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    RandomizedSearchCV,
)
from sklearn.preprocessing import StandardScaler, FunctionTransformer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    VotingClassifier,
)
from sklearn.inspection import permutation_importance
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
)

from lightgbm import LGBMClassifier, early_stopping
from xgboost import XGBClassifier

warnings.filterwarnings("ignore")
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)


# ============================================================
# 1. 기본 설정
# ============================================================

plt.rcParams["font.family"] = "Malgun Gothic"   # macOS: "AppleGothic"
plt.rcParams["axes.unicode_minus"] = False

RANDOM_STATE = 42
DATA_PATH = "steel_plates_faults.csv"

STEEL_COLS = ["TypeOfSteel_A300", "TypeOfSteel_A400"]

FAULT_COLS = [
    "Pastry",
    "Z_Scratch",
    "K_Scatch",
    "Stains",
    "Dirtiness",
    "Bumps",
    "Other_Faults",
]

TARGET = "Fault_Type"

class_mapping = {name: i for i, name in enumerate(FAULT_COLS)}
reverse_class_mapping = {i: name for name, i in class_mapping.items()}
LABELS = list(range(len(FAULT_COLS)))

# 모델 입력용 강종 코드 (고정 규칙이므로 데이터 누수 없음)
STEEL_CODE = {"A300": 0, "A400": 1}


# ============================================================
# 2. 데이터 불러오기
# ============================================================

raw_df = pd.read_csv(DATA_PATH)

print("원본 데이터 크기:", raw_df.shape)
display(raw_df.head())


# ============================================================
# 3. 데이터 기본 확인
# ============================================================

print("\n[결측치 합계]:", raw_df.isnull().sum().sum())
print("[중복 행 수]:", raw_df.duplicated().sum())

# 합치기 전에 원-핫 구조가 맞는지 검증
steel_sum = raw_df[STEEL_COLS].sum(axis=1)
fault_sum = raw_df[FAULT_COLS].sum(axis=1)

print("\n[강종 열 합계 분포] (모두 1이어야 함)")
display(steel_sum.value_counts())

print("\n[결함 열 합계 분포] (모두 1이어야 함)")
display(fault_sum.value_counts())

assert (steel_sum == 1).all(), "강종이 하나로 정해지지 않는 행이 있습니다."
assert (fault_sum == 1).all(), "결함 유형이 하나로 정해지지 않는 행이 있습니다."


# ============================================================
# 4. 강종 / 결함 유형을 각각 한 열로 합치기
# ============================================================

def merge_columns(data: pd.DataFrame) -> pd.DataFrame:
    """
    강종 2열 → Steel_Type, 결함 7열 → Fault_Type
    (새 데이터에 결함 열이 없으면 Steel_Type만 합친다)
    """

    out = data.copy()

    if set(STEEL_COLS).issubset(out.columns):

        position = out.columns.get_loc("TypeOfSteel_A300")
        steel_type = np.where(out["TypeOfSteel_A300"] == 1, "A300", "A400")

        out = out.drop(columns=STEEL_COLS)
        out.insert(position, "Steel_Type", steel_type)

    if set(FAULT_COLS).issubset(out.columns):

        out[TARGET] = out[FAULT_COLS].idxmax(axis=1)
        out = out.drop(columns=FAULT_COLS)

    return out


df = merge_columns(raw_df)

print("\n합친 후 데이터 크기:", df.shape)
print("(34열 → 강종 2열이 1열로, 결함 7열이 1열로 → 27열)")
display(df.head())

# 합친 결과 확인용 저장
df.to_csv("steel_plates_merged.csv", index=False)


# ============================================================
# 5. 결함 유형(타깃) 분포
# ============================================================

class_counts = df[TARGET].value_counts().reindex(FAULT_COLS)

class_distribution = pd.DataFrame({
    "count": class_counts,
    "ratio(%)": (class_counts / len(df) * 100).round(2),
})

print("\n[결함 유형 분포]")
display(class_distribution)
print(
    "최대/최소 비율:",
    round(class_counts.max() / class_counts.min(), 1), "배",
    "→ 불균형 데이터이므로 Macro F1을 주 지표로 사용",
)

plt.figure(figsize=(8, 4))
plt.bar(class_distribution.index, class_distribution["count"])
plt.xlabel("결함 유형")
plt.ylabel("개수")
plt.title("결함 유형별 개수")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()


# ============================================================
# 6. 강종 - 결함 유형 관계 (카이제곱 검정)
# ============================================================
#
# H0: 강종과 결함 유형은 독립이다.
# H1: 강종과 결함 유형은 독립이 아니다.
#
# Cramér's V로 연관성의 크기도 함께 본다.
# (0.1 약함 / 0.3 중간 / 0.5 이상 강함)
# ============================================================

steel_fault_table = pd.crosstab(df[TARGET], df["Steel_Type"]).reindex(FAULT_COLS)

print("\n[결함 유형 × 강종 교차표]")
display(steel_fault_table)

print("\n[결함 유형별 강종 비율]")
display(
    pd.crosstab(df[TARGET], df["Steel_Type"], normalize="index")
    .reindex(FAULT_COLS)
    .round(3)
)

chi2, p_value, dof, expected = chi2_contingency(steel_fault_table)
cramers_v = np.sqrt(chi2 / (len(df) * (min(steel_fault_table.shape) - 1)))

print("\n[카이제곱 검정]")
print(f"Chi2 = {chi2:.2f}, 자유도 = {dof}, p-value = "
      f"{'<0.001' if p_value < 0.001 else round(p_value, 4)}")
print(f"Cramér's V = {cramers_v:.3f}")

steel_fault_table.div(steel_fault_table.sum(axis=1), axis=0).plot(
    kind="bar", stacked=True, figsize=(9, 4)
)
plt.title("결함 유형별 강종 구성")
plt.xlabel("결함 유형")
plt.ylabel("비율")
plt.xticks(rotation=30)
plt.legend(title="강종")
plt.tight_layout()
plt.show()


# ============================================================
# 7. 모델 입력 변수 생성
# ============================================================
# - Steel_Type: A300 → 0, A400 → 1 (한 열 유지)
# - 기존 파생 변수 6개: 가로/세로 길이, 가로세로비, 밝기 범위,
#                      평균 밝기, 면적 대비 둘레
# - [v2 추가] 파생 변수 6개
#     X_Min_Rel / X_Max_Rel : 컨베이어 폭 대비 결함의 가로 위치
#     Edge_Dist / _Rel      : 가까운 강판 가장자리까지의 거리
#     Lum_Contrast          : 평균 밝기 대비 밝기 범위
#     Thick_x_Steel         : 두께와 강종의 상호작용
# ============================================================

def build_features(data: pd.DataFrame) -> pd.DataFrame:

    out = data.drop(columns=[TARGET], errors="ignore").copy()

    out["Steel_Type"] = out["Steel_Type"].map(STEEL_CODE)

    if out["Steel_Type"].isnull().any():
        raise ValueError("Steel_Type에 A300 / A400 이외의 값이 있습니다.")

    # 기존 파생 변수
    out["X_Range"] = out["X_Maximum"] - out["X_Minimum"]
    out["Y_Range"] = out["Y_Maximum"] - out["Y_Minimum"]
    out["Aspect_Ratio"] = (out["X_Range"] + 1) / (out["Y_Range"] + 1)
    out["Luminosity_Range"] = (
        out["Maximum_of_Luminosity"] - out["Minimum_of_Luminosity"]
    )
    out["Mean_Luminosity"] = out["Sum_of_Luminosity"] / out["Pixels_Areas"]
    out["Perimeter_Area_Ratio"] = (
        (out["X_Perimeter"] + out["Y_Perimeter"]) / out["Pixels_Areas"]
    )

    # v2 추가 파생 변수
    conveyer = out["Length_of_Conveyer"]
    out["X_Min_Rel"] = out["X_Minimum"] / conveyer
    out["X_Max_Rel"] = out["X_Maximum"] / conveyer
    out["Edge_Dist"] = np.minimum(
        out["X_Minimum"], conveyer - out["X_Maximum"]
    ).clip(lower=0)
    out["Edge_Dist_Rel"] = out["Edge_Dist"] / conveyer
    out["Lum_Contrast"] = out["Luminosity_Range"] / (out["Mean_Luminosity"] + 1)
    out["Thick_x_Steel"] = out["Steel_Plate_Thickness"] * (1 + out["Steel_Type"])

    return out


X_all = build_features(df)
y_all = df[TARGET].map(class_mapping)

feature_cols = X_all.columns.tolist()

print("\n입력 변수 수:", len(feature_cols),
      "(원본 25 + Steel_Type 1 + 기존 파생 6 + 추가 파생 6)")
print(feature_cols)

# 로지스틱 회귀에서만 로그 변환할 치우친(0 이상) 변수
SKEWED_COLS = [
    "Pixels_Areas",
    "X_Perimeter",
    "Y_Perimeter",
    "Sum_of_Luminosity",
    "X_Range",
    "Y_Range",
    "Aspect_Ratio",
    "Perimeter_Area_Ratio",
    "Edge_Dist",
]


# ============================================================
# 8. 수치 변수 - 결함 유형 연관성 (Kruskal-Wallis)
# ============================================================
# 수치 변수는 정규분포가 아니므로 비모수 검정을 사용한다.
# 표본이 커서 p-value는 대부분 매우 작으므로
# 효과크기 ε²로 크기를 비교한다.
# ============================================================

n_obs, n_class = len(df), len(FAULT_COLS)
kw_results = []

for col in feature_cols:

    if col == "Steel_Type":
        continue

    groups = [X_all.loc[y_all == k, col] for k in LABELS]
    h_stat, p_val = kruskal(*groups)

    kw_results.append({
        "Feature": col,
        "H": h_stat,
        "p-value": "<0.001" if p_val < 0.001 else round(p_val, 4),
        "Epsilon²": max(0.0, (h_stat - n_class + 1) / (n_obs - n_class)),
    })

kw_df = (
    pd.DataFrame(kw_results)
    .sort_values("Epsilon²", ascending=False)
    .reset_index(drop=True)
)

print("\n[수치 변수 Kruskal-Wallis 검정 결과]")
display(kw_df.round(4))


# ============================================================
# 9. 학습 / 검증 / 테스트 분할 (60 / 20 / 20, 층화)
# ============================================================
# 1차: 전체에서 테스트 20% 분리
# 2차: 남은 80%에서 25%를 검증으로 분리 (전체 기준 20%)
# 두 단계 모두 stratify를 걸어 결함 유형 비율을 유지한다.
# ============================================================

X_temp, X_test, y_temp, y_test = train_test_split(
    X_all, y_all,
    test_size=0.20,
    stratify=y_all,
    random_state=RANDOM_STATE,
)

X_train, X_val, y_train, y_val = train_test_split(
    X_temp, y_temp,
    test_size=0.25,
    stratify=y_temp,
    random_state=RANDOM_STATE,
)

print("\n[데이터 분할]")
print(f"학습  : {len(X_train):>5}개 ({len(X_train) / len(X_all):.0%})")
print(f"검증  : {len(X_val):>5}개 ({len(X_val) / len(X_all):.0%})")
print(f"테스트: {len(X_test):>5}개 ({len(X_test) / len(X_all):.0%})")

split_table = pd.DataFrame({
    "Train": y_train.value_counts().sort_index(),
    "Val": y_val.value_counts().sort_index(),
    "Test": y_test.value_counts().sort_index(),
}).rename(index=reverse_class_mapping)

split_ratio = split_table.div(split_table.sum(axis=0), axis=1).mul(100).round(2)
split_ratio.columns = [f"{c}(%)" for c in split_ratio.columns]

print("\n[분할별 결함 유형 개수 / 비율]")
display(pd.concat([split_table, split_ratio], axis=1))

# 분할 결과를 원본 행 번호와 함께 저장 (재현성 확인용)
split_info = pd.Series("train", index=X_all.index)
split_info[X_val.index] = "val"
split_info[X_test.index] = "test"
df.assign(Split=split_info).to_csv("data_split_info.csv", index_label="row_id")


# 학습+검증(Dev): 5-Fold CV에 사용
X_dev = pd.concat([X_train, X_val])
y_dev = pd.concat([y_train, y_val])

print(f"\n학습+검증(Dev): {len(X_dev)}개 → 5-Fold CV 및 테스트 전 재학습에 사용")


# ============================================================
# 10. 모델 정의
# ============================================================
# LR  : 기준선 (로그 변환 + 표준화)
# RF  : 이전 버전 선택 모델 (이전 탐색에서 찾은 파라미터)
# ET  : ExtraTrees - 12단계에서 CV로 탐색
# LGBM/XGB: 트리 수는 11단계에서 검증 데이터 조기 종료로 결정
# ============================================================

N_JOBS = -1

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)


def make_lr():

    other_cols = [c for c in feature_cols if c not in SKEWED_COLS]
    preprocessor = ColumnTransformer([
        ("log", FunctionTransformer(np.log1p, feature_names_out="one-to-one"),
         SKEWED_COLS),
        ("pass", "passthrough", other_cols),
    ])

    return Pipeline([
        ("preprocess", preprocessor),
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(
            C=10.0, class_weight="balanced",
            max_iter=5000, random_state=RANDOM_STATE,
        )),
    ])


def make_rf():
    return RandomForestClassifier(
        n_estimators=300, max_depth=20, max_features=0.3,
        min_samples_split=9, min_samples_leaf=1,
        class_weight="balanced_subsample",
        random_state=RANDOM_STATE, n_jobs=N_JOBS,
    )


def make_et(**params):
    base = dict(
        n_estimators=500, max_features=0.5,
        class_weight="balanced",
        random_state=RANDOM_STATE, n_jobs=N_JOBS,
    )
    base.update(params)
    return ExtraTreesClassifier(**base)


LGBM_PARAMS = dict(
    objective="multiclass",
    learning_rate=0.05,
    num_leaves=22,
    min_child_samples=30,
    subsample=0.7, subsample_freq=1,
    colsample_bytree=0.65,
    class_weight="balanced",
    random_state=RANDOM_STATE, n_jobs=N_JOBS, verbose=-1,
)

XGB_PARAMS = dict(
    learning_rate=0.05,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.7,
    random_state=RANDOM_STATE, n_jobs=N_JOBS,
)


def score_row(y_true, y_hat):
    return {
        "Accuracy": accuracy_score(y_true, y_hat),
        "Balanced Acc": balanced_accuracy_score(y_true, y_hat),
        "Precision": precision_score(y_true, y_hat, average="macro", zero_division=0),
        "Recall": recall_score(y_true, y_hat, average="macro", zero_division=0),
        "Macro F1": f1_score(y_true, y_hat, average="macro", zero_division=0),
    }


# ============================================================
# 11. [검증 데이터 역할 ①] 부스팅 모델 트리 수 결정
# ============================================================
# 학습 데이터로 학습하면서 검증 데이터 손실이 50회 연속
# 개선되지 않으면 멈추고, 그때까지의 최적 트리 수를 고정한다.
# ============================================================

print("\n" + "=" * 60)
print("부스팅 모델 조기 종료 (학습 → 검증)")
print("=" * 60)

lgbm_es = LGBMClassifier(n_estimators=3000, **LGBM_PARAMS)
lgbm_es.fit(
    X_train, y_train,
    eval_set=[(X_val, y_val)],
    eval_metric="multi_logloss",
    callbacks=[early_stopping(stopping_rounds=50, verbose=False)],
)
LGBM_N_TREES = int(lgbm_es.best_iteration_)

xgb_es = XGBClassifier(
    n_estimators=3000, eval_metric="mlogloss",
    early_stopping_rounds=50, **XGB_PARAMS,
)
xgb_es.fit(X_train, y_train, eval_set=[(X_val, y_val)], verbose=False)
XGB_N_TREES = int(xgb_es.best_iteration) + 1

print("LightGBM 트리 수:", LGBM_N_TREES)
print("XGBoost  트리 수:", XGB_N_TREES)

lgbm_curve = lgbm_es.evals_result_["valid_0"]["multi_logloss"]
xgb_curve = xgb_es.evals_result()["validation_0"]["mlogloss"]

fig, axes = plt.subplots(1, 2, figsize=(14, 4))
for ax, curve, n_best, title in [
    (axes[0], lgbm_curve, LGBM_N_TREES, "LightGBM"),
    (axes[1], xgb_curve, XGB_N_TREES, "XGBoost"),
]:
    ax.plot(range(1, len(curve) + 1), curve)
    ax.axvline(n_best, color="red", linestyle="--", label=f"최적 트리 수 ({n_best})")
    ax.set_xlabel("트리 수")
    ax.set_ylabel("검증 Log Loss")
    ax.set_title(f"{title} 검증 손실 곡선")
    ax.legend()
plt.tight_layout()
plt.show()


def make_lgbm():
    return LGBMClassifier(n_estimators=LGBM_N_TREES, **LGBM_PARAMS)


def make_xgb():
    return XGBClassifier(n_estimators=XGB_N_TREES, **XGB_PARAMS)


# ============================================================
# 12. ExtraTrees 하이퍼파라미터 탐색 (Dev 5-Fold CV)
# ============================================================

ET_SEARCH_ITER = 15

print("\n" + "=" * 60)
print(f"ExtraTrees 하이퍼파라미터 탐색 ({ET_SEARCH_ITER}개 조합 × 5-Fold)")
print("=" * 60)

start = time.time()

et_search = RandomizedSearchCV(
    estimator=make_et(),
    param_distributions={
        "n_estimators": randint(300, 1000),
        "max_features": uniform(0.2, 0.6),
        "min_samples_split": randint(2, 8),
        "min_samples_leaf": randint(1, 3),
        "max_depth": [None, 20, 30],
    },
    n_iter=ET_SEARCH_ITER,
    scoring="f1_macro",
    cv=cv,
    n_jobs=1,                  # 모델 내부에서 병렬 처리
    random_state=RANDOM_STATE,
    refit=False,
)
et_search.fit(X_dev, y_dev)

ET_BEST_PARAMS = {
    k: (v.item() if hasattr(v, "item") else v)
    for k, v in et_search.best_params_.items()
}

print(f"소요 시간: {time.time() - start:.1f}초")
print("최적 CV Macro F1:", round(et_search.best_score_, 4))
print("최적 파라미터:", ET_BEST_PARAMS)


def make_et_tuned():
    return make_et(**ET_BEST_PARAMS)


model_makers = {
    "Logistic Regression": make_lr,
    "Random Forest": make_rf,
    "ExtraTrees": make_et_tuned,
    "LightGBM": make_lgbm,
    "XGBoost": make_xgb,
}


# ============================================================
# 13. 후보 모델 5-Fold CV (Dev) - OOF 예측 확률 저장
# ============================================================
# OOF(Out-Of-Fold) 확률: 각 행을 그 행이 빠진 폴드의 모델로 예측한 확률.
# 이를 저장해 두면 앙상블 조합을 다시 학습하지 않고 평가할 수 있다.
# ============================================================

def oof_predict(make_model):

    oof = np.zeros((len(X_dev), len(LABELS)))
    fold_scores = []

    for tr_idx, va_idx in cv.split(X_dev, y_dev):

        model = make_model()
        model.fit(X_dev.iloc[tr_idx], y_dev.iloc[tr_idx])

        proba = model.predict_proba(X_dev.iloc[va_idx])
        oof[va_idx] = proba
        fold_scores.append(
            f1_score(y_dev.iloc[va_idx], proba.argmax(axis=1), average="macro")
        )

    return oof, np.array(fold_scores)


print("\n" + "=" * 60)
print("후보 모델 5-Fold CV (학습+검증)")
print("=" * 60)

oof_probs, fold_f1 = {}, {}

for name, maker in model_makers.items():

    start = time.time()
    oof_probs[name], fold_f1[name] = oof_predict(maker)
    print(f"{name:<20} Macro F1 = {fold_f1[name].mean():.4f} "
          f"± {fold_f1[name].std():.4f}  ({time.time() - start:.1f}초)")


# ============================================================
# 14. 소프트 보팅 앙상블 조합 탐색
# ============================================================
# 트리 모델 4개(RF, ET, LGBM, XGB)의 2~4개 조합에 대해
# OOF 확률을 평균한 뒤 Macro F1을 계산한다.
# 단일 모델과 앙상블을 한 표에서 비교해 CV 최고를 선택한다.
# ============================================================

def fold_scores_from_oof(proba):
    return np.array([
        f1_score(y_dev.iloc[va], proba[va].argmax(axis=1), average="macro")
        for _, va in cv.split(X_dev, y_dev)
    ])


candidate_rows = []

for name in model_makers:
    candidate_rows.append({
        "Candidate": name,
        "Members": (name,),
        "CV Macro F1": fold_f1[name].mean(),
        "CV SD": fold_f1[name].std(),
        "CV Accuracy": accuracy_score(y_dev, oof_probs[name].argmax(axis=1)),
    })

TREE_MODELS = ["Random Forest", "ExtraTrees", "LightGBM", "XGBoost"]

for r in range(2, len(TREE_MODELS) + 1):
    for combo in itertools.combinations(TREE_MODELS, r):

        avg_proba = np.mean([oof_probs[m] for m in combo], axis=0)
        scores = fold_scores_from_oof(avg_proba)

        candidate_rows.append({
            "Candidate": " + ".join(combo),
            "Members": combo,
            "CV Macro F1": scores.mean(),
            "CV SD": scores.std(),
            "CV Accuracy": accuracy_score(y_dev, avg_proba.argmax(axis=1)),
        })

candidate_df = (
    pd.DataFrame(candidate_rows)
    .sort_values("CV Macro F1", ascending=False)
    .reset_index(drop=True)
)

print("\n[단일 모델 + 앙상블 CV 비교]")
display(candidate_df.drop(columns="Members").round(4))

plot_df = candidate_df.iloc[::-1]
plt.figure(figsize=(9, 7))
plt.barh(plot_df["Candidate"], plot_df["CV Macro F1"], xerr=plot_df["CV SD"])
plt.xlabel("5-Fold CV Macro F1")
plt.title("단일 모델 및 앙상블 조합 비교")
plt.xlim(max(0, candidate_df["CV Macro F1"].min() - 0.05), 1.0)
plt.tight_layout()
plt.show()


# ============================================================
# 15. 최종 모델 선택 (CV Macro F1 기준)
# ============================================================

best_row = candidate_df.iloc[0]
best_name = best_row["Candidate"]
best_members = list(best_row["Members"])


def make_final_model():

    if len(best_members) == 1:
        return model_makers[best_members[0]]()

    return VotingClassifier(
        estimators=[(m, model_makers[m]()) for m in best_members],
        voting="soft",
        n_jobs=1,
    )


print("\n" + "=" * 60)
print("최종 모델 선정 (5-Fold CV 기준)")
print("=" * 60)
print("선택:", best_name)
print(f"CV Macro F1: {best_row['CV Macro F1']:.4f} ± {best_row['CV SD']:.4f}")


# ============================================================
# 16. [검증 데이터 역할 ②] 학습 → 검증 홀드아웃 확인
# ============================================================
# CV 결과가 한 번의 고정 분할에서도 비슷하게 나오는지 확인한다.
# 선택을 바꾸는 용도가 아니라 2차 점검용이다.
# ============================================================

holdout_rows = []

for name in list(model_makers) + ([best_name] if len(best_members) > 1 else []):

    if name == best_name and len(best_members) > 1:
        model = make_final_model()
    else:
        model = model_makers[name]()

    model.fit(X_train, y_train)
    holdout_rows.append({
        "Model": name,
        "Val Macro F1": f1_score(y_val, model.predict(X_val), average="macro"),
    })

holdout_df = pd.DataFrame(holdout_rows)
holdout_df = holdout_df.merge(
    candidate_df[["Candidate", "CV Macro F1"]].rename(columns={"Candidate": "Model"}),
    on="Model",
)

print("\n[CV vs 학습→검증 홀드아웃]")
display(holdout_df.sort_values("CV Macro F1", ascending=False).round(4))


# ============================================================
# 17. 학습+검증으로 재학습 → 테스트 평가 (마지막 1회)
# ============================================================

start = time.time()
best_model = make_final_model()
best_model.fit(X_dev, y_dev)
print(f"\n최종 모델 학습 완료 (학습+검증 {len(X_dev)}개, {time.time() - start:.1f}초)")

y_pred = best_model.predict(X_test)
test_scores = score_row(y_test, y_pred)

print("\n" + "=" * 60)
print(f"최종 테스트 결과 - {best_name}")
print("=" * 60)
for metric, value in test_scores.items():
    print(f"{metric}: {value:.4f}")

# 참고: 이전 버전 방식 (RF를 학습 60%로만 학습)
prev_model = make_rf().fit(X_train, y_train)
prev_f1 = f1_score(y_test, prev_model.predict(X_test), average="macro")

print(f"\n[참고] RF, 학습 데이터만 사용 시 테스트 Macro F1: {prev_f1:.4f}")
print(f"[참고] 최종 모델과의 차이: {test_scores['Macro F1'] - prev_f1:+.4f}")


# ============================================================
# 18. Classification Report / 결함 유형별 성능
# ============================================================

print("\n[Classification Report - Test]")
print(classification_report(
    y_test, y_pred,
    labels=LABELS, target_names=FAULT_COLS,
    digits=4, zero_division=0,
))

class_report_df = (
    pd.DataFrame(classification_report(
        y_test, y_pred,
        labels=LABELS, target_names=FAULT_COLS,
        output_dict=True, zero_division=0,
    )).T
    .loc[FAULT_COLS, ["precision", "recall", "f1-score", "support"]]
    .sort_values("f1-score")
)

print("\n[결함 유형별 성능 (F1 낮은 순)]")
display(class_report_df.round(4))


# ============================================================
# 19. 혼동행렬
# ============================================================

cm = confusion_matrix(y_test, y_pred, labels=LABELS)
cm_norm = confusion_matrix(y_test, y_pred, labels=LABELS, normalize="true")

fig, axes = plt.subplots(1, 2, figsize=(18, 7))

ConfusionMatrixDisplay(cm, display_labels=FAULT_COLS).plot(
    ax=axes[0], xticks_rotation=45, colorbar=False
)
axes[0].set_title(f"혼동행렬 (개수) - {best_name}")

ConfusionMatrixDisplay(cm_norm, display_labels=FAULT_COLS).plot(
    ax=axes[1], xticks_rotation=45, colorbar=False, values_format=".2f"
)
axes[1].set_title("혼동행렬 (실제 유형 기준 비율)")

for ax in axes:
    ax.set_xlabel("예측 결함 유형")
    ax.set_ylabel("실제 결함 유형")

plt.tight_layout()
plt.show()


# ============================================================
# 20. 오분류 분석
# ============================================================

test_result_df = pd.DataFrame({
    "Actual": y_test.map(reverse_class_mapping),
    "Predicted": pd.Series(y_pred, index=y_test.index).map(reverse_class_mapping),
    "Confidence": best_model.predict_proba(X_test).max(axis=1),
})

misclassified = test_result_df[test_result_df["Actual"] != test_result_df["Predicted"]]

print("\n" + "=" * 60)
print("오분류 분석 (테스트 데이터)")
print("=" * 60)
print("테스트 데이터:", len(y_test), "개")
print("오분류:", len(misclassified), "개", f"({len(misclassified) / len(y_test):.2%})")

misclassification_summary = (
    misclassified
    .groupby(["Actual", "Predicted"])
    .size()
    .reset_index(name="Count")
    .sort_values("Count", ascending=False)
    .reset_index(drop=True)
)

print("\n[주요 오분류 조합]")
display(misclassification_summary.head(10))

other_related = misclassified.query(
    "Actual == 'Other_Faults' or Predicted == 'Other_Faults'"
)
print("Other_Faults가 관련된 오분류 비중:",
      f"{len(other_related) / max(len(misclassified), 1):.1%}")

misclassified_rows = df.loc[misclassified.index].join(
    misclassified[["Predicted", "Confidence"]]
)

print("\n[오분류 데이터 행]")
display(misclassified_rows.head(20))


# ============================================================
# 21. 변수 중요도 (Permutation Importance, 테스트 데이터)
# ============================================================
# 앙상블에도 적용할 수 있는 모델 무관 방식.
# 상관이 높은 변수끼리는 중요도가 나뉘어 낮게 나올 수 있다.
# ============================================================

print("\nPermutation Importance 계산 중...")

perm = permutation_importance(
    best_model, X_test, y_test,
    scoring="f1_macro",
    n_repeats=10,
    random_state=RANDOM_STATE,
    n_jobs=1,
)

importance_df = (
    pd.DataFrame({
        "Feature": feature_cols,
        "Importance Mean": perm.importances_mean,
        "Importance SD": perm.importances_std,
    })
    .sort_values("Importance Mean", ascending=False)
    .reset_index(drop=True)
)
importance_df.insert(0, "Rank", range(1, len(importance_df) + 1))

NEW_FEATURES = [
    "X_Min_Rel", "X_Max_Rel", "Edge_Dist",
    "Edge_Dist_Rel", "Lum_Contrast", "Thick_x_Steel",
]
importance_df["v2 추가 변수"] = importance_df["Feature"].isin(NEW_FEATURES)

print("\n[변수 중요도]")
display(importance_df.round(4))

plot_df = importance_df.head(15).iloc[::-1]
colors = ["tab:orange" if f else "tab:blue" for f in plot_df["v2 추가 변수"]]

plt.figure(figsize=(9, 7))
plt.barh(plot_df["Feature"], plot_df["Importance Mean"],
         xerr=plot_df["Importance SD"], color=colors)
plt.xlabel("섞었을 때 Macro F1 감소량")
plt.title(f"Permutation Importance 상위 15개 (주황: v2 추가 변수)")
plt.tight_layout()
plt.show()


# ============================================================
# 22. 최종 분석 요약
# ============================================================

print("\n" + "=" * 60)
print("최종 분석 요약")
print("=" * 60)
print("전체 데이터:", len(df), "개")
print(f"분할: 학습 {len(X_train)} / 검증 {len(X_val)} / 테스트 {len(X_test)}")
print("모델 입력 변수:", len(feature_cols), "개")
print(f"강종-결함 유형 연관성: Cramér's V = {cramers_v:.3f}")
print(f"부스팅 트리 수 (검증 조기 종료): LightGBM {LGBM_N_TREES}, XGBoost {XGB_N_TREES}")
print("ExtraTrees 파라미터:", ET_BEST_PARAMS)
print("\n최종 모델:", best_name)
print(f"CV Macro F1: {best_row['CV Macro F1']:.4f} ± {best_row['CV SD']:.4f}")
print(f"테스트 Macro F1: {test_scores['Macro F1']:.4f}")
print(f"테스트 Accuracy: {test_scores['Accuracy']:.4f}")
print(f"RF(학습 데이터만) 대비: {test_scores['Macro F1'] - prev_f1:+.4f}")
print("\n가장 어려운 결함 유형:", class_report_df.index[0],
      f"(F1 {class_report_df['f1-score'].iloc[0]:.3f})")
print("가장 쉬운 결함 유형:", class_report_df.index[-1],
      f"(F1 {class_report_df['f1-score'].iloc[-1]:.3f})")
print("중요 변수 Top 5:", importance_df["Feature"].head(5).tolist())


# ============================================================
# 23. 최종 모델을 전체 데이터로 재학습 및 저장
# ============================================================
# 평가가 끝났으므로 학습+검증+테스트 전체(1,941개)로 다시 학습한다.
# 부스팅 모델의 트리 수는 11단계에서 정한 값을 그대로 사용한다.
# ============================================================

final_model = make_final_model()
final_model.fit(X_all, y_all)

joblib.dump(
    {
        "model": final_model,
        "model_name": best_name,
        "members": best_members,
        "feature_cols": feature_cols,
        "class_names": FAULT_COLS,
    },
    "steel_fault_model_v2.joblib",
)

print("\n최종 모델 저장 완료: steel_fault_model_v2.joblib")


# ============================================================
# 24. 새로운 데이터 예측 함수
# ============================================================
# 원본 형식(TypeOfSteel_A300/A400 두 열) 또는
# 합친 형식(Steel_Type 한 열) 둘 다 입력 가능.
#
# 사용 예:
#   new_df = pd.read_csv("new_plates.csv")
#   pred = predict_faults(new_df)
# ============================================================

def predict_faults(new_data: pd.DataFrame,
                   model_path: str = "steel_fault_model_v2.joblib") -> pd.DataFrame:

    bundle = joblib.load(model_path)
    model, names = bundle["model"], bundle["class_names"]

    merged = merge_columns(new_data.drop(columns=names, errors="ignore"))
    X_new = build_features(merged)[bundle["feature_cols"]]

    proba = model.predict_proba(X_new)

    output = pd.DataFrame(
        proba, columns=[f"prob_{n}" for n in names], index=new_data.index
    )
    output.insert(0, "predicted_fault", [names[i] for i in proba.argmax(axis=1)])
    output.insert(1, "confidence", proba.max(axis=1))

    return output


# 동작 확인 (전체 데이터로 재학습한 모델이므로 성능 확인용이 아님)
demo = predict_faults(raw_df.loc[X_test.index[:5]])
demo.insert(0, "actual_fault", df.loc[X_test.index[:5], TARGET].values)

print("\n[예측 함수 동작 확인]")
display(demo.round(3))
```
