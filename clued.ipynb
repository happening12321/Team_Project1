```python
# %% [markdown]
# # 강판 결함 유형 분류 — EDA 코드
# 보고서 「강판 결함 유형 분류 — 탐색적 데이터 분석(EDA) 보고서」의 수치와 그래프를 재현합니다.
# 섹션 번호는 보고서 목차와 같습니다. 원본 UCI 파일(Faults.NNA, Faults27x7_var)에서 시작합니다.

# %% 0. 라이브러리 및 설정
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import font_manager
from scipy import stats
from sklearn.model_selection import train_test_split
from IPython.display import display

FIG_DIR = "figures"
RANDOM_STATE = 42
os.makedirs(FIG_DIR, exist_ok=True)

# 한글 폰트 (설치된 것 중 첫 번째 사용)
for f in ["Malgun Gothic", "AppleGothic", "NanumGothic", "Noto Sans CJK KR"]:
    if any(f in x.name for x in font_manager.fontManager.ttflist):
        plt.rcParams["font.family"] = f
        break
plt.rcParams["axes.unicode_minus"] = False
pd.set_option("display.max_columns", 40)

# %% 1. 원본 파일 → CSV 변환
# 1-1. 열 이름 읽기 (Windows 줄바꿈 \r\n 대비 strip 처리)
with open("Faults27x7_var", "r") as f:
    cols = [line.strip() for line in f if line.strip()]

# 1-2. 데이터 읽기 (탭/공백 구분, 헤더 없음)
data = pd.read_csv("Faults.NNA", sep=r"\s+", header=None, names=cols)

# 1-3. 확인
print(data.shape)                 # (1941, 34)
print(data.isna().sum().sum())    # 0

# 1-4. CSV로 저장 (엑셀에서 열어도 깨지지 않도록 utf-8-sig)
data.to_csv("steel_plates_faults.csv", index=False, encoding="utf-8-sig")

# %% 3. 데이터 구조 파악
df = pd.read_csv("steel_plates_faults.csv")
display(df.head())
display(df.shape)                 # (1941, 34)
display(df.isna().sum().sum())    # 0
df.info()
display(df.dtypes.value_counts()) # int64 21, float64 13

TARGET_COLS = ["Pastry", "Z_Scratch", "K_Scatch", "Stains",
               "Dirtiness", "Bumps", "Other_Faults"]
FEATURES = [c for c in df.columns if c not in TARGET_COLS]
print("입력 변수 수:", len(FEATURES))   # 27

# 그래프·표 표시 순서: 건수 내림차순
CLASS_ORDER = ["Other_Faults", "Bumps", "K_Scatch", "Z_Scratch",
               "Pastry", "Stains", "Dirtiness"]

# %% 4. 컬럼정의서용 요약 (자료형·범위·고유값 수·왜도)
col_summary = pd.DataFrame({
    "dtype": df[FEATURES].dtypes.astype(str),
    "min": df[FEATURES].min(),
    "max": df[FEATURES].max(),
    "nunique": df[FEATURES].nunique(),
    "skew": df[FEATURES].skew().round(3),
})
display(col_summary)

# %% 5-1. 자료형 점검 — 정답 열 검증 및 Target 생성
is_binary = df[TARGET_COLS].isin([0, 1]).all().all()
row_sum = df[TARGET_COLS].sum(axis=1)
print("정답 열 0/1 여부:", is_binary)                         # True
print("행별 정답 합 분포:", row_sum.value_counts().to_dict())  # {1: 1941}
assert is_binary and (row_sum == 1).all(), "정답 열 무결성 오류"

# 값이 1인 정답 열의 이름 → 단일 Target
df["fault_type"] = df[TARGET_COLS].idxmax(axis=1)

# 강종 두 열의 완전 보완 관계
print("A300 + A400 분포:", (df["TypeOfSteel_A300"] + df["TypeOfSteel_A400"]).value_counts().to_dict())  # {1: 1941}
print("A300–A400 상관:", df["TypeOfSteel_A300"].corr(df["TypeOfSteel_A400"]))                      # -1.0

# 이산값 변수
print("두께 고유값 수:", df["Steel_Plate_Thickness"].nunique())                      # 24
print("Outside_Global_Index 값:", df["Outside_Global_Index"].value_counts().to_dict())

# 논리 점검 (모두 0건이어야 함)
logic_checks = pd.Series({
    "X_Minimum > X_Maximum": (df.X_Minimum > df.X_Maximum).sum(),
    "Y_Minimum > Y_Maximum": (df.Y_Minimum > df.Y_Maximum).sum(),
    "최소 밝기 > 최대 밝기": (df.Minimum_of_Luminosity > df.Maximum_of_Luminosity).sum(),
    "X_Maximum > 컨베이어 길이": (df.X_Maximum > df.Length_of_Conveyer).sum(),
}, name="위반 건수")
display(logic_checks)

# %% 5-2. 결측치 점검
missing = df[cols].isna().sum()
display(missing[missing > 0] if missing.sum() else "결측 없음")

# %% 6-1. 중복 데이터 확인
dup_all = df[cols].duplicated().sum()
dup_x = df[FEATURES].duplicated().sum()
conflict = (df.groupby(FEATURES)["fault_type"].nunique() > 1).sum()   # 입력 같고 정답 다른 경우
print(f"전체 열 완전 중복: {dup_all} / 입력 27열 중복: {dup_x} / 라벨 충돌: {conflict}")

# %% 6-2. 이상치 확인 (IQR 1.5배) 및 유형별 구성
q1 = df[FEATURES].quantile(0.25)
q3 = df[FEATURES].quantile(0.75)
iqr = q3 - q1
outlier_mask = (df[FEATURES] < q1 - 1.5 * iqr) | (df[FEATURES] > q3 + 1.5 * iqr)

outlier_table = pd.DataFrame({
    "이상치 건수": outlier_mask.sum(),
    "이상치 비율(%)": (outlier_mask.mean() * 100).round(2),
})
top_cls, top_share = [], []
for c in FEATURES:
    vc = df.loc[outlier_mask[c], "fault_type"].value_counts()
    top_cls.append(vc.index[0] if len(vc) else "-")
    top_share.append(round(vc.iloc[0] / vc.sum() * 100, 1) if len(vc) else 0)
outlier_table["최다 유형"] = top_cls
outlier_table["최다 유형 비중(%)"] = top_share
outlier_table = outlier_table.sort_values("이상치 건수", ascending=False)
display(outlier_table[outlier_table["이상치 건수"] > 0])

any_outlier = outlier_mask.any(axis=1)
print(f"한 열 이상 이상치를 가진 행: {any_outlier.sum()} ({any_outlier.mean() * 100:.1f}%)")  # 824 (42.5%)
display(df.loc[any_outlier, "fault_type"].value_counts().reindex(CLASS_ORDER, fill_value=0))
display(df.nlargest(3, "Pixels_Areas")[["Pixels_Areas", "fault_type"]])

# %% 6-3. 전처리 결과 — 분석용 데이터셋
# 정답 7열 제외(누수 방지), 강종 중복 열 1개 제외. 이상치는 유형 고유 특징이므로 유지.
DROP_COLS = TARGET_COLS + ["TypeOfSteel_A400"]
X = df.drop(columns=DROP_COLS + ["fault_type"])
y = df["fault_type"]
print("전처리 후:", X.shape[0], "행 /", X.shape[1], "입력 + Target 1")   # 1941 행 / 26 입력

# %% 7-1. 범주형 그래프 — 클래스 분포 및 층화 분할 건수
cls_cnt = y.value_counts().reindex(CLASS_ORDER)
cls_pct = (cls_cnt / len(y) * 100).round(2)
display(pd.DataFrame({"건수": cls_cnt, "비율(%)": cls_pct}))
print("최대/최소 비:", round(cls_cnt.max() / cls_cnt.min(), 2))   # 12.24

# Train/Validation/Test = 60/20/20 층화 분할 (모든 모델에 동일하게 사용)
X_tmp, X_test, y_tmp, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=RANDOM_STATE)
X_train, X_val, y_train, y_val = train_test_split(
    X_tmp, y_tmp, test_size=0.25, stratify=y_tmp, random_state=RANDOM_STATE)
split_cnt = pd.DataFrame({
    "train": y_train.value_counts(),
    "val": y_val.value_counts(),
    "test": y_test.value_counts(),
}).reindex(CLASS_ORDER)
display(split_cnt)
print("합계:", len(y_train), len(y_val), len(y_test))   # 1164 388 389

fig, ax = plt.subplots(figsize=(9, 4.5))
colors = ["#2a78d6" if p < 5 else "#c2c2b8" for p in cls_pct]
ax.barh(CLASS_ORDER[::-1], cls_cnt[::-1], color=colors[::-1])
eq = len(y) / len(CLASS_ORDER)
ax.axvline(eq, ls="--", c="gray", lw=1)
ax.text(eq, len(CLASS_ORDER) - 0.4, f"균등 분포 시 {eq:.0f}건", ha="center", color="gray")
for i, cls in enumerate(CLASS_ORDER[::-1]):
    ax.text(cls_cnt[cls] + 8, i,
            f"{cls_cnt[cls]}건 · {cls_pct[cls]:.1f}% (test {split_cnt.loc[cls, 'test']})", va="center")
ax.set_title(f"결함 유형별 건수: 최다 유형이 최소 유형의 {cls_cnt.max() / cls_cnt.min():.1f}배")
ax.set_xlim(0, cls_cnt.max() * 1.35)
ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
plt.savefig(f"{FIG_DIR}/7-1_class_distribution.png", dpi=150)
plt.show()

# %% 7-1. 강종(A300) 비율 — 유형별
steel_ct = pd.crosstab(df["fault_type"], df["TypeOfSteel_A300"]).reindex(CLASS_ORDER)
steel_ct.columns = ["A400", "A300"]
steel_ct["A300 비율(%)"] = (steel_ct["A300"] / (steel_ct["A300"] + steel_ct["A400"]) * 100).round(1)
display(steel_ct)
overall_a300 = df["TypeOfSteel_A300"].mean() * 100   # 40.0%

s = steel_ct["A300 비율(%)"].sort_values()
fig, ax = plt.subplots(figsize=(9, 4.5))
ax.barh(s.index, 100, color="#ececec")
ax.barh(s.index, s.values, color=["#2a78d6" if (v > 85 or v < 5) else "#c2c2b8" for v in s.values])
ax.axvline(overall_a300, ls="--", c="gray", lw=1)
ax.text(overall_a300, len(s) - 0.4, f"전체 {overall_a300:.1f}%", ha="center", color="gray")
for i, cls in enumerate(s.index):
    tot = steel_ct.loc[cls, "A300"] + steel_ct.loc[cls, "A400"]
    ax.text(102, i, f"{s[cls]:.1f}% ({steel_ct.loc[cls, 'A300']}/{tot})", va="center")
ax.set_xlim(0, 135)
ax.set_title("강종 A300 비율: Z_Scratch 90.5%, K_Scatch·Stains 2% 미만")
ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
plt.savefig(f"{FIG_DIR}/7-1_steel_type_share.png", dpi=150)
plt.show()

# %% 7-2. 수치형 — 유형 구분력 (Kruskal-Wallis 효과크기 ε²)
k, n = len(CLASS_ORDER), len(df)
rows = []
for c in FEATURES:
    groups = [df.loc[df.fault_type == cls, c] for cls in CLASS_ORDER]
    H, p = stats.kruskal(*groups)
    rows.append({"변수": c, "H": round(H, 1), "p": p, "eps2": round((H - k + 1) / (n - k), 3)})
eps_df = pd.DataFrame(rows).sort_values("eps2", ascending=False).reset_index(drop=True)
display(eps_df)

fig, ax = plt.subplots(figsize=(9, 9))
e = eps_df.iloc[::-1]
ax.barh(e["변수"], e["eps2"], color=["#2a78d6" if v >= 0.4 else "#c2c2b8" for v in e["eps2"]])
for i, v in enumerate(e["eps2"]):
    ax.text(v + 0.004, i, f"{v:.3f}", va="center", fontsize=9)
ax.set_title("유형 구분력 상위: 밝기 합계·X 외곽 지수·면적 계열 (ε² 0.40 이상)")
ax.set_xlabel("Kruskal-Wallis ε²")
ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
plt.savefig(f"{FIG_DIR}/7-2_feature_eps2.png", dpi=150)
plt.show()

# %% 7-2. 유형별 주요 변수 중앙값 표 및 박스플롯
KEY_VARS = ["Pixels_Areas", "X_Perimeter", "X_Minimum", "Minimum_of_Luminosity",
            "Maximum_of_Luminosity", "Steel_Plate_Thickness", "Edges_Index",
            "Square_Index", "Orientation_Index", "Length_of_Conveyer"]
display(df.groupby("fault_type")[KEY_VARS].median().reindex(CLASS_ORDER).round(3))

fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for ax, c in zip(axes.ravel(), ["Pixels_Areas", "Sum_of_Luminosity", "X_Minimum",
                                "Minimum_of_Luminosity", "Edges_Index", "Orientation_Index"]):
    ax.boxplot([df.loc[df.fault_type == cls, c] for cls in CLASS_ORDER], showfliers=False)
    ax.set_xticks(range(1, k + 1), CLASS_ORDER, rotation=30, fontsize=8)
    if c in ["Pixels_Areas", "Sum_of_Luminosity"]:
        ax.set_yscale("log")   # 치우친 크기 변수는 로그 축
    ax.set_title(c)
plt.suptitle("유형별 주요 변수 분포 (이상치 표시 생략)")
plt.tight_layout()
plt.savefig(f"{FIG_DIR}/7-2_boxplots_by_class.png", dpi=150)
plt.show()

# %% 7-3. 변수 간 상관관계 — |r| ≥ 0.9 쌍
corr = df[FEATURES].corr()
pairs = [(a, b, round(corr.loc[a, b], 3))
         for i, a in enumerate(FEATURES) for b in FEATURES[i + 1:]
         if abs(corr.loc[a, b]) >= 0.9]
high_corr = (pd.DataFrame(pairs, columns=["변수1", "변수2", "r"])
             .sort_values("r", key=abs, ascending=False).reset_index(drop=True))
display(high_corr)   # 7쌍

fig, ax = plt.subplots(figsize=(11, 9))
im = ax.imshow(corr, cmap="RdBu_r", vmin=-1, vmax=1)
ax.set_xticks(range(len(FEATURES)), FEATURES, rotation=90, fontsize=7)
ax.set_yticks(range(len(FEATURES)), FEATURES, fontsize=7)
plt.colorbar(im, ax=ax, shrink=0.8)
ax.set_title("입력 변수 상관행렬 (Pearson)")
plt.tight_layout()
plt.savefig(f"{FIG_DIR}/7-3_correlation_heatmap.png", dpi=150)
plt.show()

# 모델링 단계 변수 실험용: 중복 변수 제외 조건 (8-2)
REDUNDANT_DROP = ["Y_Maximum", "X_Maximum", "Sum_of_Luminosity"]   # 강종 A400은 이미 제외
X_reduced = X.drop(columns=REDUNDANT_DROP)
print("중복 제외 조건 입력 변수 수:", X_reduced.shape[1])   # 23

# %% 6-3 / 8. 전처리 결과 요약 및 분할 저장
summary = pd.DataFrame({
    "전처리 전": [len(data), data.shape[1], int(data.isna().sum().sum()), int(dup_all), int(any_outlier.sum())],
    "전처리 후": [len(X), X.shape[1] + 1, int(X.isna().sum().sum()), 0, 0],
}, index=["행 수", "컬럼 수", "결측치", "중복 제거 건수", "이상치 행 / 제거 건수"])
display(summary)

# 모델링 단계에서 동일 분할 재사용
for name, (Xs, ys) in {"train": (X_train, y_train), "val": (X_val, y_val),
                       "test": (X_test, y_test)}.items():
    Xs.assign(fault_type=ys).to_csv(f"split_{name}.csv", index=False, encoding="utf-8-sig")
print("저장 완료: split_train.csv / split_val.csv / split_test.csv, figures/*.png")
```
