```python
# ============================================================
# 강판 표면 결함 분류 - GPT 코드 + Claude 코드 장점 통합 버전
# ============================================================
# [GPT 코드에서 가져온 것]
#   - 하이퍼파라미터 튜닝을 단일 검증셋이 아닌 5-Fold 교차검증으로 수행
#   - class_weight를 고정하지 않고 튜닝 대상에 포함
#   - 여러 모델을 넓게 비교한 뒤 상위 모델 선별
#
# [Claude 코드에서 가져온 것]
#   - 원-핫 구조 검증 후 열 병합, 강종-결함 카이제곱 검정
#   - Permutation Importance, 오분류 조합 분석
#   - 전체 데이터 재학습 + 새 데이터 예측 함수
#
# [새로 추가한 것]
#   - 파생변수 사용 여부를 교차검증으로 결정 (중복 변수는 제외)
#   - 상위 모델 확률 평균(소프트 보팅) 앙상블.
#   - 클래스별 확률 보정: Out-of-fold 예측으로 Macro F1이 최대가 되도록
#     클래스별 배율을 찾아, 정밀도/재현율 균형을 데이터가 결정
#   - 테스트 Macro F1의 부트스트랩 95% 신뢰구간 (테스트셋 운 확인용)
#
# 분할: 개발(학습+교차검증) 80% / 테스트 20%
#   별도 검증셋 대신 교차검증을 쓰므로 학습에 쓰는 데이터가 60% → 80%로 늘어난다.
# ============================================================


# ============================================================
# 0. 라이브러리 및 설정
# ============================================================

import time
import warnings

import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

try:
    from IPython.display import display
except ImportError:
    display = print

from scipy.stats import chi2_contingency
from sklearn.base import BaseEstimator, ClassifierMixin, clone
from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    RepeatedStratifiedKFold,
    RandomizedSearchCV,
    cross_val_score,
    cross_val_predict,
)
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.inspection import permutation_importance
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
)
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier

warnings.filterwarnings("ignore")
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)

plt.rcParams["font.family"] = "Malgun Gothic"   # macOS: "AppleGothic"
plt.rcParams["axes.unicode_minus"] = False

RANDOM_STATE = 42
DATA_PATH = "steel_plates_faults.csv"
N_ITER = 20            # 모델별 무작위 탐색 조합 수 (시간이 부족하면 줄이기)

STEEL_COLS = ["TypeOfSteel_A300", "TypeOfSteel_A400"]
FAULT_COLS = ["Pastry", "Z_Scratch", "K_Scatch", "Stains",
              "Dirtiness", "Bumps", "Other_Faults"]
TARGET = "Fault_Type"

LABELS = list(range(len(FAULT_COLS)))
class_to_int = {name: i for i, name in enumerate(FAULT_COLS)}
int_to_class = {i: name for name, i in class_to_int.items()}
STEEL_CODE = {"A300": 0, "A400": 1}


def macro_f1(y_true, y_pred):
    return f1_score(y_true, y_pred, labels=LABELS, average="macro", zero_division=0)


def score_row(y_true, y_pred):
    return {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Balanced Acc": balanced_accuracy_score(y_true, y_pred),
        "Macro Precision": precision_score(y_true, y_pred, labels=LABELS,
                                           average="macro", zero_division=0),
        "Macro Recall": recall_score(y_true, y_pred, labels=LABELS,
                                     average="macro", zero_division=0),
        "Macro F1": macro_f1(y_true, y_pred),
    }


# ============================================================
# 1. 데이터 불러오기 및 열 병합 (원-핫 구조 검증 포함)
# ============================================================

raw_df = pd.read_csv(DATA_PATH)
print("원본 데이터 크기:", raw_df.shape)
print("결측치:", raw_df.isna().sum().sum(), "/ 중복 행:", raw_df.duplicated().sum())

for cols, label in [(STEEL_COLS, "강종"), (FAULT_COLS, "결함 유형")]:
    missing = [c for c in cols if c not in raw_df.columns]
    if missing:
        raise ValueError(f"{label} 열이 없습니다: {missing}")
    if not raw_df[cols].isin([0, 1]).all().all():
        raise ValueError(f"{label} 열은 0 또는 1이어야 합니다.")
    if not (raw_df[cols].sum(axis=1) == 1).all():
        raise ValueError(f"{label}이(가) 정확히 하나로 정해지지 않는 행이 있습니다.")


def merge_columns(data: pd.DataFrame) -> pd.DataFrame:
    """강종 2열 → Steel_Type, 결함 7열 → Fault_Type (결함 열이 없으면 강종만)"""
    out = data.copy()
    if set(STEEL_COLS).issubset(out.columns):
        pos = out.columns.get_loc("TypeOfSteel_A300")
        steel = np.where(out["TypeOfSteel_A300"] == 1, "A300", "A400")
        out = out.drop(columns=STEEL_COLS)
        out.insert(pos, "Steel_Type", steel)
    if set(FAULT_COLS).issubset(out.columns):
        out[TARGET] = out[FAULT_COLS].idxmax(axis=1)
        out = out.drop(columns=FAULT_COLS)
    return out


df = merge_columns(raw_df)
print("병합 후 데이터 크기:", df.shape)

print("\n[결함 유형 분포]")
display(df[TARGET].value_counts().reindex(FAULT_COLS).to_frame("count"))


# ============================================================
# 2. 강종 - 결함 유형 관계 (카이제곱 검정, Cramér's V)
# ============================================================

ct = pd.crosstab(df[TARGET], df["Steel_Type"]).reindex(FAULT_COLS)
chi2, p_value, dof, _ = chi2_contingency(ct)
cramers_v = np.sqrt(chi2 / (len(df) * (min(ct.shape) - 1)))

print("\n[결함 유형 × 강종]")
display(ct)
print(f"Chi2 = {chi2:.1f}, df = {dof}, p = "
      f"{'<0.001' if p_value < 0.001 else round(p_value, 4)}, Cramér's V = {cramers_v:.3f}")


# ============================================================
# 3. 입력 변수 생성
# ============================================================
# Claude 코드의 파생변수 중 원본과 단조 관계라 트리 모델에 정보가 없는
# X_Range(≈Log_X_Index), Y_Range(≈Log_Y_Index), Mean_Luminosity(≈Luminosity_Index)는 제외.
# 정보가 새로 생기는 3개만 후보로 두고, 쓸지 말지는 4단계 교차검증으로 결정한다.
# ============================================================

EXTRA_FEATURES = ["Luminosity_Range", "Aspect_Ratio", "Perimeter_Area_Ratio"]


def build_features(data: pd.DataFrame, use_extra: bool) -> pd.DataFrame:
    out = data.drop(columns=[TARGET], errors="ignore").copy()
    out["Steel_Type"] = out["Steel_Type"].map(STEEL_CODE)
    if out["Steel_Type"].isna().any():
        raise ValueError("Steel_Type에 A300 / A400 이외의 값이 있습니다.")

    if use_extra:
        x_len = out["X_Maximum"] - out["X_Minimum"] + 1
        y_len = out["Y_Maximum"] - out["Y_Minimum"] + 1
        out["Luminosity_Range"] = out["Maximum_of_Luminosity"] - out["Minimum_of_Luminosity"]
        out["Aspect_Ratio"] = x_len / y_len
        out["Perimeter_Area_Ratio"] = (
            (out["X_Perimeter"] + out["Y_Perimeter"]) / out["Pixels_Areas"]
        )
    return out


y_all = df[TARGET].map(class_to_int).to_numpy()


# ============================================================
# 4. 개발 / 테스트 분할 (80 / 20, 층화)
# ============================================================

idx_dev, idx_test = train_test_split(
    np.arange(len(df)), test_size=0.20, stratify=y_all, random_state=RANDOM_STATE
)
y_dev, y_test = y_all[idx_dev], y_all[idx_test]
print(f"\n개발: {len(idx_dev)}개 / 테스트: {len(idx_test)}개")

cv5 = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
cv_rep = RepeatedStratifiedKFold(n_splits=5, n_repeats=3, random_state=RANDOM_STATE)


# ============================================================
# 5. 파생변수 사용 여부 결정 (개발 데이터 반복 교차검증)
# ============================================================

probe = RandomForestClassifier(n_estimators=300, random_state=RANDOM_STATE, n_jobs=-1)
feature_check = {}
for use_extra in [False, True]:
    X_tmp = build_features(df, use_extra).iloc[idx_dev]
    s = cross_val_score(probe, X_tmp, y_dev, cv=cv_rep, scoring="f1_macro", n_jobs=-1)
    feature_check["원본 + 파생 3개" if use_extra else "원본만"] = (s.mean(), s.std())
    print(f"{'원본 + 파생 3개' if use_extra else '원본만':<12} "
          f"CV Macro F1 = {s.mean():.4f} ± {s.std():.4f}")

# 평균이 표준편차의 절반 이상 좋아질 때만 파생변수 채택 (작은 차이는 단순한 쪽 선택)
gain = feature_check["원본 + 파생 3개"][0] - feature_check["원본만"][0]
USE_EXTRA = gain > 0.5 * feature_check["원본만"][1]
print("→ 파생변수 사용:", USE_EXTRA)

X_all = build_features(df, USE_EXTRA)
feature_cols = X_all.columns.tolist()
X_dev, X_test = X_all.iloc[idx_dev], X_all.iloc[idx_test]


# ============================================================
# 6. 모델 및 탐색 공간
# ============================================================
# GPT 코드의 기본 비교에서 선형/거리 기반 모델(로지스틱, SVM, KNN)은
# 트리 계열보다 Macro F1이 0.07~0.17 낮았으므로 트리 계열만 사용한다.
# class_weight는 모든 모델에서 튜닝 대상이다.
# ============================================================

models = {
    "Random Forest": RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=1),
    "Extra Trees": ExtraTreesClassifier(random_state=RANDOM_STATE, n_jobs=1),
    "LightGBM": LGBMClassifier(objective="multiclass", random_state=RANDOM_STATE,
                               n_jobs=1, verbose=-1),
    "CatBoost": CatBoostClassifier(loss_function="MultiClass", random_seed=RANDOM_STATE,
                                   verbose=0, allow_writing_files=False, thread_count=1),
}

param_spaces = {
    "Random Forest": {
        "n_estimators": [300, 500, 700],
        "max_depth": [None, 10, 15, 20, 30],
        "min_samples_split": [2, 5, 9],
        "min_samples_leaf": [1, 2, 4],
        "max_features": ["sqrt", 0.3, 0.5, 0.8],
        "class_weight": [None, "balanced", "balanced_subsample"],
    },
    "Extra Trees": {
        "n_estimators": [300, 500, 700],
        "max_depth": [None, 10, 15, 20, 30],
        "min_samples_split": [2, 5, 9],
        "min_samples_leaf": [1, 2, 4],
        "max_features": ["sqrt", 0.3, 0.5, 0.8],
        "class_weight": [None, "balanced", "balanced_subsample"],
    },
    "LightGBM": {
        "n_estimators": [200, 400, 700, 1000],
        "learning_rate": [0.02, 0.03, 0.05, 0.1],
        "num_leaves": [7, 15, 22, 31],
        "max_depth": [-1, 5, 7, 10],
        "min_child_samples": [5, 10, 20, 33],
        "subsample": [0.6, 0.8, 1.0],
        "subsample_freq": [1],
        "colsample_bytree": [0.6, 0.8, 1.0],
        "reg_lambda": [0, 0.01, 1, 5],
        "class_weight": [None, "balanced"],
    },
    "CatBoost": {
        "iterations": [300, 500, 700],
        "depth": [4, 5, 6, 7],
        "learning_rate": [0.03, 0.05, 0.1],
        "l2_leaf_reg": [1, 3, 5, 7],
        "auto_class_weights": [None, "Balanced", "SqrtBalanced"],
    },
}


# ============================================================
# 7. 하이퍼파라미터 튜닝 (개발 데이터 5-Fold, Macro F1)
# ============================================================

tuned = {}
tuning_rows = []

for name, model in models.items():
    print(f"\n{name} 튜닝 중... ({N_ITER}개 조합 × 5-Fold)")
    start = time.time()
    search = RandomizedSearchCV(
        model, param_spaces[name], n_iter=N_ITER, scoring="f1_macro",
        cv=cv5, random_state=RANDOM_STATE, n_jobs=-1, refit=False,
    )
    search.fit(X_dev, y_dev)
    tuned[name] = clone(model).set_params(**search.best_params_)

    tuning_rows.append({
        "Model": name,
        "CV Macro F1": search.best_score_,
        "CV SD": search.cv_results_["std_test_score"][search.best_index_],
        "Best Params": search.best_params_,
        "Time(s)": round(time.time() - start, 1),
    })
    print(f"  CV Macro F1 = {search.best_score_:.4f} | {search.best_params_}")

tuning_df = (pd.DataFrame(tuning_rows)
             .sort_values("CV Macro F1", ascending=False).reset_index(drop=True))
print("\n[튜닝 결과]")
display(tuning_df)


# ============================================================
# 8. Out-of-fold 확률 예측
# ============================================================
# 개발 데이터의 모든 행에 대해 "그 행을 학습에 쓰지 않은 모델"의 예측 확률을 얻는다.
# 앙상블 조합과 클래스별 확률 보정은 이 확률로만 결정하고 테스트는 건드리지 않는다.
# (튜닝에 쓴 fold와 같으므로 약간 낙관적일 수 있음 → 최종 판단은 테스트에서)
# ============================================================

oof = {}
for name, est in tuned.items():
    oof[name] = cross_val_predict(
        clone(est), X_dev, y_dev, cv=cv5, method="predict_proba", n_jobs=-1
    )


def tune_class_scale(P, y, grid=np.round(np.arange(0.5, 3.01, 0.1), 2), n_rounds=3):
    """
    예측 확률에 클래스별 배율을 곱해 argmax 했을 때 Macro F1이 최대가 되는 배율 탐색.
    배율 > 1 이면 그 클래스를 더 자주 예측(재현율↑ 정밀도↓), < 1 이면 반대.
    → GPT 코드의 '보수적 예측'과 Claude 코드의 '소수 클래스 적극 예측' 사이 균형점을
      사람이 정하지 않고 데이터로 찾는다.
    """
    w = np.ones(P.shape[1])
    best = macro_f1(y, P.argmax(axis=1))
    for _ in range(n_rounds):
        improved = False
        for k in range(P.shape[1]):
            for g in grid:
                trial = w.copy()
                trial[k] = g
                s = macro_f1(y, (P * trial).argmax(axis=1))
                if s > best + 1e-9:
                    best, w, improved = s, trial, True
        if not improved:
            break
    return w, best


ranked = tuning_df["Model"].tolist()
candidates = {name: [name] for name in ranked}
candidates["앙상블(상위 2)"] = ranked[:2]
candidates["앙상블(상위 3)"] = ranked[:3]
candidates["앙상블(전체 4)"] = ranked

cand_rows = []
cand_scale = {}
for cname, members in candidates.items():
    P = np.mean([oof[m] for m in members], axis=0)
    raw = macro_f1(y_dev, P.argmax(axis=1))
    w, scaled = tune_class_scale(P, y_dev)
    cand_scale[cname] = w
    cand_rows.append({"Candidate": cname, "OOF Macro F1": raw,
                      "OOF Macro F1 (확률 보정)": scaled})

cand_df = (pd.DataFrame(cand_rows)
           .sort_values("OOF Macro F1 (확률 보정)", ascending=False).reset_index(drop=True))
print("\n[후보별 Out-of-fold 성능]")
display(cand_df.round(4))


# ============================================================
# 9. 최종 후보 선택
# ============================================================
# 확률 보정은 OOF에 맞춰 찾은 값이라 그 자체로 약간 낙관적이다.
# 보정 이득이 0.005 미만이면 보정 없이 사용한다.
# ============================================================

class ScaledSoftVoting(BaseEstimator, ClassifierMixin):
    """여러 모델의 확률 평균 + 클래스별 배율 보정"""

    def __init__(self, estimators, class_scale=None):
        self.estimators = estimators
        self.class_scale = class_scale

    def fit(self, X, y):
        self.estimators_ = [clone(e).fit(X, y) for e in self.estimators]
        self.classes_ = np.unique(y)
        return self

    def predict_proba(self, X):
        P = np.mean([e.predict_proba(X) for e in self.estimators_], axis=0)
        if self.class_scale is not None:
            P = P * np.asarray(self.class_scale)
            P = P / P.sum(axis=1, keepdims=True)
        return P

    def predict(self, X):
        return self.classes_[self.predict_proba(X).argmax(axis=1)]


best_row = cand_df.iloc[0]
best_cand = best_row["Candidate"]
use_scale = (best_row["OOF Macro F1 (확률 보정)"] - best_row["OOF Macro F1"]) >= 0.005
final_scale = cand_scale[best_cand] if use_scale else None

print("\n최종 후보:", best_cand, "| 구성:", candidates[best_cand])
print("확률 보정 사용:", use_scale)
if use_scale:
    display(pd.Series(final_scale, index=FAULT_COLS, name="클래스 배율").round(2))

final_model = ScaledSoftVoting(
    [tuned[m] for m in candidates[best_cand]], class_scale=final_scale
).fit(X_dev, y_dev)


# ============================================================
# 10. 테스트 평가 (마지막 1회)
# ============================================================

y_pred = final_model.predict(X_test)
test_scores = score_row(y_test, y_pred)

print("\n" + "=" * 60)
print(f"최종 테스트 결과 - {best_cand}{' + 확률 보정' if use_scale else ''}")
print("=" * 60)
display(pd.DataFrame([test_scores]).round(4))

# 부트스트랩 95% 신뢰구간: 테스트셋 구성에 따라 점수가 얼마나 흔들리는지 확인
rng = np.random.default_rng(RANDOM_STATE)
boot = []
for _ in range(1000):
    b = rng.integers(0, len(y_test), len(y_test))
    boot.append(macro_f1(y_test[b], y_pred[b]))
lo, hi = np.percentile(boot, [2.5, 97.5])
print(f"테스트 Macro F1 95% 신뢰구간: [{lo:.3f}, {hi:.3f}]")
print("  → 다른 코드와의 점수 차이가 이 구간 폭보다 작으면 우열을 단정하기 어렵다.")

print("\n[Classification Report - Test]")
print(classification_report(y_test, y_pred, labels=LABELS,
                            target_names=FAULT_COLS, digits=4, zero_division=0))

# 참고용: 다른 후보들의 테스트 성능 (이것을 보고 선택을 바꾸지 않는다)
ref_rows = []
for cname, members in candidates.items():
    m = ScaledSoftVoting([tuned[x] for x in members]).fit(X_dev, y_dev)
    ref_rows.append({"Candidate": cname, "Test Macro F1 (보정 없음)":
                     macro_f1(y_test, m.predict(X_test))})
print("[참고] 후보별 테스트 Macro F1")
display(pd.DataFrame(ref_rows).round(4))


# ============================================================
# 11. 혼동행렬 및 오분류 분석
# ============================================================

fig, axes = plt.subplots(1, 2, figsize=(18, 7))
ConfusionMatrixDisplay(confusion_matrix(y_test, y_pred, labels=LABELS),
                       display_labels=FAULT_COLS).plot(
    ax=axes[0], xticks_rotation=45, colorbar=False)
axes[0].set_title("혼동행렬 (개수)")
ConfusionMatrixDisplay(confusion_matrix(y_test, y_pred, labels=LABELS, normalize="true"),
                       display_labels=FAULT_COLS).plot(
    ax=axes[1], xticks_rotation=45, colorbar=False, values_format=".2f")
axes[1].set_title("혼동행렬 (실제 유형 기준 비율)")
for ax in axes:
    ax.set_xlabel("예측 결함 유형")
    ax.set_ylabel("실제 결함 유형")
plt.tight_layout()
plt.show()

test_result = pd.DataFrame({
    "Actual": [int_to_class[i] for i in y_test],
    "Predicted": [int_to_class[i] for i in y_pred],
    "Confidence": final_model.predict_proba(X_test).max(axis=1),
}, index=X_test.index)
misclassified = test_result[test_result["Actual"] != test_result["Predicted"]]

print(f"\n오분류: {len(misclassified)} / {len(y_test)} ({len(misclassified) / len(y_test):.2%})")
display(misclassified.groupby(["Actual", "Predicted"]).size()
        .sort_values(ascending=False).head(10).to_frame("Count"))


# ============================================================
# 12. 변수 중요도 (Permutation Importance, 테스트 데이터)
# ============================================================

perm = permutation_importance(final_model, X_test, y_test, scoring="f1_macro",
                              n_repeats=20, random_state=RANDOM_STATE, n_jobs=-1)
importance_df = (pd.DataFrame({
    "Feature": feature_cols,
    "Importance Mean": perm.importances_mean,
    "Importance SD": perm.importances_std,
}).sort_values("Importance Mean", ascending=False).reset_index(drop=True))
display(importance_df.head(15).round(4))

top = importance_df.head(15).iloc[::-1]
plt.figure(figsize=(9, 7))
plt.barh(top["Feature"], top["Importance Mean"], xerr=top["Importance SD"])
plt.xlabel("섞었을 때 Macro F1 감소량")
plt.title("Permutation Importance 상위 15개")
plt.tight_layout()
plt.show()


# ============================================================
# 13. 전체 데이터로 재학습 및 저장
# ============================================================

deploy_model = ScaledSoftVoting(
    [tuned[m] for m in candidates[best_cand]], class_scale=final_scale
).fit(X_all, y_all)

joblib.dump({
    "model": deploy_model,
    "candidate": best_cand,
    "members": candidates[best_cand],
    "params": {m: tuned[m].get_params() for m in candidates[best_cand]},
    "class_scale": None if final_scale is None else final_scale.tolist(),
    "use_extra_features": USE_EXTRA,
    "feature_cols": feature_cols,
    "class_names": FAULT_COLS,
    "test_scores": test_scores,
}, "steel_fault_model_combined.joblib")
print("\n저장 완료: steel_fault_model_combined.joblib")


# ============================================================
# 14. 새 데이터 예측 함수
# ============================================================
# 원본 형식(TypeOfSteel_A300/A400) 또는 병합 형식(Steel_Type) 모두 입력 가능.
# ScaledSoftVoting 클래스가 정의된 환경에서 불러와야 한다.
# ============================================================

def predict_faults(new_data: pd.DataFrame,
                   model_path: str = "steel_fault_model_combined.joblib") -> pd.DataFrame:
    bundle = joblib.load(model_path)
    names = bundle["class_names"]
    merged = merge_columns(new_data.drop(columns=names, errors="ignore"))
    X_new = build_features(merged, bundle["use_extra_features"])[bundle["feature_cols"]]

    proba = bundle["model"].predict_proba(X_new)
    out = pd.DataFrame(proba, columns=[f"prob_{n}" for n in names], index=new_data.index)
    out.insert(0, "predicted_fault", [names[i] for i in proba.argmax(axis=1)])
    out.insert(1, "confidence", proba.max(axis=1))
    return out


demo = predict_faults(raw_df.iloc[idx_test[:5]])
demo.insert(0, "actual_fault", df.iloc[idx_test[:5]][TARGET].values)
print("\n[예측 함수 동작 확인 - 전체 데이터로 재학습한 모델이므로 성능 지표 아님]")
display(demo.round(3))
```

## 성능 개선 실험 (데이터를 늘릴 수 없는 경우)

0번 셀만 실행한 뒤 아래 셀을 위에서부터 차례로 실행하면 됩니다. 과적합 점검 셀(17~21)은 다시 돌릴 필요가 없습니다.

```python
# ============================================================
# 22. 성능 개선 실험 - 공통 평가 틀
# ============================================================
# 0번 셀(노트북 본체)만 실행돼 있으면 된다. (과적합 점검 17~21번 셀은 필요 없음)
#
# [비교 방법]
#   - 개발 데이터(1,552개)에서 5-Fold × 3회 반복 = 15개 fold로 평가한다.
#   - 모든 실험을 같은 15개 fold에서 기준 모델과 짝지어 비교한다.
#   - 테스트셋은 이미 결과를 본 데이터라 실험 비교에는 쓰지 않는다.
#   - 새 변수 선택, 전문 모델 학습, 오버샘플링, 의사 라벨처럼 "데이터를 보고 정하는 일"은
#     전부 fold 안의 학습 데이터에서만 한다. 검증 데이터는 평가에만 쓰인다.
#
# [기준 모델] 최종 모델과 같은 구성(같은 앙상블 구성원, 같은 하이퍼파라미터).
#   확률 보정은 효과가 0 근처였고 배율 탐색이 실험 효과를 가리므로 끄고 비교한다.
#
# [판정 기준] 기준 대비 평균 +0.005 이상이고 15개 fold 중 2/3(10개) 이상에서 나아졌을 때만
#   "효과 있음". 테스트 점수가 분할에 따라 ±0.018 흔들리므로 그보다 작은 차이는
#   단일 테스트셋으로는 확인할 수 없다는 점을 기억해 둔다.
#
# [소요 시간] 실험 1개 = 모델 15번 학습. 코어 수에 따라 실험당 수십 초~2분.
# ============================================================

from joblib import Parallel, delayed
from sklearn.metrics import roc_auc_score
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

BASE_NAMES = candidates[best_cand]
BASE_MEMBERS = [tuned[m] for m in BASE_NAMES]
EXP_SPLITS = list(RepeatedStratifiedKFold(n_splits=5, n_repeats=3,
                                          random_state=RANDOM_STATE + 7).split(X_dev, y_dev))
BUMPS, OTHER = class_to_int["Bumps"], class_to_int["Other_Faults"]
exp_results = {}          # 실험 이름 → fold별 결과 DataFrame


def identity_prep(X_tr, y_tr, X_va):
    """아무것도 바꾸지 않는 전처리 (기준 모델)"""
    return X_tr, y_tr, X_va


def compose(*preps):
    """여러 전처리를 순서대로 적용 (예: 새 변수 추가 → 오버샘플링)"""
    def prep(X_tr, y_tr, X_va):
        info = {}
        for p in preps:
            res = p(X_tr, y_tr, X_va)
            X_tr, y_tr, X_va = res[:3]
            if len(res) == 4:
                info.update(res[3])
        return X_tr, y_tr, X_va, info
    return prep


def _exp_fold(i, tr, va, prep, members):
    res = prep(X_dev.iloc[tr].copy(), y_dev[tr].copy(), X_dev.iloc[va].copy())
    X_tr, y_tr, X_va = res[:3]
    info = res[3] if len(res) == 4 else {}
    m = ScaledSoftVoting(members).fit(X_tr, y_tr)
    pred = m.predict(X_va)
    per_class = f1_score(y_dev[va], pred, labels=LABELS, average=None, zero_division=0)
    row = {"fold": i, "valid": macro_f1(y_dev[va], pred),
           "train": macro_f1(y_tr, m.predict(X_tr)), "n_train": len(y_tr),
           **{f"f1_{c}": v for c, v in zip(FAULT_COLS, per_class)}}
    row.update(info)
    return row


def run_experiment(name, prep=identity_prep, members=None, verbose=True):
    """15개 fold에서 (전처리 → 앙상블 학습 → 검증) 을 수행하고 결과를 저장"""
    members = BASE_MEMBERS if members is None else members
    start = time.time()
    rows = Parallel(n_jobs=-1)(delayed(_exp_fold)(i, tr, va, prep, members)
                               for i, (tr, va) in enumerate(EXP_SPLITS))
    r = pd.DataFrame(rows).set_index("fold")
    exp_results[name] = r
    if verbose:
        print(f"{name:<32} 검증 {r['valid'].mean():.4f} ± {r['valid'].std():.4f} "
              f"| 학습 {r['train'].mean():.4f} | {time.time() - start:.0f}초")
    return r


def verdict(d):
    need = int(np.ceil(len(d) * 2 / 3))
    if d.mean() >= 0.005 and (d > 0).sum() >= need:
        return "효과 있음"
    if d.mean() <= -0.005 and (d < 0).sum() >= need:
        return "악화"
    return "차이 없음"


def compare_experiments(names=None, baseline="기준 모델"):
    """기준 모델과 같은 fold끼리 짝지어 비교한 표"""
    base = exp_results[baseline]["valid"]
    rows = []
    for name in (names or list(exp_results)):
        r = exp_results[name]
        d = r["valid"] - base
        rows.append({
            "실험": name, "검증 Macro F1": r["valid"].mean(), "SD": r["valid"].std(),
            "학습 Macro F1": r["train"].mean(),
            "기준 대비": d.mean(),
            "나아진 fold": "-" if name == baseline else f"{int((d > 0).sum())}/{len(d)}",
            "Bumps F1": r["f1_Bumps"].mean(), "Other F1": r["f1_Other_Faults"].mean(),
            "Pastry F1": r["f1_Pastry"].mean(), "Dirtiness F1": r["f1_Dirtiness"].mean(),
            "판정": "-" if name == baseline else verdict(d),
        })
    out = (pd.DataFrame(rows).set_index("실험")
           .sort_values("검증 Macro F1", ascending=False))
    display(out.round(4))
    return out


def pair_auc(x, y):
    """두 클래스를 한 변수로 얼마나 잘 나누는지 (0.5 = 못 나눔, 1.0 = 완벽, 방향 무관)"""
    if pd.Series(x).nunique() < 2:
        return 0.5
    auc = roc_auc_score(y, x)
    return max(auc, 1 - auc)


print(f"기준 모델 구성: {BASE_NAMES} (확률 보정 없음) | 평가 fold {len(EXP_SPLITS)}개")
run_experiment("기준 모델")
```

```python
# ============================================================
# 23. [2] Bumps vs Other_Faults - 어떤 변수가 두 클래스를 가르는가 (개발 데이터만)
# ============================================================
# 구분력(AUC): 그 변수 하나로 두 클래스를 나눌 때의 성능. 0.5면 전혀 못 나누고 1.0이면 완벽.
# 헷갈린 샘플: 노트북 8단계 OOF 예측에서 Bumps ↔ Other_Faults 로 틀린 개발 데이터 행.
#   "맞힌 Bumps / 헷갈린 Bumps / 헷갈린 Other / 맞힌 Other" 의 중앙값을 나란히 보면
#   헷갈린 샘플이 상대 클래스 쪽으로 치우친 변수가 무엇인지 보인다.
# 이 표와 그림을 현장 담당자와 함께 보면서 두 결함을 실제로 어떻게 구분하는지
# 물어보면 24번 셀의 domain_features 에 넣을 아이디어를 얻을 수 있다.
# ============================================================

pair_mask = np.isin(y_dev, [BUMPS, OTHER])
X_bo = X_dev[pair_mask]
y_bo = (y_dev[pair_mask] == BUMPS).astype(int)          # 1 = Bumps, 0 = Other_Faults

P_oof = np.mean([oof[m] for m in BASE_NAMES], axis=0)    # 기준 앙상블의 OOF 예측
pred_oof = P_oof.argmax(axis=1)[pair_mask]
group = np.select(
    [(y_bo == 1) & (pred_oof == BUMPS), (y_bo == 1) & (pred_oof == OTHER),
     (y_bo == 0) & (pred_oof == BUMPS), (y_bo == 0) & (pred_oof == OTHER)],
    ["맞힌 Bumps", "헷갈린 Bumps(→Other)", "헷갈린 Other(→Bumps)", "맞힌 Other"],
    default="다른 클래스로 예측")
print("[개발 데이터 OOF 기준 Bumps / Other_Faults 예측 결과]")
display(pd.Series(group).value_counts().to_frame("개수"))

bo_rows = []
for c in feature_cols:
    row = {"변수": c, "구분력(AUC)": pair_auc(X_bo[c], y_bo)}
    for g in ["맞힌 Bumps", "헷갈린 Bumps(→Other)", "헷갈린 Other(→Bumps)", "맞힌 Other"]:
        row[g] = X_bo.loc[group == g, c].median()
    bo_rows.append(row)
bo_df = (pd.DataFrame(bo_rows).sort_values("구분력(AUC)", ascending=False)
         .reset_index(drop=True))
print("\n[변수별 Bumps vs Other_Faults 구분력과 그룹별 중앙값]")
display(bo_df.round(3))

top_feats = bo_df["변수"].head(8).tolist()
fig, axes = plt.subplots(2, 4, figsize=(20, 8))
for ax, c in zip(axes.ravel(), top_feats):
    lo_q, hi_q = X_bo[c].quantile([0.01, 0.99])
    bins = np.linspace(lo_q, hi_q, 30) if hi_q > lo_q else 10
    for g, color in [("맞힌 Bumps", "tab:blue"), ("맞힌 Other", "tab:orange")]:
        ax.hist(X_bo.loc[group == g, c].clip(lo_q, hi_q), bins=bins, alpha=0.45,
                density=True, color=color, label=g)
    for g, color in [("헷갈린 Bumps(→Other)", "navy"), ("헷갈린 Other(→Bumps)", "darkred")]:
        ax.axvline(X_bo.loc[group == g, c].median(), color=color, ls="--", lw=1.5,
                   label=f"{g} 중앙값")
    ax.set_title(f"{c} (AUC {pair_auc(X_bo[c], y_bo):.3f})")
axes[0, 0].legend(fontsize=8)
plt.suptitle("Bumps vs Other_Faults 구분력 상위 8개 변수 (1~99% 구간)")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 24. [2] 두 클래스 구분에 초점을 맞춘 변수 실험
# ============================================================
# 세 종류를 시험한다. 모두 fold의 학습 데이터만 보고 만든다.
#   (a) 전문 점수: Bumps와 Other_Faults만으로 "Bumps일 확률"을 내는 작은 모델을 학습해
#       그 점수를 새 변수로 추가. 학습 데이터 행에는 내부 5-Fold OOF 점수를 넣어
#       자기 자신을 외운 점수가 들어가지 않게 한다. 로지스틱 / LightGBM 두 가지.
#   (b) 자동 조합 변수: 구분력 상위 변수끼리의 표준화 합·차, 로그 비율을 만들고
#       두 원본 변수보다 구분력이 PAIR_MIN_GAIN 이상 높은 것만 추가.
#   (c) 현장 지식 변수: 아래 domain_features 함수에 직접 작성 (비어 있으면 건너뜀).
# ============================================================

PAIR_TOP_K = 8         # 조합에 쓸 구분력 상위 변수 수
PAIR_N_KEEP = 5        # 추가할 조합 변수 최대 개수
PAIR_MIN_GAIN = 0.02   # 원본 두 변수의 구분력보다 이만큼 높아야 후보


def make_specialist_prep(kind="logistic"):
    col = f"Spec_{kind}_Bumps"

    def prep(X_tr, y_tr, X_va):
        mask = np.isin(y_tr, [BUMPS, OTHER])
        Xp, yp = X_tr[mask], (y_tr[mask] == BUMPS).astype(int)
        if kind == "logistic":
            est = make_pipeline(StandardScaler(), LogisticRegression(C=1.0, max_iter=5000))
        else:
            est = LGBMClassifier(n_estimators=300, learning_rate=0.03, num_leaves=15,
                                 min_child_samples=20, subsample=0.8, subsample_freq=1,
                                 colsample_bytree=0.8, random_state=RANDOM_STATE,
                                 n_jobs=1, verbose=-1)
        est.fit(Xp, yp)
        tr_score = est.predict_proba(X_tr)[:, 1]
        inner = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
        tr_score[mask] = cross_val_predict(clone(est), Xp, yp, cv=inner,
                                           method="predict_proba")[:, 1]
        X_tr, X_va = X_tr.copy(), X_va.copy()
        X_tr[col] = tr_score
        X_va[col] = est.predict_proba(X_va)[:, 1]
        return X_tr, y_tr, X_va
    return prep


def pair_feature_prep(X_tr, y_tr, X_va):
    mask = np.isin(y_tr, [BUMPS, OTHER])
    yp = (y_tr[mask] == BUMPS).astype(int)
    base_auc = {c: pair_auc(X_tr.loc[mask, c], yp)
                for c in X_tr.columns if X_tr[c].nunique() > 2}
    top = sorted(base_auc, key=base_auc.get, reverse=True)[:PAIR_TOP_K]
    mu, sd = X_tr[top].mean(), X_tr[top].std().replace(0, 1)
    nonneg = [c for c in top if X_tr[c].min() >= 0]

    def make(X):
        Z = (X[top] - mu) / sd
        feats, parts = {}, {}
        names = sorted(top)                     # fold마다 같은 조합이 같은 이름을 갖도록 정렬
        for i, a in enumerate(names):
            for b in names[i + 1:]:
                feats[f"Z({a})-Z({b})"] = Z[a] - Z[b]
                feats[f"Z({a})+Z({b})"] = Z[a] + Z[b]
                parts[f"Z({a})-Z({b})"] = parts[f"Z({a})+Z({b})"] = (a, b)
                if a in nonneg and b in nonneg:
                    feats[f"log({a}/{b})"] = np.log1p(X[a].clip(lower=0)) - np.log1p(X[b].clip(lower=0))
                    parts[f"log({a}/{b})"] = (a, b)
        return pd.DataFrame(feats, index=X.index), parts

    F_tr, parts = make(X_tr)
    gains = {}
    for c in F_tr.columns:
        a, b = parts[c]
        g = pair_auc(F_tr.loc[mask, c], yp) - max(base_auc[a], base_auc[b])
        if g >= PAIR_MIN_GAIN:
            gains[c] = g
    keep = sorted(gains, key=gains.get, reverse=True)[:PAIR_N_KEEP]
    F_va, _ = make(X_va)
    X_tr = pd.concat([X_tr, F_tr[keep]], axis=1)
    X_va = pd.concat([X_va, F_va[keep]], axis=1)
    return X_tr, y_tr, X_va, {"추가 변수": " | ".join(keep)}


def domain_features(X):
    """
    현장 지식으로 만든 변수를 여기에 추가한다.
    행 하나만 보고 계산되는 변수만 넣는다 (다른 행이나 라벨을 쓰면 안 됨).
    아무것도 추가하지 않으면 (c) 실험은 건너뛴다. 아래는 형식 예시.
    """
    out = pd.DataFrame(index=X.index)
    # 예) 결함 영역의 평균 밝기 대비 최대 밝기
    # out["Peak_vs_Mean_Lum"] = X["Maximum_of_Luminosity"] / (X["Sum_of_Luminosity"] / X["Pixels_Areas"] + 1)
    # 예) 경계 상자 대비 실제 결함 면적 비율
    # out["Fill_Ratio"] = X["Pixels_Areas"] / ((X["X_Maximum"] - X["X_Minimum"] + 1) * (X["Y_Maximum"] - X["Y_Minimum"] + 1))
    return out


def domain_prep(X_tr, y_tr, X_va):
    return (pd.concat([X_tr, domain_features(X_tr)], axis=1), y_tr,
            pd.concat([X_va, domain_features(X_va)], axis=1))


run_experiment("[2] 전문 점수 (로지스틱)", make_specialist_prep("logistic"))
run_experiment("[2] 전문 점수 (LightGBM)", make_specialist_prep("lgbm"))
run_experiment("[2] 자동 조합 변수", pair_feature_prep)
if domain_features(X_dev.head()).shape[1] > 0:
    run_experiment("[2] 현장 지식 변수", domain_prep)
else:
    print("[2] 현장 지식 변수: domain_features 가 비어 있어 건너뜀")

if "추가 변수" in exp_results["[2] 자동 조합 변수"]:
    picked = (exp_results["[2] 자동 조합 변수"]["추가 변수"].str.split(" | ", regex=False)
              .explode().replace("", np.nan).dropna().value_counts())
    print(f"\n[자동 조합 변수 - 15개 fold에서 선택된 횟수]  (선택된 변수가 없으면 비어 있음)")
    display(picked.head(10).to_frame("선택 횟수"))

compare_experiments([n for n in exp_results if n == "기준 모델" or n.startswith("[2]")])
```

```python
# ============================================================
# 25. [3] 규제 강화 - 덜 외우게 하면 새 데이터에서 나아지는가
# ============================================================
# 최종 모델의 하이퍼파라미터에서 "외우는 정도"를 정하는 값 하나씩만 바꿔 본다.
#   Extra Trees : min_samples_leaf (잎 하나에 최소 샘플 수, 클수록 덜 외움)
#   LightGBM    : min_child_samples (위와 같은 의미), reg_lambda (L2 규제, 클수록 덜 외움)
# 앙상블에 없는 모델의 파라미터는 건너뛴다.
# [주의] 여러 값 중 가장 좋은 값을 같은 15개 fold로 고르므로 "최적 조합" 점수는
#        약간 낙관적이다. 기준 대비 차이가 작으면 효과가 없다고 보는 것이 안전하다.
# ============================================================

REG_SWEEP = [
    ("Extra Trees", "min_samples_leaf", [1, 2, 4, 8, 16]),
    ("LightGBM", "min_child_samples", [5, 10, 20, 40, 80]),
    ("LightGBM", "reg_lambda", [0, 1, 5, 20, 50]),
]


def members_with(changes):
    """changes = {모델 이름: {파라미터: 값}} 을 적용한 앙상블 구성원"""
    out = []
    for name in BASE_NAMES:
        est = clone(tuned[name])
        if name in changes:
            est.set_params(**changes[name])
        out.append(est)
    return out


sweep_rows = []
for model_name, param, values in REG_SWEEP:
    if model_name not in BASE_NAMES:
        print(f"{model_name} 은(는) 앙상블에 없어 건너뜀")
        continue
    current = tuned[model_name].get_params()[param]
    for v in values:
        r = run_experiment(f"[3] {model_name} {param}={v}",
                           members=members_with({model_name: {param: v}}), verbose=False)
        d = r["valid"] - exp_results["기준 모델"]["valid"]
        sweep_rows.append({"모델": model_name, "파라미터": param, "값": v,
                           "현재 값": "◀" if v == current else "",
                           "학습": r["train"].mean(), "검증": r["valid"].mean(),
                           "격차": r["train"].mean() - r["valid"].mean(),
                           "기준 대비": d.mean()})
sweep_df = pd.DataFrame(sweep_rows)
print("[규제 값별 결과 - 15개 fold 평균]")
display(sweep_df.round(4))

groups = list(sweep_df.groupby(["모델", "파라미터"], sort=False))
fig, axes = plt.subplots(1, len(groups), figsize=(6 * len(groups), 4.5), squeeze=False)
for ax, ((model_name, param), g) in zip(axes[0], groups):
    x = np.arange(len(g))
    ax.plot(x, g["학습"], "o-", label="학습")
    ax.plot(x, g["검증"], "o-", label="검증")
    ax.axhline(exp_results["기준 모델"]["valid"].mean(), color="gray", ls="--", lw=1, label="기준 검증")
    ax.set_xticks(x)
    ax.set_xticklabels(g["값"].astype(str))
    ax.set_title(f"{model_name} {param}")
    ax.set_ylabel("Macro F1")
    ax.grid(alpha=0.3)
    ax.legend()
plt.tight_layout()
plt.show()

# 파라미터별 최적 값을 모아 한 번에 적용
best_changes = {}
for (model_name, param), g in groups:
    best_changes.setdefault(model_name, {})[param] = g.loc[g["검증"].idxmax(), "값"]
best_changes = {m: {k: (int(v) if float(v).is_integer() else float(v)) for k, v in p.items()}
                for m, p in best_changes.items()}
print("최적 조합:", best_changes)
REG_MEMBERS = members_with(best_changes)
run_experiment("[3] 규제 최적 조합", members=REG_MEMBERS)
compare_experiments(["기준 모델", "[3] 규제 최적 조합"])
```

```python
# ============================================================
# 26. [4] 오버샘플링 - fold의 학습 데이터에만 적용
# ============================================================
# 개수가 OS_TARGET 보다 적은 클래스만 OS_TARGET 개까지 늘린다.
#   (학습 fold 기준 Dirtiness 약 35, Stains 약 46, Pastry 약 101, Z_Scratch 약 122개)
#   RandomOverSampler : 기존 샘플을 그대로 복제
#   SMOTE             : 같은 클래스 이웃 사이를 보간해 새 샘플 생성
#   BorderlineSMOTE   : 다른 클래스와의 경계 근처 샘플 위주로 생성
# 값 종류가 3개 이하인 변수(Steel_Type 등)는 보간 후 가장 가까운 원래 값으로 되돌린다.
# 검증 데이터에는 절대 적용하지 않는다. (학습 점수는 가짜 샘플이 포함된 값이라 참고만)
# 필요 패키지: pip install imbalanced-learn
# ============================================================

try:
    from imblearn.over_sampling import RandomOverSampler, SMOTE, BorderlineSMOTE
    HAS_IMBLEARN = True
except ImportError:
    HAS_IMBLEARN = False
    print("imbalanced-learn 이 없습니다. pip install imbalanced-learn 후 다시 실행하세요.")


def make_oversample_prep(kind="smote", target=150):
    def prep(X_tr, y_tr, X_va):
        counts = pd.Series(y_tr).value_counts()
        strategy = {int(k): target for k, v in counts.items() if v < target}
        if not strategy:
            return X_tr, y_tr, X_va
        if kind == "random":
            sampler = RandomOverSampler(sampling_strategy=strategy, random_state=RANDOM_STATE)
        elif kind == "borderline":
            sampler = BorderlineSMOTE(sampling_strategy=strategy, k_neighbors=5,
                                      random_state=RANDOM_STATE)
        else:
            sampler = SMOTE(sampling_strategy=strategy, k_neighbors=5, random_state=RANDOM_STATE)
        X_res, y_res = sampler.fit_resample(X_tr, y_tr)
        X_res = pd.DataFrame(np.asarray(X_res), columns=X_tr.columns)
        for c in X_tr.columns:                       # 범주형 값 되돌리기
            levels = np.sort(X_tr[c].unique())
            if len(levels) <= 3:
                idx = np.abs(X_res[c].to_numpy()[:, None] - levels[None, :]).argmin(axis=1)
                X_res[c] = levels[idx]
        return X_res, np.asarray(y_res), X_va
    return prep


if HAS_IMBLEARN:
    run_experiment("[4] RandomOverSampler (150)", make_oversample_prep("random", 150))
    run_experiment("[4] SMOTE (150)", make_oversample_prep("smote", 150))
    run_experiment("[4] SMOTE (300)", make_oversample_prep("smote", 300))
    run_experiment("[4] BorderlineSMOTE (150)", make_oversample_prep("borderline", 150))
    compare_experiments([n for n in exp_results if n == "기준 모델" or n.startswith("[4]")])
```

```python
# ============================================================
# 27. [5] 반지도 학습 (자기 학습, self-training)
# ============================================================
# 라벨 없는 데이터에 대해 모델이 확신하는 예측(확률 ≥ ST_THRESHOLD)을 라벨로 삼아
# 학습 데이터에 추가하고 다시 학습한다. ST_ROUNDS 번 반복.
#
# UNLABELED_PATH 가 None 이면 "모의 실험"을 한다:
#   각 fold의 학습 데이터 중 절반만 라벨이 있다고 가정하고
#     (a) 라벨 50%만 학습  (b) 라벨 50% + 나머지 50%에 의사 라벨  (c) 라벨 100% 학습(상한)
#   을 비교한다. (b)가 (a)→(c) 격차를 얼마나 메우는지(회수율)와 의사 라벨 정확도를 보면
#   이 데이터에서 자기 학습이 쓸모 있을지 미리 판단할 수 있다.
# UNLABELED_PATH 에 실제 파일을 넣으면 그 데이터를 의사 라벨로 추가한 모델과 기준 모델을 비교한다.
#   (라벨 없는 데이터는 학습에만 쓰이고 검증 fold에는 들어가지 않는다)
# ============================================================

UNLABELED_PATH = None      # 예: "steel_unlabeled.csv" (원본과 같은 열 구성, 결함 열은 없어도 됨)
ST_THRESHOLD = 0.90
ST_ROUNDS = 2


def self_train(X_lab, y_lab, X_unlab, y_unlab_true=None):
    """의사 라벨을 붙여 늘린 학습 데이터와 정보(추가 개수, 정확도)를 돌려준다"""
    X_aug, y_aug, info = X_lab, y_lab, {}
    for rnd in range(1, ST_ROUNDS + 1):
        m = ScaledSoftVoting(BASE_MEMBERS).fit(X_aug, y_aug)
        P = m.predict_proba(X_unlab)
        sel = P.max(axis=1) >= ST_THRESHOLD
        pseudo = m.classes_[P.argmax(axis=1)]
        X_aug = pd.concat([X_lab, X_unlab[sel]], axis=0)
        y_aug = np.concatenate([y_lab, pseudo[sel]])
        info[f"추가 개수 {rnd}회차"] = int(sel.sum())
        if y_unlab_true is not None and sel.any():
            info[f"의사 라벨 정확도 {rnd}회차"] = float((pseudo[sel] == y_unlab_true[sel]).mean())
    return X_aug, y_aug, info


def _half_split(X_tr, y_tr):
    lab, unl = train_test_split(np.arange(len(y_tr)), train_size=0.5, stratify=y_tr,
                                random_state=RANDOM_STATE)
    return lab, unl


def half_labeled_prep(X_tr, y_tr, X_va):
    lab, _ = _half_split(X_tr, y_tr)
    return X_tr.iloc[lab], y_tr[lab], X_va


def half_plus_pseudo_prep(X_tr, y_tr, X_va):
    lab, unl = _half_split(X_tr, y_tr)
    X_aug, y_aug, info = self_train(X_tr.iloc[lab], y_tr[lab], X_tr.iloc[unl], y_tr[unl])
    return X_aug, y_aug, X_va, info


if UNLABELED_PATH is None:
    print("[모의 실험] 학습 데이터의 50%만 라벨이 있다고 가정")
    r_a = run_experiment("[5] 모의: 라벨 50%만", half_labeled_prep)
    r_b = run_experiment("[5] 모의: 라벨 50% + 의사 라벨", half_plus_pseudo_prep)
    r_c = exp_results["기준 모델"]
    gap = r_c["valid"].mean() - r_a["valid"].mean()
    gain = r_b["valid"].mean() - r_a["valid"].mean()
    print(f"\n라벨 50% → 100% 격차: {gap:+.4f}")
    print(f"의사 라벨로 얻은 개선: {gain:+.4f}  (회수율 {gain / gap:.0%})" if gap > 0 else
          f"의사 라벨로 얻은 개선: {gain:+.4f}")
    info_cols = [c for c in r_b.columns if c.startswith(("추가 개수", "의사 라벨 정확도"))]
    print("\n[의사 라벨 정보 - 15개 fold 평균]  (모의 실험이라 실제 라벨과 비교 가능)")
    display(r_b[info_cols].mean().round(3).to_frame("평균"))
    compare_experiments(["기준 모델", "[5] 모의: 라벨 50%만", "[5] 모의: 라벨 50% + 의사 라벨"])
else:
    unlab_raw = pd.read_csv(UNLABELED_PATH)
    X_unlab = build_features(merge_columns(unlab_raw.drop(columns=FAULT_COLS, errors="ignore")),
                             USE_EXTRA)[feature_cols]
    print(f"라벨 없는 데이터: {len(X_unlab)}개")

    def unlabeled_prep(X_tr, y_tr, X_va):
        X_aug, y_aug, info = self_train(X_tr, y_tr, X_unlab.reset_index(drop=True))
        return X_aug, y_aug, X_va, info

    r_u = run_experiment("[5] 실제 라벨 없는 데이터 + 의사 라벨", unlabeled_prep)
    display(r_u[[c for c in r_u.columns if c.startswith("추가 개수")]].mean().round(1).to_frame("평균"))
    compare_experiments(["기준 모델", "[5] 실제 라벨 없는 데이터 + 의사 라벨"])
```

```python
# ============================================================
# 28. 전체 실험 비교 및 다음 단계
# ============================================================
# [5] 모의 실험은 학습 데이터를 절반만 쓴 것이라 비교 대상에서 뺀다.
# "효과 있음"이 나온 실험이 있으면:
#   1) 그 설정을 노트북 본체(0번 셀)에 반영하고
#   2) 과적합 점검의 반복 분할 평가(20번 셀)로 다시 확인한 뒤
#   3) 테스트 점수는 "이미 한 번 본 테스트셋"이라는 점을 밝히고 참고로만 보고한다.
# ============================================================

final_names = ["기준 모델"] + [n for n in exp_results
                              if n != "기준 모델" and not n.startswith("[5] 모의")
                              and not (n.startswith("[3]") and n != "[3] 규제 최적 조합")]
print("[전체 실험 비교 - 개발 데이터 5-Fold × 3회, 기준 모델과 같은 fold끼리 비교]")
exp_summary = compare_experiments(final_names)

winners = exp_summary.index[exp_summary["판정"] == "효과 있음"].tolist()
if winners:
    print("효과 있음:", winners)
else:
    print("판정 기준을 넘은 실험이 없습니다. 현재 모델이 이 데이터에서 얻을 수 있는 성능에 "
          "가깝다고 볼 수 있습니다.")

pd.concat({k: v for k, v in exp_results.items()}, names=["실험", "fold"]).to_csv(
    "improve_experiments.csv", encoding="utf-8-sig")
print("저장 완료: improve_experiments.csv")
```

```python
# ============================================================
# 27. 오분류 행 추출 (테스트 + 개발 OOF, 전체 1,941행)
# ============================================================
# 0번 셀만 실행돼 있으면 된다.
#
# [테스트]  final_model(앙상블 + 확률 보정)의 예측 → 10단계 테스트 결과와 같은 기준
# [개발]    8단계 OOF 확률에 같은 앙상블 구성과 같은 클래스 배율을 적용
#           → 각 행을 학습에 쓰지 않은 모델의 예측이므로 학습 데이터를 외운 영향이 없다.
#           (단, 배율을 이 OOF로 찾았으므로 개발 쪽 오분류 수는 약간 적게 나올 수 있다)
#
# [오분류 유형]
#   보정 유발   : 확률 보정 전에는 맞혔는데 보정 후 틀린 행 (배율 부작용)
#   확신 오분류 : 확신도 ≥ MIS_CONF_HIGH 이고 구성원 모델이 모두 같은 오답 → 라벨 재확인 후보
#   경계선      : 1·2순위 확률 차이 < MIS_GAP_LOW → 애매한 샘플
#   일반 오분류 : 나머지
# ============================================================

from sklearn.metrics import roc_auc_score

MIS_CONF_HIGH = 0.70
MIS_GAP_LOW = 0.10
KEY_FEATS = importance_df["Feature"].head(6).tolist()   # 12단계 중요도 상위 변수

members = candidates[best_cand]


def apply_scale(P, scale):
    if scale is None:
        return P
    P = P * np.asarray(scale)
    return P / P.sum(axis=1, keepdims=True)


def build_pred_table(P, P_raw, y_true, member_P, row_ids, split):
    """확률 행렬 → 행 단위 예측 결과 표 (P: 보정 후, P_raw: 보정 전)"""
    n = np.arange(len(y_true))
    order = np.argsort(P, axis=1)
    pred, second = order[:, -1], order[:, -2]

    t = pd.DataFrame({
        "row_id": row_ids,
        "split": split,
        "Actual": [int_to_class[i] for i in y_true],
        "Predicted": [int_to_class[i] for i in pred],
        "Second": [int_to_class[i] for i in second],
        "Pred_No_Scale": [int_to_class[i] for i in P_raw.argmax(axis=1)],
        "Confidence": P[n, pred],
        "True_Prob": P[n, y_true],
        "Top2_Gap": P[n, pred] - P[n, second],
    })
    t["Correct"] = t["Actual"] == t["Predicted"]

    for name, Pm in member_P.items():                 # 구성원 개별 예측 (보정 전)
        t[f"pred_{name}"] = [int_to_class[i] for i in Pm.argmax(axis=1)]
    member_cols = [f"pred_{m}" for m in member_P]
    t["Members_Agree"] = t[member_cols].eq(t["Predicted"], axis=0).all(axis=1)

    prob = pd.DataFrame(P, columns=[f"prob_{c}" for c in FAULT_COLS])
    return pd.concat([t, prob.round(3)], axis=1)


# ── 1. 테스트 / 개발(OOF) 예측 표 ─────────────────────
member_test = {m: est.predict_proba(X_test)
               for m, est in zip(members, final_model.estimators_)}
P_test_raw = np.mean(list(member_test.values()), axis=0)
test_tbl = build_pred_table(apply_scale(P_test_raw, final_scale), P_test_raw,
                            y_test, member_test, X_test.index.to_numpy(), "test")

member_dev = {m: oof[m] for m in members}
P_dev_raw = np.mean(list(member_dev.values()), axis=0)
dev_tbl = build_pred_table(apply_scale(P_dev_raw, final_scale), P_dev_raw,
                           y_dev, member_dev, X_dev.index.to_numpy(), "dev(OOF)")

pred_all = (pd.concat([dev_tbl, test_tbl], ignore_index=True)
            .sort_values("row_id").reset_index(drop=True))

wrong = ~pred_all["Correct"]
pred_all["Error_Type"] = np.select(
    [~wrong,
     wrong & (pred_all["Pred_No_Scale"] == pred_all["Actual"]),
     wrong & (pred_all["Confidence"] >= MIS_CONF_HIGH) & pred_all["Members_Agree"],
     wrong & (pred_all["Top2_Gap"] < MIS_GAP_LOW)],
    ["정답", "보정 유발", "확신 오분류", "경계선"],
    default="일반 오분류")

# 테스트 결과가 10단계와 같은지 확인
assert (test_tbl["Predicted"].map(class_to_int).to_numpy()
        == final_model.predict(X_test)).all()


# ── 2. 오분류 요약 ────────────────────────────────────
mis_all = pred_all[wrong].join(df[["Steel_Type"]], on="row_id")

print("[분할별 오분류]")
display(pred_all.groupby("split")["Correct"]
        .agg(전체="size", 오분류=lambda s: (~s).sum())
        .assign(오류율=lambda d: (d["오분류"] / d["전체"]).round(4)))

print("[오분류 유형 × 분할]")
display(pd.crosstab(mis_all["Error_Type"], mis_all["split"], margins=True, margins_name="합계"))

fixed = (~pred_all["Correct"].to_numpy()) == False
fixed_by_scale = (pred_all["Correct"] & (pred_all["Pred_No_Scale"] != pred_all["Actual"])).sum()
print(f"확률 보정으로 고쳐진 행: {fixed_by_scale}개 / 보정 때문에 틀린 행: "
      f"{(pred_all['Error_Type'] == '보정 유발').sum()}개")


# ── 3. 클래스별 오류율 (개발 OOF vs 테스트) ────────────
cls_err = (pred_all.groupby(["Actual", "split"])["Correct"]
           .agg(n="size", wrong=lambda s: (~s).sum())
           .assign(error_rate=lambda d: (d["wrong"] / d["n"]).round(3))
           ["error_rate"].unstack("split").reindex(FAULT_COLS))
print("\n[클래스별 오류율]  두 분할의 값이 크게 다르면 테스트셋 구성의 운이 섞인 것")
display(cls_err)


# ── 4. 혼동 쌍 (전체 1,941행 기준) ────────────────────
pair_tbl = (mis_all.groupby(["Actual", "Predicted"])
            .agg(개수=("row_id", "size"),
                 테스트=("split", lambda s: (s == "test").sum()),
                 평균확신도=("Confidence", "mean"),
                 확신오분류=("Error_Type", lambda s: (s == "확신 오분류").sum()),
                 경계선=("Error_Type", lambda s: (s == "경계선").sum()),
                 보정유발=("Error_Type", lambda s: (s == "보정 유발").sum()))
            .sort_values("개수", ascending=False))
print("\n[혼동 쌍 (실제 → 예측)]")
display(pair_tbl.head(12).round(3))


# ── 5. 확신 오분류 (라벨 재확인 후보) ─────────────────
view_cols = (["row_id", "split", "Actual", "Predicted", "Confidence", "True_Prob",
              "Steel_Type"] + [c for c in KEY_FEATS if c != "Steel_Type"])
confident = (mis_all[mis_all["Error_Type"] == "확신 오분류"]
             .join(X_all[[c for c in KEY_FEATS if c != "Steel_Type"]], on="row_id")
             .sort_values("Confidence", ascending=False))
print(f"\n[확신 오분류 {len(confident)}건] 구성원 모델이 모두 같은 오답을 높은 확신으로 냄")
display(confident[view_cols].head(20).round(3))


# ── 6. 혼동 쌍별 변수 비교 ────────────────────────────
def pair_profile(actual, predicted, top_n=10):
    """
    actual 을 predicted 로 틀린 행이 어떤 변수에서 predicted 쪽을 닮았는지.
    위치: 오분류 그룹 중앙값이 두 정답 그룹 중앙값 사이 어디에 있는지
          (0 = 실제 클래스 쪽, 1 = 예측 클래스 쪽, 1 초과 = 예측 클래스보다도 더 멀리)
    """
    ok_a = pred_all.loc[(pred_all["Actual"] == actual) & pred_all["Correct"], "row_id"]
    ok_p = pred_all.loc[(pred_all["Actual"] == predicted) & pred_all["Correct"], "row_id"]
    bad = pred_all.loc[(pred_all["Actual"] == actual)
                       & (pred_all["Predicted"] == predicted), "row_id"]
    if len(bad) == 0:
        print(f"{actual} → {predicted} 오분류 없음")
        return None

    cols = [c for c in feature_cols if X_all[c].nunique() > 2]
    lab = np.r_[np.zeros(len(ok_a)), np.ones(len(ok_p))]
    rows = []
    for c in cols:
        a, b, w = X_all.loc[ok_a, c].median(), X_all.loc[ok_p, c].median(), X_all.loc[bad, c].median()
        auc = roc_auc_score(lab, np.r_[X_all.loc[ok_a, c], X_all.loc[ok_p, c]])
        rows.append({"변수": c, f"맞힌 {actual}": a, f"{actual}→{predicted}": w,
                     f"맞힌 {predicted}": b, "정답끼리 분리도(AUC)": max(auc, 1 - auc),
                     "위치(0=실제,1=예측)": (w - a) / (b - a) if b != a else np.nan})
    out = (pd.DataFrame(rows).sort_values("정답끼리 분리도(AUC)", ascending=False)
           .head(top_n).reset_index(drop=True))
    st = df.loc[bad, "Steel_Type"].value_counts().to_dict()
    print(f"\n[{actual} → {predicted}] 오분류 {len(bad)}건 | 강종 {st}")
    display(out.round(3))
    return out


profiles = {}
for (a, p) in pair_tbl.head(3).index:
    profiles[(a, p)] = pair_profile(a, p)


# ── 7. 저장 ───────────────────────────────────────────
type_order = {"확신 오분류": 0, "보정 유발": 1, "경계선": 2, "일반 오분류": 3}
export = (mis_all.assign(_o=mis_all["Error_Type"].map(type_order))
          .sort_values(["_o", "Confidence"], ascending=[True, False])
          .drop(columns="_o")
          .join(raw_df, on="row_id"))          # 원본 34개 열 전체 붙이기

export.to_csv("misclassified_rows.csv", index=False, encoding="utf-8-sig")
pred_all.to_csv("all_predictions_oof_test.csv", index=False, encoding="utf-8-sig")
print(f"\n저장 완료: misclassified_rows.csv ({len(export)}행), "
      f"all_predictions_oof_test.csv ({len(pred_all)}행)")
```
