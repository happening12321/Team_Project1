```python

## 기준 코드

# ============================================================
# 강판 결함 유형 분류 프로젝트 - 통합 코드
# 데이터: steel_plates_faults.csv
# 목표: 7개 결함 유형의 발생 여부 예측
# ============================================================

import os
import time
import json
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.multioutput import MultiOutputClassifier
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    hamming_loss,
    classification_report,
    multilabel_confusion_matrix
)

warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# 1. 환경 설정 및 데이터 불러오기
# ------------------------------------------------------------

RANDOM_STATE = 42
DATA_PATH = "steel_plates_faults.csv"
OUTPUT_DIR = "steel_faults_results"

os.makedirs(OUTPUT_DIR, exist_ok=True)

plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False

df = pd.read_csv(DATA_PATH)

target_cols = [
    "Pastry", "Z_Scratch", "K_Scatch",
    "Stains", "Dirtiness", "Bumps", "Other_Faults"
]

missing_targets = [col for col in target_cols if col not in df.columns]

if missing_targets:
    raise ValueError(f"타깃 컬럼이 없습니다: {missing_targets}")

print("=" * 60)
print("1. 데이터 기본 점검")
print("=" * 60)

print("데이터 크기:", df.shape)
print("중복 행 수:", df.duplicated().sum())
print("전체 결측치 수:", df.isna().sum().sum())
print("중복 컬럼명 여부:", df.columns.duplicated().any())

display(df.head())
display(df.dtypes.value_counts())
display(df.describe().T.round(3))

# ------------------------------------------------------------
# 2. 타깃 및 설명 변수 분리
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("2. 타깃 및 데이터 누수 점검")
print("=" * 60)

X_raw = df.drop(columns=target_cols).copy()
Y_raw = df[target_cols].copy()

print("설명 변수 수:", X_raw.shape[1])
print("타깃 변수 수:", Y_raw.shape[1])

# 숫자형 데이터 변환
X_numeric = X_raw.apply(pd.to_numeric, errors="coerce")
Y_numeric = Y_raw.apply(pd.to_numeric, errors="coerce")

# 타깃이 0 또는 1인지 검증
if not Y_numeric.isin([0, 1]).all().all():
    raise ValueError("타깃에 0과 1 이외의 값이 존재합니다.")

# 타깃이 모두 0인 행도 정상 데이터일 수 있으므로 자동 삭제하지 않음
print("타깃 이진값 확인: 통과")
print("동시 결함 라벨 개수가 2개 이상인 행:",
      (Y_numeric.sum(axis=1) >= 2).sum())

# 타깃과 완전히 동일한 설명 변수 확인
duplicate_target_features = [
    feature
    for feature in X_numeric.columns
    if any(X_numeric[feature].equals(Y_numeric[target])
           for target in target_cols)
]

print("타깃과 완전히 동일한 설명 변수:",
      duplicate_target_features)

# ------------------------------------------------------------
# 3. 데이터 전처리
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("3. 데이터 전처리")
print("=" * 60)

# 모든 설명 변수가 수치형인지 확인
if X_numeric.shape[1] == 0:
    raise ValueError("설명 변수가 없습니다.")

# 결측치 및 무한대 확인
X_numeric = X_numeric.replace([np.inf, -np.inf], np.nan)

valid_mask = (
    X_numeric.notna().all(axis=1)
    & Y_numeric.notna().all(axis=1)
)

removed_rows = int((~valid_mask).sum())

X = X_numeric.loc[valid_mask].copy()
Y = Y_numeric.loc[valid_mask].astype(int).copy()

print("결측치·무한대 포함으로 제외한 행:", removed_rows)

# 상수 변수 제거
constant_cols = X.columns[X.nunique() <= 1].tolist()
X = X.drop(columns=constant_cols)

# 타깃과 완전히 동일한 설명 변수 제거
X = X.drop(columns=duplicate_target_features, errors="ignore")

print("제거한 상수 변수:", constant_cols)
print("제거한 누수 의심 변수:", duplicate_target_features)
print("전처리 후 X:", X.shape)
print("전처리 후 Y:", Y.shape)

# 데이터 행 인덱스 정리
X = X.reset_index(drop=True)
Y = Y.reset_index(drop=True)

# 완전 중복 행은 자동 제거하지 않고 개수만 기록
print("원본 데이터 완전 중복 행 수:", df.duplicated().sum())

# ------------------------------------------------------------
# 4. 타깃 분포 분석
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("4. 타깃 분포 분석")
print("=" * 60)

label_distribution = pd.DataFrame({
    "발생 건수": Y.sum(),
    "발생 비율(%)": Y.mean() * 100,
    "미발생 건수": (Y == 0).sum()
}).round(2)

display(label_distribution)

label_distribution.to_csv(
    f"{OUTPUT_DIR}/target_distribution.csv",
    encoding="utf-8-sig"
)

plt.figure(figsize=(10, 5))
Y.sum().sort_values(ascending=False).plot(kind="bar")
plt.title("결함 유형별 발생 건수")
plt.xlabel("결함 유형")
plt.ylabel("발생 건수")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/target_distribution.png", dpi=150)
plt.show()

# 행별 동시 결함 라벨 개수
labels_per_row = Y.sum(axis=1)

print("행별 결함 라벨 개수:")
display(labels_per_row.value_counts().sort_index())

plt.figure(figsize=(7, 4))
labels_per_row.value_counts().sort_index().plot(kind="bar")
plt.title("행별 동시 결함 라벨 개수")
plt.xlabel("결함 라벨 개수")
plt.ylabel("데이터 수")
plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/labels_per_row.png", dpi=150)
plt.show()

# ------------------------------------------------------------
# 5. 설명 변수 EDA 및 상관관계
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("5. 설명 변수 EDA")
print("=" * 60)

display(X.describe().T.round(3))

corr = X.corr(method="spearman")

plt.figure(figsize=(14, 11))
plt.imshow(corr, aspect="auto", cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar(label="Spearman 상관계수")
plt.xticks(range(len(corr.columns)), corr.columns,
           rotation=90, fontsize=8)
plt.yticks(range(len(corr.index)), corr.index, fontsize=8)
plt.title("설명 변수 간 Spearman 상관관계")
plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/feature_correlation.png", dpi=150)
plt.show()

corr_pairs = corr.where(
    np.triu(np.ones(corr.shape), k=1).astype(bool)
).stack().reset_index()

corr_pairs.columns = ["변수 1", "변수 2", "상관계수"]
corr_pairs["절댓값"] = corr_pairs["상관계수"].abs()

corr_pairs = corr_pairs.sort_values("절댓값", ascending=False)

print("상관관계가 높은 변수 쌍:")
display(corr_pairs.head(15).round(4))

corr_pairs.to_csv(
    f"{OUTPUT_DIR}/feature_correlations.csv",
    index=False,
    encoding="utf-8-sig"
)

# ------------------------------------------------------------
# 6. 학습·검증·테스트 데이터 분할
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("6. 데이터 분할")
print("=" * 60)

# 학습 60%, 검증 20%, 테스트 20%
X_train, X_temp, Y_train, Y_temp = train_test_split(
    X, Y,
    test_size=0.4,
    random_state=RANDOM_STATE
)

X_valid, X_test, Y_valid, Y_test = train_test_split(
    X_temp, Y_temp,
    test_size=0.5,
    random_state=RANDOM_STATE
)

print("학습 데이터:", X_train.shape, Y_train.shape)
print("검증 데이터:", X_valid.shape, Y_valid.shape)
print("테스트 데이터:", X_test.shape, Y_test.shape)

split_distribution = pd.DataFrame({
    "전체(%)": Y.mean() * 100,
    "학습(%)": Y_train.mean() * 100,
    "검증(%)": Y_valid.mean() * 100,
    "테스트(%)": Y_test.mean() * 100
}).round(2)

display(split_distribution)

split_distribution.to_csv(
    f"{OUTPUT_DIR}/split_distribution.csv",
    encoding="utf-8-sig"
)

# ------------------------------------------------------------
# 7. 모델 학습 및 검증
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("7. 모델 학습 및 검증")
print("=" * 60)

models = {
    "Random Forest": MultiOutputClassifier(
        RandomForestClassifier(
            n_estimators=300,
            class_weight="balanced",
            random_state=RANDOM_STATE,
            n_jobs=-1
        )
    ),
    "Extra Trees": MultiOutputClassifier(
        ExtraTreesClassifier(
            n_estimators=300,
            class_weight="balanced",
            random_state=RANDOM_STATE,
            n_jobs=-1
        )
    ),
    "Logistic Regression": MultiOutputClassifier(
        LogisticRegression(
            max_iter=2000,
            class_weight="balanced",
            random_state=RANDOM_STATE
        )
    )
}

# Logistic Regression용 스케일링
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_valid_scaled = scaler.transform(X_valid)

results = []
fitted_models = {}
valid_predictions = {}

for name, model in models.items():
    start_time = time.time()

    if name == "Logistic Regression":
        model.fit(X_train_scaled, Y_train)
        pred = model.predict(X_valid_scaled)
    else:
        model.fit(X_train, Y_train)
        pred = model.predict(X_valid)

    elapsed = time.time() - start_time

    fitted_models[name] = model
    valid_predictions[name] = pred

    results.append({
        "모델": name,
        "Subset Accuracy": accuracy_score(Y_valid, pred),
        "Micro Precision": precision_score(
            Y_valid, pred, average="micro", zero_division=0
        ),
        "Micro Recall": recall_score(
            Y_valid, pred, average="micro", zero_division=0
        ),
        "Micro F1": f1_score(
            Y_valid, pred, average="micro", zero_division=0
        ),
        "Macro F1": f1_score(
            Y_valid, pred, average="macro", zero_division=0
        ),
        "Hamming Loss": hamming_loss(Y_valid, pred),
        "학습 시간(초)": round(elapsed, 3)
    })

results_df = pd.DataFrame(results).sort_values(
    "Macro F1", ascending=False
).reset_index(drop=True)

display(results_df.round(4))

results_df.to_csv(
    f"{OUTPUT_DIR}/model_comparison.csv",
    index=False,
    encoding="utf-8-sig"
)

# 검증 Macro F1 기준 최종 모델 선택
best_name = results_df.loc[0, "모델"]
best_model = fitted_models[best_name]

print("검증 기준 선택 모델:", best_name)

# ------------------------------------------------------------
# 8. 최종 테스트 평가
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("8. 최종 테스트 평가")
print("=" * 60)

if best_name == "Logistic Regression":
    X_test_input = scaler.transform(X_test)
else:
    X_test_input = X_test

test_pred = best_model.predict(X_test_input)

test_metrics = {
    "최종 모델": best_name,
    "Subset Accuracy": accuracy_score(Y_test, test_pred),
    "Micro Precision": precision_score(
        Y_test, test_pred, average="micro", zero_division=0
    ),
    "Micro Recall": recall_score(
        Y_test, test_pred, average="micro", zero_division=0
    ),
    "Micro F1": f1_score(
        Y_test, test_pred, average="micro", zero_division=0
    ),
    "Macro F1": f1_score(
        Y_test, test_pred, average="macro", zero_division=0
    ),
    "Hamming Loss": hamming_loss(Y_test, test_pred)
}

test_results = pd.DataFrame([test_metrics])
display(test_results.round(4))

test_results.to_csv(
    f"{OUTPUT_DIR}/final_test_results.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\n결함 유형별 분류 성능:")
report = classification_report(
    Y_test,
    test_pred,
    target_names=target_cols,
    output_dict=True,
    zero_division=0
)

display(pd.DataFrame(report).T.round(4))

# 라벨별 혼동행렬 저장
cms = multilabel_confusion_matrix(Y_test, test_pred)

for i, target in enumerate(target_cols):
    cm_df = pd.DataFrame(
        cms[i],
        index=["실제 0", "실제 1"],
        columns=["예측 0", "예측 1"]
    )

    print(f"\n[{target}] 혼동행렬")
    display(cm_df)

    cm_df.to_csv(
        f"{OUTPUT_DIR}/confusion_matrix_{target}.csv",
        encoding="utf-8-sig"
    )

# ------------------------------------------------------------
# 9. 변수 중요도 분석
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("9. 변수 중요도 분석")
print("=" * 60)

if best_name in ["Random Forest", "Extra Trees"]:

    importance_rows = []

    for target, estimator in zip(
        target_cols, best_model.estimators_
    ):
        for feature, importance in zip(
            X_train.columns,
            estimator.feature_importances_
        ):
            importance_rows.append({
                "결함 유형": target,
                "변수": feature,
                "중요도": importance
            })

    importance_df = pd.DataFrame(importance_rows)

    overall_importance = (
        importance_df.groupby("변수")["중요도"]
        .mean()
        .sort_values(ascending=False)
        .rename("평균 중요도")
        .reset_index()
    )

    display(overall_importance.head(15).round(4))

    overall_importance.to_csv(
        f"{OUTPUT_DIR}/feature_importance.csv",
        index=False,
        encoding="utf-8-sig"
    )

    plt.figure(figsize=(10, 6))
    overall_importance.head(15).sort_values(
        "평균 중요도"
    ).plot(
        x="변수",
        y="평균 중요도",
        kind="barh",
        legend=False
    )

    plt.title("평균 변수 중요도 상위 15개")
    plt.xlabel("평균 중요도")
    plt.ylabel("변수")
    plt.tight_layout()
    plt.savefig(f"{OUTPUT_DIR}/feature_importance.png", dpi=150)
    plt.show()

elif best_name == "Logistic Regression":

    coef_rows = []

    for target, estimator in zip(
        target_cols, best_model.estimators_
    ):
        for feature, coef in zip(
            X_train.columns, estimator.coef_[0]
        ):
            coef_rows.append({
                "결함 유형": target,
                "변수": feature,
                "계수": coef,
                "계수 절댓값": abs(coef)
            })

    coef_df = pd.DataFrame(coef_rows)

    display(
        coef_df.sort_values(
            "계수 절댓값", ascending=False
        ).head(20).round(4)
    )

    coef_df.to_csv(
        f"{OUTPUT_DIR}/logistic_coefficients.csv",
        index=False,
        encoding="utf-8-sig"
    )

# ------------------------------------------------------------
# 10. 오분류 사례 저장
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("10. 오분류 사례 분석")
print("=" * 60)

# 행별 실제 라벨과 예측 라벨 비교
error_mask = (Y_test.to_numpy() != test_pred).any(axis=1)

error_rows = []

for position in np.where(error_mask)[0]:
    actual_labels = [
        target_cols[j]
        for j, value in enumerate(Y_test.iloc[position])
        if value == 1
    ]

    predicted_labels = [
        target_cols[j]
        for j, value in enumerate(test_pred[position])
        if value == 1
    ]

    error_rows.append({
        "테스트 행 인덱스": int(Y_test.index[position]),
        "실제 결함": ", ".join(actual_labels) if actual_labels else "없음",
        "예측 결함": ", ".join(predicted_labels) if predicted_labels else "없음",
        "불일치 라벨 수": int(
            (Y_test.iloc[position].to_numpy()
             != test_pred[position]).sum()
        )
    })

errors_df = pd.DataFrame(error_rows)

print("오분류 행 수:", len(errors_df))
display(errors_df.head(20))

errors_df.to_csv(
    f"{OUTPUT_DIR}/misclassified_cases.csv",
    index=False,
    encoding="utf-8-sig"
)

# ------------------------------------------------------------
# 11. 분석 결과 및 실행 정보 저장
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("11. 결과 저장")
print("=" * 60)

run_info = {
    "data_path": DATA_PATH,
    "original_rows": int(len(df)),
    "original_columns": int(df.shape[1]),
    "processed_rows": int(len(X)),
    "feature_count": int(X.shape[1]),
    "target_columns": target_cols,
    "train_rows": int(len(X_train)),
    "validation_rows": int(len(X_valid)),
    "test_rows": int(len(X_test)),
    "random_state": RANDOM_STATE,
    "selected_model": best_name,
    "selection_metric": "Validation Macro F1",
    "test_metrics": {
        key: float(value)
        for key, value in test_metrics.items()
        if key != "최종 모델"
    },
    "excluded_constant_features": constant_cols,
    "excluded_target_duplicate_features": duplicate_target_features
}

with open(
    f"{OUTPUT_DIR}/run_info.json",
    "w",
    encoding="utf-8"
) as file:
    json.dump(run_info, file, ensure_ascii=False, indent=4)

# 테스트 예측 결과 저장
prediction_df = Y_test.reset_index(drop=True).add_prefix("실제_")
predicted_df = pd.DataFrame(
    test_pred, columns=[f"예측_{col}" for col in target_cols]
)

pd.concat(
    [prediction_df, predicted_df],
    axis=1
).to_csv(
    f"{OUTPUT_DIR}/test_predictions.csv",
    index=False,
    encoding="utf-8-sig"
)

print("최종 모델:", best_name)
print("결과 저장 폴더:", OUTPUT_DIR)
print("생성 파일 수 및 내용은 해당 폴더에서 확인하세요.")
print("전체 분석 완료")
```
