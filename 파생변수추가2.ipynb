```python
#claude

# ============================================================
# 강판 표면 결함 분류 - 파생변수 확장 통합 버전
# ============================================================
# [기존 통합 코드]
#   - 5-Fold CV 하이퍼파라미터 튜닝 (class_weight 포함)
#   - 원-핫 검증 후 열 병합, 강종-결함 카이제곱 검정
#   - 소프트 보팅 앙상블 + OOF 기반 클래스별 확률 보정
#   - Permutation Importance, 오분류 분석, 부트스트랩 신뢰구간
#   - 전체 데이터 재학습 + 새 데이터 예측 함수
#
# [이번에 추가한 것]
#   - EDA(결함별 크기·밝기·모양·강종·두께·위치 특성) 기반 파생변수 5개 그룹
#       형상 / 밝기 / 위치·가장자리 / 강종·두께 / 상호작용
#   - 원본과 사실상 같은 정보인 파생변수 자동 제거 (Spearman |rho| > 0.995)
#   - 그룹 단위 반복 교차검증으로 채택 여부 결정 (개발 데이터만 사용)
#   - 결함 유형별 파생변수 중앙값 확인 (EDA 표와 방향이 맞는지 점검)
#
# 분할: 개발(학습+교차검증) 80% / 테스트 20%
# ============================================================


# ============================================================
# 0. 라이브러리 및 설정
# ============================================================

import time
import warnings

import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

try:
    from IPython.display import display
except ImportError:
    display = print

from scipy.stats import chi2_contingency
from sklearn.base import BaseEstimator, ClassifierMixin, clone
from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    RepeatedStratifiedKFold,
    RandomizedSearchCV,
    cross_val_score,
    cross_val_predict,
)
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.inspection import permutation_importance
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
)
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier

warnings.filterwarnings("ignore")
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)

plt.rcParams["font.family"] = "Malgun Gothic"   # macOS: "AppleGothic"
plt.rcParams["axes.unicode_minus"] = False

RANDOM_STATE = 42
DATA_PATH = "steel_plates_faults.csv"
N_ITER = 20             # 모델별 무작위 탐색 조합 수 (시간이 부족하면 줄이기)
REDUNDANT_RHO = 0.995   # 이 이상 순위상관이면 같은 정보로 보고 제거

STEEL_COLS = ["TypeOfSteel_A300", "TypeOfSteel_A400"]
FAULT_COLS = ["Pastry", "Z_Scratch", "K_Scatch", "Stains",
              "Dirtiness", "Bumps", "Other_Faults"]
TARGET = "Fault_Type"

LABELS = list(range(len(FAULT_COLS)))
class_to_int = {name: i for i, name in enumerate(FAULT_COLS)}
int_to_class = {i: name for name, i in class_to_int.items()}
STEEL_CODE = {"A300": 0, "A400": 1}


def macro_f1(y_true, y_pred):
    return f1_score(y_true, y_pred, labels=LABELS, average="macro", zero_division=0)


def score_row(y_true, y_pred):
    return {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Balanced Acc": balanced_accuracy_score(y_true, y_pred),
        "Macro Precision": precision_score(y_true, y_pred, labels=LABELS,
                                           average="macro", zero_division=0),
        "Macro Recall": recall_score(y_true, y_pred, labels=LABELS,
                                     average="macro", zero_division=0),
        "Macro F1": macro_f1(y_true, y_pred),
    }


# ============================================================
# 1. 데이터 불러오기 및 열 병합 (원-핫 구조 검증 포함)
# ============================================================

raw_df = pd.read_csv(DATA_PATH)
print("원본 데이터 크기:", raw_df.shape)
print("결측치:", raw_df.isna().sum().sum(), "/ 중복 행:", raw_df.duplicated().sum())

for cols, label in [(STEEL_COLS, "강종"), (FAULT_COLS, "결함 유형")]:
    missing = [c for c in cols if c not in raw_df.columns]
    if missing:
        raise ValueError(f"{label} 열이 없습니다: {missing}")
    if not raw_df[cols].isin([0, 1]).all().all():
        raise ValueError(f"{label} 열은 0 또는 1이어야 합니다.")
    if not (raw_df[cols].sum(axis=1) == 1).all():
        raise ValueError(f"{label}이(가) 정확히 하나로 정해지지 않는 행이 있습니다.")


def merge_columns(data: pd.DataFrame) -> pd.DataFrame:
    """강종 2열 → Steel_Type, 결함 7열 → Fault_Type (결함 열이 없으면 강종만)"""
    out = data.copy()
    if set(STEEL_COLS).issubset(out.columns):
        pos = out.columns.get_loc("TypeOfSteel_A300")
        steel = np.where(out["TypeOfSteel_A300"] == 1, "A300", "A400")
        out = out.drop(columns=STEEL_COLS)
        out.insert(pos, "Steel_Type", steel)
    if set(FAULT_COLS).issubset(out.columns):
        out[TARGET] = out[FAULT_COLS].idxmax(axis=1)
        out = out.drop(columns=FAULT_COLS)
    return out


df = merge_columns(raw_df)
print("병합 후 데이터 크기:", df.shape)

print("\n[결함 유형 분포]")
display(df[TARGET].value_counts().reindex(FAULT_COLS).to_frame("count"))


# ============================================================
# 2. 강종 - 결함 유형 관계 (카이제곱 검정, Cramér's V)
# ============================================================

ct = pd.crosstab(df[TARGET], df["Steel_Type"]).reindex(FAULT_COLS)
chi2, p_value, dof, _ = chi2_contingency(ct)
cramers_v = np.sqrt(chi2 / (len(df) * (min(ct.shape) - 1)))

print("\n[결함 유형 × 강종]")
display(ct)
print(f"Chi2 = {chi2:.1f}, df = {dof}, p = "
      f"{'<0.001' if p_value < 0.001 else round(p_value, 4)}, Cramér's V = {cramers_v:.3f}")


# ============================================================
# 3. 입력 변수 생성 (원본 + 파생변수 후보 전체)
# ============================================================
# 트리 모델은 단조 변환(로그, 스케일 변경)에서 새 정보를 얻지 못하므로
# 변수 2개 이상을 조합해 원본에 없는 정보가 생기는 것만 후보로 둔다.
# 실제 사용 여부는 5단계에서 개발 데이터 교차검증으로 결정한다.
#
# 두께 구간은 EDA 표의 경계를 사용했다. 표가 전체 데이터 기준이므로
# 테스트 정보가 아주 약간 섞인다 (구간이 거칠어 영향은 작음).
# ============================================================

THICK_BINS = [-np.inf, 40, 60, 70, 80, 150, np.inf]   # 40 / 41~60 / 61~70 / 71~80 / 81~150 / 151~

FEATURE_GROUPS = {
    # K_Scatch 가로로 김, Dirtiness/Pastry 세로로 김, Bumps 정사각형에 가까움
    "형상": ["Log_Aspect_Ratio", "Elongation", "Compactness",
             "Perimeter_Area_Ratio", "Boundary_Complexity", "Fill_Ratio"],
    # K_Scatch 최소 밝기 매우 낮음, Stains 가장 밝고 범위 좁음, Pastry 평균 가장 어두움
    "밝기": ["Mean_Luminosity", "Luminosity_Range", "Dark_Depth",
             "Lum_Mean_Position", "Lum_Contrast"],
    # K_Scatch Edges_Y 낮음·가장자리, Stains Edges_X ≈ 1·판 안쪽, Z_Scratch 가장자리 근처
    "위치·가장자리": ["Edge_XY_Diff", "X_Center_Rel", "Edge_Dist_Rel"],
    # Z_Scratch = A300 + 61~70, K_Scatch = A400 + 40, Stains = A400 + 41~60 ...
    "강종·두께": ["Steel_Thick_Group"],
    # 크고 어두움(K_Scatch) vs 작고 밝음(Stains), 길쭉+어두움(Pastry) vs 길쭉+밝음(Dirtiness)
    "상호작용": ["LogArea_x_LumIdx", "LogArea_x_MinLum", "Elong_x_Darkness"],
}
ALL_DERIVED = [c for cols in FEATURE_GROUPS.values() for c in cols]


def build_features(data: pd.DataFrame) -> pd.DataFrame:
    """원본 입력 변수 + 파생변수 후보 전체를 반환 (사용할 열은 호출하는 쪽에서 선택)"""
    out = data.drop(columns=[TARGET], errors="ignore").copy()
    out["Steel_Type"] = out["Steel_Type"].map(STEEL_CODE)
    if out["Steel_Type"].isna().any():
        raise ValueError("Steel_Type에 A300 / A400 이외의 값이 있습니다.")

    x_len = (out["X_Maximum"] - out["X_Minimum"] + 1).clip(lower=1)
    y_len = (out["Y_Maximum"] - out["Y_Minimum"] + 1).clip(lower=1)
    area = out["Pixels_Areas"].clip(lower=1)
    perim = (out["X_Perimeter"] + out["Y_Perimeter"]).clip(lower=1)
    lum_min = out["Minimum_of_Luminosity"]
    lum_max = out["Maximum_of_Luminosity"]
    lum_mean = out["Sum_of_Luminosity"] / area
    lum_range = lum_max - lum_min
    conveyer = out["Length_of_Conveyer"].clip(lower=1)

    d = {}
    # --- 형상 ---
    d["Log_Aspect_Ratio"] = np.log(x_len / y_len)              # +: 가로로 김, -: 세로로 김
    d["Elongation"] = np.maximum(x_len, y_len) / np.minimum(x_len, y_len)
    d["Compactness"] = 4 * np.pi * area / perim ** 2           # 원형에 가까울수록 큼
    d["Perimeter_Area_Ratio"] = perim / area
    d["Boundary_Complexity"] = perim / (2 * (x_len + y_len))   # 1 = 매끈한 사각형
    d["Fill_Ratio"] = area / (x_len * y_len)                   # Empty_Index와 중복 가능 → 자동 점검

    # --- 밝기 ---
    d["Mean_Luminosity"] = lum_mean                            # Luminosity_Index와 중복 가능 → 자동 점검
    d["Luminosity_Range"] = lum_range
    d["Dark_Depth"] = lum_mean - lum_min                       # 평균 대비 가장 어두운 부분의 깊이
    d["Lum_Mean_Position"] = (lum_mean - lum_min) / (lum_range + 1)
    d["Lum_Contrast"] = lum_range / (lum_mean + 1)

    # --- 위치·가장자리 ---
    d["Edge_XY_Diff"] = out["Edges_X_Index"] - out["Edges_Y_Index"]
    d["X_Center_Rel"] = ((out["X_Minimum"] + out["X_Maximum"]) / 2) / conveyer
    d["Edge_Dist_Rel"] = np.clip(
        np.minimum(out["X_Minimum"], conveyer - out["X_Maximum"]), 0, None) / conveyer

    # --- 강종·두께 ---
    thick_bin = pd.cut(out["Steel_Plate_Thickness"], THICK_BINS, labels=False)
    d["Steel_Thick_Group"] = out["Steel_Type"] * 10 + thick_bin

    # --- 상호작용 ---
    d["LogArea_x_LumIdx"] = out["LogOfAreas"] * out["Luminosity_Index"]
    d["LogArea_x_MinLum"] = out["LogOfAreas"] * (lum_min / 255)
    d["Elong_x_Darkness"] = np.log(d["Elongation"]) * (1 - lum_mean / 255)

    out = pd.concat([out, pd.DataFrame(d, index=out.index)], axis=1)
    out = out.replace([np.inf, -np.inf], np.nan)
    if out.isna().any().any():
        bad = out.columns[out.isna().any()].tolist()
        print(f"[경고] 결측/무한값 발생 열 {bad} → 0으로 대체")
        out = out.fillna(0)
    return out


y_all = df[TARGET].map(class_to_int).to_numpy()
X_full = build_features(df)
BASE_COLS = [c for c in X_full.columns if c not in ALL_DERIVED]
print(f"\n원본 입력 변수 {len(BASE_COLS)}개 / 파생변수 후보 {len(ALL_DERIVED)}개")

# Edge_Dist_Rel / X_Center_Rel 가정 점검: Length_of_Conveyer를 판 폭으로 봐도 되는지
inside = (df["X_Maximum"] <= df["Length_of_Conveyer"]).mean()
print(f"X_Maximum ≤ Length_of_Conveyer 인 비율: {inside:.1%} "
      "(낮으면 위치 상대값 가정이 맞지 않음)")


# ============================================================
# 4. 개발 / 테스트 분할 (80 / 20, 층화)
# ============================================================

idx_dev, idx_test = train_test_split(
    np.arange(len(df)), test_size=0.20, stratify=y_all, random_state=RANDOM_STATE
)
y_dev, y_test = y_all[idx_dev], y_all[idx_test]
print(f"\n개발: {len(idx_dev)}개 / 테스트: {len(idx_test)}개")

cv5 = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
cv_rep = RepeatedStratifiedKFold(n_splits=5, n_repeats=3, random_state=RANDOM_STATE)

X_full_dev = X_full.iloc[idx_dev]


# ============================================================
# 5. 파생변수 선택 (개발 데이터만 사용)
# ============================================================

# 5-1. 결함 유형별 파생변수 중앙값 (EDA 표와 방향이 맞는지 눈으로 확인)
print("\n[결함 유형별 파생변수 중앙값 - 개발 데이터]")
check = X_full_dev[ALL_DERIVED].copy()
check[TARGET] = [int_to_class[i] for i in y_dev]
display(check.groupby(TARGET).median().reindex(FAULT_COLS).T.round(3))

# 5-2. 중복 제거: 원본 또는 앞서 남긴 파생변수와 순위상관이 거의 1이면 제외
#      (트리 모델 입장에서 같은 변수이므로 분할 후보만 늘리고 정보는 없음)
rho = X_full_dev.corr(method="spearman").abs()
kept_derived, dropped = [], []
for col in ALL_DERIVED:
    ref = BASE_COLS + kept_derived
    best_ref = rho.loc[col, ref].idxmax()
    if rho.loc[col, best_ref] > REDUNDANT_RHO:
        dropped.append({"Feature": col, "Same as": best_ref,
                        "|rho|": round(rho.loc[col, best_ref], 4)})
    else:
        kept_derived.append(col)

print("\n[중복으로 제외된 파생변수]")
display(pd.DataFrame(dropped) if dropped else "없음")

groups_kept = {g: [c for c in cols if c in kept_derived]
               for g, cols in FEATURE_GROUPS.items()}
groups_kept = {g: cols for g, cols in groups_kept.items() if cols}

# 5-3. 그룹별 반복 교차검증 (같은 fold를 써서 비교 조건을 맞춤)
probe = RandomForestClassifier(n_estimators=300, random_state=RANDOM_STATE, n_jobs=-1)


def cv_f1(cols):
    s = cross_val_score(probe, X_full_dev[cols], y_dev, cv=cv_rep,
                        scoring="f1_macro", n_jobs=-1)
    return s.mean(), s.std()


base_mean, base_sd = cv_f1(BASE_COLS)
group_rows = [{"구성": "원본만", "변수 수": len(BASE_COLS),
               "CV Macro F1": base_mean, "SD": base_sd, "이득": 0.0, "채택": "-"}]
group_scores = {}

for g, cols in groups_kept.items():
    m, s = cv_f1(BASE_COLS + cols)
    gain = m - base_mean
    adopt = gain > 0.5 * base_sd          # 기존 코드와 같은 기준
    group_scores[g] = (m, adopt)
    group_rows.append({"구성": f"원본 + {g}", "변수 수": len(BASE_COLS) + len(cols),
                       "CV Macro F1": m, "SD": s, "이득": gain,
                       "채택": "O" if adopt else "X"})

selected_groups = [g for g, (_, adopt) in group_scores.items() if adopt]

# 5-4. 채택 그룹 결합 확인: 합친 것이 최고 단일 그룹보다 못하면 단일 그룹만 사용
if len(selected_groups) >= 2:
    comb_cols = [c for g in selected_groups for c in groups_kept[g]]
    m, s = cv_f1(BASE_COLS + comb_cols)
    group_rows.append({"구성": "원본 + " + " + ".join(selected_groups),
                       "변수 수": len(BASE_COLS) + len(comb_cols),
                       "CV Macro F1": m, "SD": s, "이득": m - base_mean, "채택": "결합"})
    best_single = max(selected_groups, key=lambda g: group_scores[g][0])
    if m < group_scores[best_single][0]:
        print(f"→ 결합 성능이 단일 그룹({best_single})보다 낮아 단일 그룹만 사용")
        selected_groups = [best_single]

print("\n[파생변수 그룹 평가 - 개발 데이터 5-Fold × 3회]")
display(pd.DataFrame(group_rows).round(4))

DERIVED_USED = [c for g in selected_groups for c in groups_kept[g]]
print("→ 채택 그룹:", selected_groups if selected_groups else "없음 (원본만 사용)")
print("→ 사용 파생변수:", DERIVED_USED)

feature_cols = BASE_COLS + DERIVED_USED
X_all = X_full[feature_cols]
X_dev, X_test = X_all.iloc[idx_dev], X_all.iloc[idx_test]


# ============================================================
# 6. 모델 및 탐색 공간
# ============================================================
# 선형/거리 기반 모델은 트리 계열보다 Macro F1이 0.07~0.17 낮았으므로 트리 계열만 사용.
# class_weight는 모든 모델에서 튜닝 대상이다.
# 파생변수로 변수 수가 늘었으므로 max_features 후보는 비율 위주로 둔다.
# ============================================================

models = {
    "Random Forest": RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=1),
    "Extra Trees": ExtraTreesClassifier(random_state=RANDOM_STATE, n_jobs=1),
    "LightGBM": LGBMClassifier(objective="multiclass", random_state=RANDOM_STATE,
                               n_jobs=1, verbose=-1),
    "CatBoost": CatBoostClassifier(loss_function="MultiClass", random_seed=RANDOM_STATE,
                                   verbose=0, allow_writing_files=False, thread_count=1),
}

param_spaces = {
    "Random Forest": {
        "n_estimators": [300, 500, 700],
        "max_depth": [None, 10, 15, 20, 30],
        "min_samples_split": [2, 5, 9],
        "min_samples_leaf": [1, 2, 4],
        "max_features": ["sqrt", 0.3, 0.5, 0.8],
        "class_weight": [None, "balanced", "balanced_subsample"],
    },
    "Extra Trees": {
        "n_estimators": [300, 500, 700],
        "max_depth": [None, 10, 15, 20, 30],
        "min_samples_split": [2, 5, 9],
        "min_samples_leaf": [1, 2, 4],
        "max_features": ["sqrt", 0.3, 0.5, 0.8],
        "class_weight": [None, "balanced", "balanced_subsample"],
    },
    "LightGBM": {
        "n_estimators": [200, 400, 700, 1000],
        "learning_rate": [0.02, 0.03, 0.05, 0.1],
        "num_leaves": [7, 15, 22, 31],
        "max_depth": [-1, 5, 7, 10],
        "min_child_samples": [5, 10, 20, 33],
        "subsample": [0.6, 0.8, 1.0],
        "subsample_freq": [1],
        "colsample_bytree": [0.5, 0.6, 0.8, 1.0],
        "reg_lambda": [0, 0.01, 1, 5],
        "class_weight": [None, "balanced"],
    },
    "CatBoost": {
        "iterations": [300, 500, 700],
        "depth": [4, 5, 6, 7],
        "learning_rate": [0.03, 0.05, 0.1],
        "l2_leaf_reg": [1, 3, 5, 7],
        "auto_class_weights": [None, "Balanced", "SqrtBalanced"],
    },
}


# ============================================================
# 7. 하이퍼파라미터 튜닝 (개발 데이터 5-Fold, Macro F1)
# ============================================================

tuned = {}
tuning_rows = []

for name, model in models.items():
    print(f"\n{name} 튜닝 중... ({N_ITER}개 조합 × 5-Fold)")
    start = time.time()
    search = RandomizedSearchCV(
        model, param_spaces[name], n_iter=N_ITER, scoring="f1_macro",
        cv=cv5, random_state=RANDOM_STATE, n_jobs=-1, refit=False,
    )
    search.fit(X_dev, y_dev)
    tuned[name] = clone(model).set_params(**search.best_params_)

    tuning_rows.append({
        "Model": name,
        "CV Macro F1": search.best_score_,
        "CV SD": search.cv_results_["std_test_score"][search.best_index_],
        "Best Params": search.best_params_,
        "Time(s)": round(time.time() - start, 1),
    })
    print(f"  CV Macro F1 = {search.best_score_:.4f} | {search.best_params_}")

tuning_df = (pd.DataFrame(tuning_rows)
             .sort_values("CV Macro F1", ascending=False).reset_index(drop=True))
print("\n[튜닝 결과]")
display(tuning_df)


# ============================================================
# 8. Out-of-fold 확률 예측 및 후보 비교
# ============================================================
# 앙상블 조합과 클래스별 확률 보정은 OOF 확률로만 결정하고 테스트는 건드리지 않는다.
# (튜닝에 쓴 fold와 같으므로 약간 낙관적일 수 있음 → 최종 판단은 테스트에서)
# ============================================================

oof = {}
for name, est in tuned.items():
    oof[name] = cross_val_predict(
        clone(est), X_dev, y_dev, cv=cv5, method="predict_proba", n_jobs=-1
    )


def tune_class_scale(P, y, grid=np.round(np.arange(0.5, 3.01, 0.1), 2), n_rounds=3):
    """
    예측 확률에 클래스별 배율을 곱해 argmax 했을 때 Macro F1이 최대가 되는 배율 탐색.
    배율 > 1 이면 그 클래스를 더 자주 예측(재현율↑ 정밀도↓), < 1 이면 반대.
    """
    w = np.ones(P.shape[1])
    best = macro_f1(y, P.argmax(axis=1))
    for _ in range(n_rounds):
        improved = False
        for k in range(P.shape[1]):
            for g in grid:
                trial = w.copy()
                trial[k] = g
                s = macro_f1(y, (P * trial).argmax(axis=1))
                if s > best + 1e-9:
                    best, w, improved = s, trial, True
        if not improved:
            break
    return w, best


ranked = tuning_df["Model"].tolist()
candidates = {name: [name] for name in ranked}
candidates["앙상블(상위 2)"] = ranked[:2]
candidates["앙상블(상위 3)"] = ranked[:3]
candidates["앙상블(전체 4)"] = ranked

cand_rows = []
cand_scale = {}
for cname, members in candidates.items():
    P = np.mean([oof[m] for m in members], axis=0)
    raw = macro_f1(y_dev, P.argmax(axis=1))
    w, scaled = tune_class_scale(P, y_dev)
    cand_scale[cname] = w
    cand_rows.append({"Candidate": cname, "OOF Macro F1": raw,
                      "OOF Macro F1 (확률 보정)": scaled})

cand_df = (pd.DataFrame(cand_rows)
           .sort_values("OOF Macro F1 (확률 보정)", ascending=False).reset_index(drop=True))
print("\n[후보별 Out-of-fold 성능]")
display(cand_df.round(4))


# ============================================================
# 9. 최종 후보 선택
# ============================================================
# 보정 이득이 0.005 미만이면 보정 없이 사용한다.
# ============================================================

class ScaledSoftVoting(BaseEstimator, ClassifierMixin):
    """여러 모델의 확률 평균 + 클래스별 배율 보정"""

    def __init__(self, estimators, class_scale=None):
        self.estimators = estimators
        self.class_scale = class_scale

    def fit(self, X, y):
        self.estimators_ = [clone(e).fit(X, y) for e in self.estimators]
        self.classes_ = np.unique(y)
        return self

    def predict_proba(self, X):
        P = np.mean([e.predict_proba(X) for e in self.estimators_], axis=0)
        if self.class_scale is not None:
            P = P * np.asarray(self.class_scale)
            P = P / P.sum(axis=1, keepdims=True)
        return P

    def predict(self, X):
        return self.classes_[self.predict_proba(X).argmax(axis=1)]


best_row = cand_df.iloc[0]
best_cand = best_row["Candidate"]
use_scale = (best_row["OOF Macro F1 (확률 보정)"] - best_row["OOF Macro F1"]) >= 0.005
final_scale = cand_scale[best_cand] if use_scale else None

print("\n최종 후보:", best_cand, "| 구성:", candidates[best_cand])
print("확률 보정 사용:", use_scale)
if use_scale:
    display(pd.Series(final_scale, index=FAULT_COLS, name="클래스 배율").round(2))

final_model = ScaledSoftVoting(
    [tuned[m] for m in candidates[best_cand]], class_scale=final_scale
).fit(X_dev, y_dev)


# ============================================================
# 10. 테스트 평가 (마지막 1회)
# ============================================================

y_pred = final_model.predict(X_test)
test_scores = score_row(y_test, y_pred)

print("\n" + "=" * 60)
print(f"최종 테스트 결과 - {best_cand}{' + 확률 보정' if use_scale else ''}"
      f" | 파생변수 {len(DERIVED_USED)}개")
print("=" * 60)
display(pd.DataFrame([test_scores]).round(4))

rng = np.random.default_rng(RANDOM_STATE)
boot = []
for _ in range(1000):
    b = rng.integers(0, len(y_test), len(y_test))
    boot.append(macro_f1(y_test[b], y_pred[b]))
lo, hi = np.percentile(boot, [2.5, 97.5])
print(f"테스트 Macro F1 95% 신뢰구간: [{lo:.3f}, {hi:.3f}]")
print("  → 이전 버전과의 점수 차이가 이 구간 폭보다 작으면 우열을 단정하기 어렵다.")

print("\n[Classification Report - Test]")
print(classification_report(y_test, y_pred, labels=LABELS,
                            target_names=FAULT_COLS, digits=4, zero_division=0))

# 참고용: 후보별 테스트 성능 + 파생변수 없는 같은 최종 구성 (선택을 바꾸는 데 쓰지 않는다)
ref_rows = []
for cname, members in candidates.items():
    m = ScaledSoftVoting([tuned[x] for x in members]).fit(X_dev, y_dev)
    ref_rows.append({"Candidate": cname, "Test Macro F1 (보정 없음)":
                     macro_f1(y_test, m.predict(X_test))})
if DERIVED_USED:
    base_model = ScaledSoftVoting([tuned[m] for m in candidates[best_cand]],
                                  class_scale=final_scale)
    base_model.fit(X_full[BASE_COLS].iloc[idx_dev], y_dev)
    ref_rows.append({"Candidate": f"{best_cand} (원본 변수만)",
                     "Test Macro F1 (보정 없음)":
                     macro_f1(y_test, base_model.predict(X_full[BASE_COLS].iloc[idx_test]))})
print("[참고] 후보별 테스트 Macro F1")
display(pd.DataFrame(ref_rows).round(4))


# ============================================================
# 11. 혼동행렬 및 오분류 분석
# ============================================================

fig, axes = plt.subplots(1, 2, figsize=(18, 7))
ConfusionMatrixDisplay(confusion_matrix(y_test, y_pred, labels=LABELS),
                       display_labels=FAULT_COLS).plot(
    ax=axes[0], xticks_rotation=45, colorbar=False)
axes[0].set_title("혼동행렬 (개수)")
ConfusionMatrixDisplay(confusion_matrix(y_test, y_pred, labels=LABELS, normalize="true"),
                       display_labels=FAULT_COLS).plot(
    ax=axes[1], xticks_rotation=45, colorbar=False, values_format=".2f")
axes[1].set_title("혼동행렬 (실제 유형 기준 비율)")
for ax in axes:
    ax.set_xlabel("예측 결함 유형")
    ax.set_ylabel("실제 결함 유형")
plt.tight_layout()
plt.show()

test_result = pd.DataFrame({
    "Actual": [int_to_class[i] for i in y_test],
    "Predicted": [int_to_class[i] for i in y_pred],
    "Confidence": final_model.predict_proba(X_test).max(axis=1),
}, index=X_test.index)
misclassified = test_result[test_result["Actual"] != test_result["Predicted"]]

print(f"\n오분류: {len(misclassified)} / {len(y_test)} ({len(misclassified) / len(y_test):.2%})")
display(misclassified.groupby(["Actual", "Predicted"]).size()
        .sort_values(ascending=False).head(10).to_frame("Count"))


# ============================================================
# 12. 변수 중요도 (Permutation Importance, 테스트 데이터)
# ============================================================
# 상관이 높은 변수끼리는 서로 정보를 대신하므로 개별 중요도가 낮게 나올 수 있다.
# ============================================================

perm = permutation_importance(final_model, X_test, y_test, scoring="f1_macro",
                              n_repeats=20, random_state=RANDOM_STATE, n_jobs=-1)
importance_df = (pd.DataFrame({
    "Feature": feature_cols,
    "구분": ["파생" if c in DERIVED_USED else "원본" for c in feature_cols],
    "Importance Mean": perm.importances_mean,
    "Importance SD": perm.importances_std,
}).sort_values("Importance Mean", ascending=False).reset_index(drop=True))
display(importance_df.head(20).round(4))

top = importance_df.head(15).iloc[::-1]
colors = ["tab:orange" if g == "파생" else "tab:blue" for g in top["구분"]]
plt.figure(figsize=(9, 7))
plt.barh(top["Feature"], top["Importance Mean"], xerr=top["Importance SD"], color=colors)
plt.xlabel("섞었을 때 Macro F1 감소량")
plt.title("Permutation Importance 상위 15개 (주황 = 파생변수)")
plt.tight_layout()
plt.show()


# ============================================================
# 13. 전체 데이터로 재학습 및 저장
# ============================================================

deploy_model = ScaledSoftVoting(
    [tuned[m] for m in candidates[best_cand]], class_scale=final_scale
).fit(X_all, y_all)

joblib.dump({
    "model": deploy_model,
    "candidate": best_cand,
    "members": candidates[best_cand],
    "params": {m: tuned[m].get_params() for m in candidates[best_cand]},
    "class_scale": None if final_scale is None else final_scale.tolist(),
    "derived_groups": selected_groups,
    "derived_features": DERIVED_USED,
    "feature_cols": feature_cols,
    "class_names": FAULT_COLS,
    "test_scores": test_scores,
}, "steel_fault_model_derived.joblib")
print("\n저장 완료: steel_fault_model_derived.joblib")


# ============================================================
# 14. 새 데이터 예측 함수
# ============================================================
# 원본 형식(TypeOfSteel_A300/A400) 또는 병합 형식(Steel_Type) 모두 입력 가능.
# ScaledSoftVoting, build_features가 정의된 환경에서 불러와야 한다.
# ============================================================

def predict_faults(new_data: pd.DataFrame,
                   model_path: str = "steel_fault_model_derived.joblib") -> pd.DataFrame:
    bundle = joblib.load(model_path)
    names = bundle["class_names"]
    merged = merge_columns(new_data.drop(columns=names, errors="ignore"))
    X_new = build_features(merged)[bundle["feature_cols"]]

    proba = bundle["model"].predict_proba(X_new)
    out = pd.DataFrame(proba, columns=[f"prob_{n}" for n in names], index=new_data.index)
    out.insert(0, "predicted_fault", [names[i] for i in proba.argmax(axis=1)])
    out.insert(1, "confidence", proba.max(axis=1))
    return out


demo = predict_faults(raw_df.iloc[idx_test[:5]])
demo.insert(0, "actual_fault", df.iloc[idx_test[:5]][TARGET].values)
print("\n[예측 함수 동작 확인 - 전체 데이터로 재학습한 모델이므로 성능 지표 아님]")
display(demo.round(3))
```

```python
# ============================================================
# 11-1. 오분류 행 추출
# ============================================================

proba_df = pd.DataFrame(final_model.predict_proba(X_test),
                        columns=[f"prob_{n}" for n in FAULT_COLS], index=X_test.index)

mis_rows = test_result.loc[misclassified.index].join(proba_df).join(X_test)
mis_rows.insert(2, "Pair", mis_rows["Actual"] + " → " + mis_rows["Predicted"])

# 정답 클래스에 준 확률, 예측 확률과의 차이(작을수록 아깝게 틀린 것)
mis_rows.insert(4, "Prob_Actual",
                [mis_rows.at[i, f"prob_{a}"] for i, a in zip(mis_rows.index, mis_rows["Actual"])])
mis_rows.insert(5, "Margin", mis_rows["Confidence"] - mis_rows["Prob_Actual"])

# 조합 빈도순 → 조합 안에서는 아깝게 틀린 순
pair_order = mis_rows["Pair"].value_counts().index
mis_rows["Pair"] = pd.Categorical(mis_rows["Pair"], categories=pair_order, ordered=True)
mis_rows = mis_rows.sort_values(["Pair", "Margin"])

display(mis_rows.round(3))
mis_rows.to_csv("misclassified_rows.csv", encoding="utf-8-sig")

# 특정 조합만 보기
display(mis_rows[mis_rows["Pair"] == "Other_Faults → Bumps"].round(3))
```

```python
def compare_pair(actual, predicted, cols=None):
    """실제 actual인데 predicted로 틀린 행 vs 각 클래스를 맞힌 행의 중앙값 비교"""
    cols = cols or feature_cols
    correct = test_result["Actual"] == test_result["Predicted"]
    groups = {
        f"{actual}→{predicted} (오분류)": (test_result["Actual"] == actual) & (test_result["Predicted"] == predicted),
        f"{actual} (정답)": correct & (test_result["Actual"] == actual),
        f"{predicted} (정답)": correct & (test_result["Actual"] == predicted),
    }
    table = pd.DataFrame({k: X_test.loc[m, cols].median() for k, m in groups.items()})
    table["n"] = None
    table.loc["n", :] = [m.sum() for m in groups.values()] + [None]
    return table.drop(columns="n")

top_feats = importance_df["Feature"].head(15).tolist()
display(compare_pair("Other_Faults", "Bumps", top_feats).round(3))
display(compare_pair("Bumps", "Other_Faults", top_feats).round(3))
```
