```python
# gpt

# ============================================================
# 강판 표면 결함 분류 - 파생변수 확장 통합 모델링
# CatBoost auto_class_weights=None 오류 수정 버전
# ============================================================

import time
import warnings
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import chi2_contingency
from sklearn.base import ClassifierMixin, BaseEstimator, clone
from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    RepeatedStratifiedKFold,
    RandomizedSearchCV,
    cross_val_score,
    cross_val_predict,
)
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.inspection import permutation_importance
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
)
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier

try:
    from IPython.display import display
except ImportError:
    display = print

warnings.filterwarnings("ignore")
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)
plt.rcParams["font.family"] = "Malgun Gothic"  # macOS: AppleGothic
plt.rcParams["axes.unicode_minus"] = False

RANDOM_STATE = 42
DATA_PATH = "steel_plates_faults.csv"
N_ITER = 20  # 모델별 탐색 조합 수. 테스트가 오래 걸리면 10으로 줄이기

STEEL_COLS = ["TypeOfSteel_A300", "TypeOfSteel_A400"]
FAULT_COLS = [
    "Pastry", "Z_Scratch", "K_Scatch", "Stains",
    "Dirtiness", "Bumps", "Other_Faults"
]
TARGET = "Fault_Type"
class_to_int = {name: i for i, name in enumerate(FAULT_COLS)}
int_to_class = {i: name for name, i in class_to_int.items()}
STEEL_CODE = {"A300": 0, "A400": 1}
LABELS = list(range(len(FAULT_COLS)))


def macro_f1(y_true, y_pred):
    return f1_score(
        y_true, y_pred, labels=LABELS,
        average="macro", zero_division=0
    )


def score_row(y_true, y_pred):
    return {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Balanced Acc": balanced_accuracy_score(y_true, y_pred),
        "Macro Precision": precision_score(
            y_true, y_pred, labels=LABELS,
            average="macro", zero_division=0
        ),
        "Macro Recall": recall_score(
            y_true, y_pred, labels=LABELS,
            average="macro", zero_division=0
        ),
        "Macro F1": macro_f1(y_true, y_pred),
    }


# ============================================================
# 1. 데이터 불러오기 및 원-핫 구조 검증
# ============================================================
raw_df = pd.read_csv(DATA_PATH)
raw_df = raw_df.loc[:, ~raw_df.columns.astype(str).str.startswith("Unnamed:")].copy()

print("원본 데이터 크기:", raw_df.shape)
print("전체 결측치:", raw_df.isna().sum().sum())
print("중복 행:", raw_df.duplicated().sum())

for cols, label in [(STEEL_COLS, "강종"), (FAULT_COLS, "결함 유형")]:
    missing = [c for c in cols if c not in raw_df.columns]
    if missing:
        raise ValueError(f"{label} 열이 없습니다: {missing}")
    if not raw_df[cols].isin([0, 1]).all().all():
        raise ValueError(f"{label} 열은 0 또는 1이어야 합니다.")
    if not (raw_df[cols].sum(axis=1) == 1).all():
        raise ValueError(f"{label} 원-핫 열의 합이 1이 아닌 행이 있습니다.")


def merge_columns(data):
    """강종 원-핫 2열을 Steel_Type으로, 결함 원-핫 7열을 Fault_Type으로 변환."""
    out = data.copy()
    if set(STEEL_COLS).issubset(out.columns):
        pos = out.columns.get_loc("TypeOfSteel_A300")
        steel = np.where(out["TypeOfSteel_A300"].eq(1), "A300", "A400")
        out = out.drop(columns=STEEL_COLS)
        out.insert(pos, "Steel_Type", steel)
    if set(FAULT_COLS).issubset(out.columns):
        out[TARGET] = out[FAULT_COLS].idxmax(axis=1)
        out = out.drop(columns=FAULT_COLS)
    return out


df = merge_columns(raw_df)
print("병합 후 데이터 크기:", df.shape)
print("\n[결함 유형 분포]")
display(df[TARGET].value_counts().reindex(FAULT_COLS).to_frame("count"))


# ============================================================
# 2. 강종-결함 유형 관계: 카이제곱 검정 및 Cramer's V
# ============================================================
ct = pd.crosstab(df[TARGET], df["Steel_Type"]).reindex(FAULT_COLS, fill_value=0)
chi2, p_value, dof, _ = chi2_contingency(ct)
n = ct.to_numpy().sum()
denominator = n * min(ct.shape[0] - 1, ct.shape[1] - 1)
cramers_v = np.sqrt(chi2 / denominator) if denominator > 0 else np.nan

print("\n[결함 유형 × 강종]")
display(ct)
print(
    f"Chi2 = {chi2:.2f}, df = {dof}, "
    f"p-value = {p_value:.6g}, Cramér's V = {cramers_v:.3f}"
)


# ============================================================
# 3. 파생변수 생성
# ============================================================
# 좌표 가장자리 변수는 데이터에서 관측된 좌표 범위 기준이다.
# 실제 강판의 물리적 경계와 같다고 보장할 수 없으므로 해석에 주의한다.
# ============================================================

BOUNDS = (
    df["X_Minimum"].min(), df["X_Maximum"].max(),
    df["Y_Minimum"].min(), df["Y_Maximum"].max()
)


def build_features(data, use_extra=True, bounds=None):
    out = data.drop(columns=[TARGET] + FAULT_COLS, errors="ignore").copy()

    if "Steel_Type" in out.columns:
        out["Steel_A300"] = out["Steel_Type"].map(STEEL_CODE)
        if out["Steel_A300"].isna().any():
            raise ValueError("Steel_Type에 A300/A400 이외의 값이 있습니다.")
        out = out.drop(columns=["Steel_Type"])
    elif set(STEEL_COLS).issubset(out.columns):
        out["Steel_A300"] = out["TypeOfSteel_A300"]
        out = out.drop(columns=STEEL_COLS)
    else:
        raise ValueError("강종 정보가 없습니다.")

    if not use_extra:
        return out

    required = [
        "X_Minimum", "X_Maximum", "Y_Minimum", "Y_Maximum",
        "Pixels_Areas", "X_Perimeter", "Y_Perimeter",
        "Minimum_of_Luminosity", "Maximum_of_Luminosity",
        "Steel_Plate_Thickness"
    ]
    missing = [c for c in required if c not in out.columns]
    if missing:
        raise ValueError(f"파생변수 생성에 필요한 열이 없습니다: {missing}")

    # 크기
    x_len = (out["X_Maximum"] - out["X_Minimum"] + 1).clip(lower=1)
    y_len = (out["Y_Maximum"] - out["Y_Minimum"] + 1).clip(lower=1)
    area = out["Pixels_Areas"].clip(lower=0)
    perimeter = (out["X_Perimeter"] + out["Y_Perimeter"]).clip(lower=1)

    out["X_Range"] = x_len
    out["Y_Range"] = y_len
    out["Defect_Area_Log"] = np.log1p(area)
    out["Area_Per_BBox"] = area / (x_len * y_len).clip(lower=1)

    # 밝기
    lum_min = out["Minimum_of_Luminosity"]
    lum_max = out["Maximum_of_Luminosity"]
    out["Luminosity_Range"] = lum_max - lum_min
    out["Luminosity_Midpoint"] = (lum_max + lum_min) / 2
    out["Luminosity_Range_Log"] = np.log1p(
        out["Luminosity_Range"].clip(lower=0)
    )

    # 모양
    out["Aspect_Ratio"] = x_len / y_len
    out["Aspect_Ratio_Log"] = np.log(out["Aspect_Ratio"].clip(lower=1e-6))
    out["BBox_Aspect_Ratio"] = (
        np.maximum(x_len, y_len) / np.minimum(x_len, y_len).clip(lower=1)
    )
    out["Perimeter_Area_Ratio"] = perimeter / area.clip(lower=1)
    out["Compactness"] = area / (perimeter ** 2).clip(lower=1)

    # 위치
    out["Center_X"] = (out["X_Minimum"] + out["X_Maximum"]) / 2
    out["Center_Y"] = (out["Y_Minimum"] + out["Y_Maximum"]) / 2
    if bounds is None:
        bounds = BOUNDS
    xmin, xmax, ymin, ymax = bounds
    out["Edge_Distance_X"] = np.minimum(
        (out["X_Minimum"] - xmin).clip(lower=0),
        (xmax - out["X_Maximum"]).clip(lower=0)
    )
    out["Edge_Distance_Y"] = np.minimum(
        (out["Y_Minimum"] - ymin).clip(lower=0),
        (ymax - out["Y_Maximum"]).clip(lower=0)
    )
    out["Edge_Distance_Min"] = np.minimum(
        out["Edge_Distance_X"], out["Edge_Distance_Y"]
    )

    # 강종 및 두께
    thickness = out["Steel_Plate_Thickness"]
    out["Thickness_Log"] = np.log1p(thickness.clip(lower=0))
    out["Steel_Thickness_Interaction"] = out["Steel_A300"] * thickness
    out["Thickness_Group"] = pd.cut(
        thickness,
        bins=[-np.inf, 40, 60, 80, 150, 300, np.inf],
        labels=False,
        right=True
    ).astype(float)

    return out


EXTRA_FEATURES = [
    "X_Range", "Y_Range", "Defect_Area_Log", "Area_Per_BBox",
    "Luminosity_Range", "Luminosity_Midpoint", "Luminosity_Range_Log",
    "Aspect_Ratio", "Aspect_Ratio_Log", "BBox_Aspect_Ratio",
    "Perimeter_Area_Ratio", "Compactness", "Center_X", "Center_Y",
    "Edge_Distance_X", "Edge_Distance_Y", "Edge_Distance_Min",
    "Thickness_Log", "Steel_Thickness_Interaction", "Thickness_Group"
]

y_all = df[TARGET].map(class_to_int).to_numpy()


# ============================================================
# 4. 개발 데이터 / 테스트 데이터 분할
# ============================================================
idx_dev, idx_test = train_test_split(
    np.arange(len(df)), test_size=0.20,
    stratify=y_all, random_state=RANDOM_STATE
)
y_dev, y_test = y_all[idx_dev], y_all[idx_test]
print(f"\n개발 데이터: {len(idx_dev)}개 / 테스트 데이터: {len(idx_test)}개")

cv5 = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
cv_rep = RepeatedStratifiedKFold(
    n_splits=5, n_repeats=3, random_state=RANDOM_STATE
)


# ============================================================
# 5. 파생변수 사용 여부 비교
# ============================================================
probe = RandomForestClassifier(
    n_estimators=300, random_state=RANDOM_STATE, n_jobs=1
)
feature_check = {}

for use_extra in [False, True]:
    X_tmp = build_features(df, use_extra=use_extra).iloc[idx_dev]
    scores = cross_val_score(
        probe, X_tmp, y_dev, cv=cv_rep,
        scoring="f1_macro", n_jobs=-1
    )
    key = "원본 + 파생변수" if use_extra else "원본만"
    feature_check[key] = {"mean": scores.mean(), "std": scores.std()}
    print(f"{key}: CV Macro F1 = {scores.mean():.4f} ± {scores.std():.4f}")

gain = feature_check["원본 + 파생변수"]["mean"] - feature_check["원본만"]["mean"]
# 0.002는 실험용 실질 개선 기준이며, 보편적인 통계적 유의성 기준은 아니다.
USE_EXTRA = gain > 0.002
print(f"파생변수 추가에 따른 Macro F1 차이: {gain:+.4f}")
print("최종 파생변수 사용:", USE_EXTRA)

X_all = build_features(df, use_extra=USE_EXTRA)
feature_cols = X_all.columns.tolist()
X_dev, X_test = X_all.iloc[idx_dev], X_all.iloc[idx_test]


# ============================================================
# 6. 모델 및 탐색 공간
# ============================================================
models = {
    "Random Forest": RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=1),
    "Extra Trees": ExtraTreesClassifier(random_state=RANDOM_STATE, n_jobs=1),
    "LightGBM": LGBMClassifier(
        objective="multiclass", random_state=RANDOM_STATE,
        n_jobs=1, verbosity=-1
    ),
    "CatBoost": CatBoostClassifier(
        loss_function="MultiClass", random_seed=RANDOM_STATE,
        verbose=0, allow_writing_files=False, thread_count=1
    ),
}

param_spaces = {
    "Random Forest": {
        "n_estimators": [300, 500, 700],
        "max_depth": [None, 10, 15, 20, 30],
        "min_samples_split": [2, 5, 9],
        "min_samples_leaf": [1, 2, 4],
        "max_features": ["sqrt", 0.3, 0.5, 0.8],
        "class_weight": [None, "balanced", "balanced_subsample"],
    },
    "Extra Trees": {
        "n_estimators": [300, 500, 700],
        "max_depth": [None, 10, 15, 20, 30],
        "min_samples_split": [2, 5, 9],
        "min_samples_leaf": [1, 2, 4],
        "max_features": ["sqrt", 0.3, 0.5, 0.8],
        "class_weight": [None, "balanced", "balanced_subsample"],
    },
    "LightGBM": {
        "n_estimators": [200, 400, 700, 1000],
        "learning_rate": [0.02, 0.03, 0.05, 0.1],
        "num_leaves": [7, 15, 22, 31],
        "max_depth": [-1, 5, 7, 10],
        "min_child_samples": [5, 10, 20, 33],
        "subsample": [0.6, 0.8, 1.0],
        "subsample_freq": [1],
        "colsample_bytree": [0.6, 0.8, 1.0],
        "reg_lambda": [0, 0.01, 1, 5],
        "class_weight": [None, "balanced"],
    },
    "CatBoost": {
        "iterations": [300, 500, 700],
        "depth": [4, 5, 6, 7],
        "learning_rate": [0.03, 0.05, 0.1],
        "l2_leaf_reg": [1, 3, 5, 7],
        # 수정: None을 제거해 auto_class_weights=null 오류 방지
        "auto_class_weights": ["Balanced", "SqrtBalanced"],
    },
}


# ============================================================
# 7. 하이퍼파라미터 튜닝: 개발 데이터 5-Fold
# ============================================================
tuned = {}
tuning_rows = []

for name, model in models.items():
    print(f"\n{name} 튜닝 중... ({N_ITER}개 조합 × 5-Fold)")
    start = time.time()
    search = RandomizedSearchCV(
        estimator=model,
        param_distributions=param_spaces[name],
        n_iter=N_ITER,
        scoring="f1_macro",
        cv=cv5,
        random_state=RANDOM_STATE,
        n_jobs=-1,
        refit=False,
        error_score="raise",
    )
    search.fit(X_dev, y_dev)
    tuned[name] = clone(model).set_params(**search.best_params_)
    tuning_rows.append({
        "Model": name,
        "CV Macro F1": search.best_score_,
        "CV SD": search.cv_results_["std_test_score"][search.best_index_],
        "Best Params": search.best_params_,
        "Time(s)": round(time.time() - start, 1),
    })
    print(f"CV Macro F1 = {search.best_score_:.4f} | {search.best_params_}")

tuning_df = pd.DataFrame(tuning_rows).sort_values(
    "CV Macro F1", ascending=False
).reset_index(drop=True)
print("\n[모델별 튜닝 결과]")
display(tuning_df)


# ============================================================
# 8. OOF 확률 예측
# ============================================================
oof = {}
for name, est in tuned.items():
    print(f"{name} OOF 예측 중...")
    oof[name] = cross_val_predict(
        clone(est), X_dev, y_dev,
        cv=cv5, method="predict_proba", n_jobs=-1
    )


# ============================================================
# 9. 앙상블 후보 및 클래스별 확률 배율 탐색
# ============================================================
def tune_class_scale(
    P, y,
    grid=np.round(np.arange(0.5, 3.01, 0.1), 2),
    n_rounds=3
):
    """OOF 확률의 클래스별 배율을 곱해 Macro F1을 탐색한다."""
    w = np.ones(P.shape[1])
    best = macro_f1(y, P.argmax(axis=1))
    for _ in range(n_rounds):
        improved = False
        for k in range(P.shape[1]):
            for g in grid:
                trial = w.copy()
                trial[k] = g
                score = macro_f1(y, (P * trial).argmax(axis=1))
                if score > best + 1e-9:
                    best, w, improved = score, trial, True
        if not improved:
            break
    return w, best


ranked = tuning_df["Model"].tolist()
candidates = {name: [name] for name in ranked}
candidates["앙상블(상위 2)"] = ranked[:2]
candidates["앙상블(상위 3)"] = ranked[:3]
candidates["앙상블(전체 4)"] = ranked

cand_rows = []
cand_scale = {}
for cname, members in candidates.items():
    P = np.mean([oof[m] for m in members], axis=0)
    raw_score = macro_f1(y_dev, P.argmax(axis=1))
    scale, scaled_score = tune_class_scale(P, y_dev)
    cand_scale[cname] = scale
    cand_rows.append({
        "Candidate": cname,
        "OOF Macro F1": raw_score,
        "OOF Macro F1 (보정)": scaled_score,
        "Improvement": scaled_score - raw_score,
    })

cand_df = pd.DataFrame(cand_rows).sort_values(
    "OOF Macro F1 (보정)", ascending=False
).reset_index(drop=True)
print("\n[앙상블 및 확률 보정 비교]")
display(cand_df.round(4))


# ============================================================
# 10. 최종 후보 선택 및 테스트 평가
# ============================================================
class ScaledSoftVoting(ClassifierMixin, BaseEstimator):
    """여러 분류기의 확률 평균과 선택적 클래스별 배율 보정."""
    def __init__(self, estimators, class_scale=None):
        self.estimators = estimators
        self.class_scale = class_scale

    def fit(self, X, y):
        self.estimators_ = [clone(est).fit(X, y) for est in self.estimators]
        self.classes_ = np.unique(y)
        return self

    def predict_proba(self, X):
        P = np.mean([est.predict_proba(X) for est in self.estimators_], axis=0)
        if self.class_scale is not None:
            P = P * np.asarray(self.class_scale)
            P = P / P.sum(axis=1, keepdims=True)
        return P

    def predict(self, X):
        return self.classes_[self.predict_proba(X).argmax(axis=1)]


best_row = cand_df.iloc[0]
best_cand = best_row["Candidate"]
use_scale = best_row["Improvement"] >= 0.005
final_scale = cand_scale[best_cand] if use_scale else None

print("\n최종 후보:", best_cand)
print("구성 모델:", candidates[best_cand])
print("클래스별 확률 보정:", use_scale)
if use_scale:
    display(pd.Series(final_scale, index=FAULT_COLS, name="클래스별 배율").round(2))

final_model = ScaledSoftVoting(
    [tuned[m] for m in candidates[best_cand]],
    class_scale=final_scale
).fit(X_dev, y_dev)

y_pred = final_model.predict(X_test)
test_scores = score_row(y_test, y_pred)
print("\n[최종 테스트 성능]")
display(pd.DataFrame([test_scores]).round(4))
print("\n[Classification Report - Test]")
print(classification_report(
    y_test, y_pred, labels=LABELS,
    target_names=FAULT_COLS, digits=4, zero_division=0
))


# ============================================================
# 11. 테스트 Macro F1 부트스트랩 95% 신뢰구간
# ============================================================
rng = np.random.default_rng(RANDOM_STATE)
boot = []
for _ in range(1000):
    b = rng.integers(0, len(y_test), len(y_test))
    boot.append(macro_f1(y_test[b], y_pred[b]))
lo, hi = np.percentile(boot, [2.5, 97.5])
print(f"테스트 Macro F1 부트스트랩 95% 구간: [{lo:.3f}, {hi:.3f}]")


# ============================================================
# 12. 혼동행렬 및 오분류 분석
# ============================================================
fig, axes = plt.subplots(1, 2, figsize=(18, 7))
ConfusionMatrixDisplay(
    confusion_matrix(y_test, y_pred, labels=LABELS),
    display_labels=FAULT_COLS
).plot(ax=axes[0], xticks_rotation=45, colorbar=False)
axes[0].set_title("혼동행렬 (개수)")
ConfusionMatrixDisplay(
    confusion_matrix(y_test, y_pred, labels=LABELS, normalize="true"),
    display_labels=FAULT_COLS
).plot(ax=axes[1], xticks_rotation=45, colorbar=False, values_format=".2f")
axes[1].set_title("혼동행렬 (실제 유형 기준 비율)")
for ax in axes:
    ax.set_xlabel("예측 결함 유형")
    ax.set_ylabel("실제 결함 유형")
plt.tight_layout()
plt.show()

test_result = pd.DataFrame({
    "Actual": [int_to_class[i] for i in y_test],
    "Predicted": [int_to_class[i] for i in y_pred],
    "Confidence": final_model.predict_proba(X_test).max(axis=1),
}, index=X_test.index)
misclassified = test_result[test_result["Actual"] != test_result["Predicted"]]
print(f"\n오분류: {len(misclassified)} / {len(y_test)} ({len(misclassified) / len(y_test):.2%})")
display(
    misclassified.groupby(["Actual", "Predicted"]).size()
    .sort_values(ascending=False).head(10).to_frame("Count")
)


# ============================================================
# 13. Permutation Importance
# ============================================================
perm = permutation_importance(
    final_model, X_test, y_test,
    scoring="f1_macro", n_repeats=20,
    random_state=RANDOM_STATE, n_jobs=-1
)
importance_df = pd.DataFrame({
    "Feature": feature_cols,
    "Importance Mean": perm.importances_mean,
    "Importance SD": perm.importances_std,
}).sort_values("Importance Mean", ascending=False).reset_index(drop=True)
print("\n[Permutation Importance 상위 20개]")
display(importance_df.head(20).round(4))
top = importance_df.head(15).iloc[::-1]
plt.figure(figsize=(10, 7))
plt.barh(top["Feature"], top["Importance Mean"], xerr=top["Importance SD"])
plt.xlabel("변수를 섞었을 때 Macro F1 감소량")
plt.title("Permutation Importance 상위 15개")
plt.tight_layout()
plt.show()


# ============================================================
# 14. 전체 데이터 재학습 및 모델 저장
# ============================================================
deploy_model = ScaledSoftVoting(
    [tuned[m] for m in candidates[best_cand]],
    class_scale=final_scale
).fit(X_all, y_all)

model_bundle = {
    "model": deploy_model,
    "candidate": best_cand,
    "members": candidates[best_cand],
    "params": {m: tuned[m].get_params() for m in candidates[best_cand]},
    "class_scale": None if final_scale is None else final_scale.tolist(),
    "use_extra_features": USE_EXTRA,
    "feature_cols": feature_cols,
    "class_names": FAULT_COLS,
    "bounds": BOUNDS,
    "test_scores": test_scores,
}
joblib.dump(model_bundle, "steel_fault_model_combined.joblib")
print("\n모델 저장 완료: steel_fault_model_combined.joblib")


# ============================================================
# 15. 새 데이터 예측 함수
# ============================================================
def predict_faults(new_data, model_path="steel_fault_model_combined.joblib"):
    bundle = joblib.load(model_path)
    names = bundle["class_names"]
    input_data = new_data.drop(columns=[TARGET] + names, errors="ignore")
    merged = merge_columns(input_data)
    X_new = build_features(
        merged,
        use_extra=bundle["use_extra_features"],
        bounds=bundle["bounds"]
    )
    X_new = X_new.reindex(columns=bundle["feature_cols"])
    if X_new.isna().any().any():
        missing_cols = X_new.columns[X_new.isna().any()].tolist()
        raise ValueError(
            "예측 입력에 결측치가 있거나 필요한 열이 없습니다: "
            f"{missing_cols}"
        )
    proba = bundle["model"].predict_proba(X_new)
    result = pd.DataFrame(
        proba,
        columns=[f"prob_{name}" for name in names],
        index=new_data.index
    )
    result.insert(0, "predicted_fault", [names[i] for i in proba.argmax(axis=1)])
    result.insert(1, "confidence", proba.max(axis=1))
    return result


# 예측 함수 동작 확인: 전체 데이터 재학습 모델의 예측이므로 평가 점수로 사용하지 않는다.
demo = predict_faults(raw_df.iloc[idx_test[:5]])
demo.insert(0, "actual_fault", df.iloc[idx_test[:5]][TARGET].values)
print("\n[예측 함수 동작 확인 - 성능 평가용 결과 아님]")
display(demo.round(3))
```
